<div style="color:#d9cdb4;font-size:12px;letter-spacing:4px;text-transform:uppercase;margin-bottom:14px;font-family:Georgia,serif;">Statistical Research Project</div>
<h1 style="color:#ffffff;font-family:Georgia,serif;font-size:2.15em;margin:0 0 14px 0;font-weight:700;line-height:1.3;">When Does the Hot Hand Appear?</h1>
<p style="color:#e8f0e9;font-size:1.05em;margin:0 0 22px 0;font-family:Georgia,serif;line-height:1.65;max-width:840px;">A statistical investigation of how evidence for short-term dependence in NBA shooting changes under finite-sample bias correction, shot-difficulty adjustment, player heterogeneity, latent-state models, and out-of-sample validation.</p>
<div style="color:#d9cdb4;font-size:13px;font-family:Georgia,serif;border-top:1px solid rgba(255,255,255,0.16);padding-top:14px;margin-top:10px;">2014-15 tracking log, 128,069 attempts &middot; plus a 22-season archive, 4,450,789 attempts &middot; 2003-04 to 2024-25</div>
</div>
</div>

<div style="background:linear-gradient(135deg,#f4efe4 0%,#ece5d6 50%,#f4efe4 100%);border:1.5px solid #d9cdb4;border-radius:16px;padding:34px 40px;margin-bottom:26px;font-family:Georgia,serif;">
<h2 style="color:#14532d;margin:0 0 14px 0;font-size:1.22em;letter-spacing:1px;text-transform:uppercase;">Abstract</h2>
<div style="color:#1a2e22;line-height:1.85;font-size:0.98em;">

<p style="margin:0 0 12px 0;">The basketball hot hand has been argued over for forty years, and for most of
that time the argument has been about the data. This project treats it instead as a question about
estimators. Two datasets are used: shot-level tracking data from the 2014-15 NBA season, which alone
records how closely each shot was defended, and a 22-season archive of 4,450,789 attempts from 2003-04 to
2024-25, which alone carries court coordinates and shot-type labels. The question throughout is how the
evidence for short-term dependence in shooting changes as each analytical decision is made properly
rather than conventionally.</p>

<p style="margin:0 0 12px 0;">Read naively, the data look like a cold hand: shooting after three consecutive
makes is about 2.4 percentage points below a player's own average, and a logistic model with player
effects returns a streak coefficient of -0.051 with a p-value near 1e-14. Almost all of that dissolves
under scrutiny. Roughly a fifth of the raw effect is the finite-sample selection bias that Miller and
Sanjurjo (2018) identified, reproduced here from exact enumeration and confirmed to persist, attenuated,
inside generalised linear models. Two thirds of what remains is shot selection: players attempt shots
about 1.7 feet farther out after three makes than after a miss, a difference with a p-value beyond any
conventional threshold. The fully adjusted, bias-corrected coefficient is about -0.011, which is not
distinguishable from zero under the primary specification. Diagnostics then complicate that: the primary
model is poorly calibrated, and the better-fitting specification that repairs it gives about -0.014 at a
two-sided p-value near 0.039. The adjusted effect is therefore small, negative, and of borderline
significance rather than cleanly null.</p>

<p style="margin:0 0 12px 0;">Three further analyses point the same way. A hierarchical Bayesian model,
fitted by a hand-written Polya-Gamma Gibbs sampler, estimates between-player variation in the streak
effect at 0.023 and finds no player among 281 with a 95% posterior probability of a positive effect. A
two-state hidden Markov model, run to convergence, collapses exactly into a single state. Under
chronological validation, adding streak information to a predictive model changes held-out ROC-AUC by
less than 0.001, against 0.095 for shot difficulty.</p>

<p style="margin:0 0 12px 0;">Across twenty-two seasons, none of this turns out to be a property of one
year. An independent measurement of 2014-15 reproduces the same declining pattern, while revealing that
the tracking log covers only 62% of that season's shots. Fitting all twenty-two seasons separately gives
a negative corrected coefficient in nineteen of them and a pooled estimate near -0.005, with no
detectable between-season heterogeneity (Cochran's Q = 22.1 on 21 df, p = 0.39, I-squared 5%) even though
the share of attempts from beyond the arc rose from 19% to 42% over the period. The archive's zone labels
also show that the pooled coefficient averages over opposite signs: after three makes shooting improves
inside the arc, by 3.1 percentage points in the restricted area, and falls in all three three-point
zones, while the shot mix moves from corner threes toward above-the-break threes.</p>

<p style="margin:0;">Two methodological findings emerged from the checking rather than the analysis, and may
be the more durable contribution. The conventional inverse-gamma prior on the slope variance returns
more between-player heterogeneity from data simulated with none (0.173) than from the real data (0.136),
so any hierarchical hot-hand analysis using it is liable to report player differences that do not exist.
And the hidden Markov model's state separation shrinks slowly enough that a default stopping rule halts
while the two states still differ by 0.13, manufacturing a latent-state finding that longer optimisation
dissolves. Both defaults err toward finding structure that is not there.</p>

</div>
</div>

<div style="background:linear-gradient(135deg,#0d3320 0%,#14532d 55%,#2f7a4d 100%);padding:26px 32px;border-radius:14px;margin:34px 0 20px 0;box-shadow:0 6px 24px rgba(13,51,32,0.25);">
<div style="color:rgba(255,255,255,0.55);font-size:11px;letter-spacing:4px;text-transform:uppercase;margin-bottom:8px;">Section 1</div>
<h2 style="color:#ffffff;font-family:Georgia,serif;font-size:24px;margin:0;font-weight:700;">Introduction</h2>
<p style="color:#d9cdb4;margin:6px 0 0 0;font-size:13.5px;">Why a forty-year-old argument is really an argument about estimators</p>
</div>


Gilovich, Vallone and Tversky (1985) examined shooting records from the Philadelphia 76ers, free-throw
records from the Boston Celtics, and a controlled experiment with Cornell's varsity teams, and found no
evidence that a make made the next make more likely. Their conclusion, that belief in the hot hand is a
cognitive illusion, became one of the most cited results in the psychology of judgement, and Kahneman
later described the hot hand as a massive and widespread cognitive illusion.

Thirty-three years later, Miller and Sanjurjo (2018) proved that the statistic those studies relied on is
biased. Not biased in some exotic edge case, but biased under the simplest possible null: independent
coin flips. Conditioning on positions that follow a streak is a selection step, and the resulting
estimator sits systematically below the true probability. Correcting for it reverses the direction of the
canonical study's conclusions.

That is a remarkable episode in the history of applied statistics, and it is easy to draw the wrong
lesson from it. The wrong lesson is that the hot hand has now been proven real. Miller and Sanjurjo
proved a theorem about an estimator, not a fact about basketball, and the size of any real effect, in any
particular dataset, remains an empirical question. The right lesson is that the answer depended on an
analytical choice nobody had examined, and that invites a harder question: how many of the other choices
matter too?


<div style="background:linear-gradient(135deg,#f3f8f4,#e6f0e9);border:1px solid #7aa98c55;border-left:5px solid #1c6b3f;border-radius:0 12px 12px 0;padding:18px 24px;margin:14px 0;">
<div style="font-weight:700;color:#14532d;margin-bottom:10px;font-size:14px;">What this project actually asks</div>
<div style="color:#1a2e22;font-size:13.5px;line-height:1.75;">
Not "does the hot hand exist", which invites a yes or no that the data cannot support. Instead:
<br><br>
<em>How does the evidence for short-term dependence in shooting change as finite-sample bias, shot
difficulty, player heterogeneity, latent-state structure, model specification and out-of-sample
validation are each taken seriously?</em>
<br><br>
The output is not a verdict but a decomposition: how much of the apparent effect each consideration
explains, and what survives all of them. Along the way the project also asks the question one step
further out, which is when these methods can be trusted at all. Section 18 generates data from five known
processes, one with no hot hand, and checks which methods recover which truth.
</div>
</div>

<a id="sec2"></a>
<div style="background:linear-gradient(135deg,#0d3320 0%,#14532d 55%,#2f7a4d 100%);padding:26px 32px;border-radius:14px;margin:34px 0 20px 0;box-shadow:0 6px 24px rgba(13,51,32,0.25);">
<div style="color:rgba(255,255,255,0.55);font-size:11px;letter-spacing:4px;text-transform:uppercase;margin-bottom:8px;">Section 2</div>
<h2 style="color:#ffffff;font-family:Georgia,serif;font-size:24px;margin:0;font-weight:700;">Research Questions</h2>
<p style="color:#d9cdb4;margin:6px 0 0 0;font-size:13.5px;">Six questions the data can actually answer</p>
</div>


**RQ1.** How large is the streak-selection bias across sequence lengths and streak lengths, and how does
it compare with the effect sizes at issue in the literature?

**RQ2.** How do the naive estimator, a bias-corrected estimator and a permutation-calibrated test compare
on bias, variance, mean squared error and interval coverage under controlled data-generating processes?

**RQ3.** Does the apparent relationship between streaks and shooting survive conditioning on observable
shot difficulty, or is it shot selection?

**RQ4.** Is there genuine between-player heterogeneity in short-term dependence, and how much of any
apparent heterogeneity is produced by the prior on the variance component?

**RQ5.** Do the data prefer an explicit streak mechanism or a latent two-state process?

**RQ6.** Does streak information carry out-of-sample predictive value once player identity and shot
difficulty are already known?

**RQ7.** Does an independent measurement of the same season agree, and is whatever the season shows
stable across twenty-two years of a substantially changing game?


<div style="background:#f4efe4;border-radius:14px;padding:24px 30px;border:1.5px solid #d9cdb4;font-family:Georgia,serif;line-height:1.85;color:#1a2e22;margin-bottom:10px;">
<h4 style="color:#14532d;margin:0 0 10px;font-size:1.0rem;">Hypotheses and what counts as confirmatory</h4>
<p style="margin:0 0 10px 0;"><strong>Primary hypothesis, fixed before the modelling.</strong>
The quantity of interest is the coefficient on the preceding-make count in a logistic model with
player-specific intercepts and shot-difficulty controls. The null is that it equals the value it takes
under independent shooting, which Section 12.1 shows is not zero.</p>
<p style="margin:0 0 10px 0;"><strong>Secondary and exploratory.</strong> Everything concerning the shape of
the streak effect (Section 14), between-player heterogeneity (Section 15), and latent states (Section 16)
is secondary. Where a secondary analysis produces something that looks significant, the notebook reports
how many comparisons were involved and whether the finding survives them, rather than promoting it.</p>
<p style="margin:0;"><strong>Standing rule.</strong> Every coefficient is judged against a simulated or
permutation null rather than against zero, because zero is the wrong benchmark for every statistic in
this notebook.</p>

<div style="background:linear-gradient(135deg,#0d3320 0%,#14532d 55%,#2f7a4d 100%);padding:26px 32px;border-radius:14px;margin:34px 0 20px 0;box-shadow:0 6px 24px rgba(13,51,32,0.25);">
<div style="color:rgba(255,255,255,0.55);font-size:11px;letter-spacing:4px;text-transform:uppercase;margin-bottom:8px;">Section 3</div>
<h2 style="color:#ffffff;font-family:Georgia,serif;font-size:24px;margin:0;font-weight:700;">Statistical Background</h2>
<p style="color:#d9cdb4;margin:6px 0 0 0;font-size:13.5px;">The machinery, stated before it is used rather than after</p>
</div>


Each method below is introduced with what it assumes and what it cannot do, so that the results sections
can report numbers without stopping to explain the apparatus. Readers who want the results first can skip
to Section 5 and refer back.


<div style="background:linear-gradient(135deg,#2f7a4d 0%,#14532d 100%);border-radius:8px;padding:11px 20px;margin:24px 0 14px 0;">
<span style="color:#d9cdb4;font-family:Georgia,serif;font-size:12px;font-weight:700;letter-spacing:1px;">3.1</span>
<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">The conditional estimator and its selection problem</span>
</div>


For a binary sequence $Y_1,\dots,Y_n$ the quantity at issue is the probability of a make given that the
previous $k$ shots were all makes,


$$p_k \;=\; P\big(Y_{t+1}=1 \mid Y_{t-k+1}=\cdots=Y_t=1\big)$$


estimated by averaging outcomes over the positions that follow such a run. Under independence
$p_k = p$ for every $k$, so the natural test compares $\hat p_k - \hat p$ against zero.

The problem is that the positions entering the average are selected by the outcomes themselves. Within a
finite sequence, requiring $k$ preceding makes uses up makes from a limited supply, and the expectation
of the resulting ratio is not the ratio of expectations. Miller and Sanjurjo (2018) prove
$\mathbb{E}[\hat p_k] < p$ under independence, with the gap growing in $k$ and shrinking in $n$. Section
9 derives the smallest case by exact enumeration.


<div style="background:linear-gradient(135deg,#2f7a4d 0%,#14532d 100%);border-radius:8px;padding:11px 20px;margin:24px 0 14px 0;">
<span style="color:#d9cdb4;font-family:Georgia,serif;font-size:12px;font-weight:700;letter-spacing:1px;">3.2</span>
<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">Permutation inference</span>
</div>


If the exact bias is awkward to compute, it can be simulated. Holding a player's outcomes fixed and
reordering them destroys any sequential dependence while preserving how many shots were made, so the
distribution of the statistic across reorderings is its null distribution including whatever bias the
estimator carries. The exchangeability assumption is explicit: conditional on player and game, all
orderings of that game's outcomes are equally likely under the null. Permuting within game rather than
across a season deliberately preserves game-level variation, so a player with genuinely good and bad
nights does not register as streaky.


<div style="background:linear-gradient(135deg,#2f7a4d 0%,#14532d 100%);border-radius:8px;padding:11px 20px;margin:24px 0 14px 0;">
<span style="color:#d9cdb4;font-family:Georgia,serif;font-size:12px;font-weight:700;letter-spacing:1px;">3.3</span>
<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">Logistic regression with group effects</span>
</div>

$$Y_{it} \sim \text{Bernoulli}(p_{it}), \qquad
\operatorname{logit}(p_{it}) = \log\frac{p_{it}}{1-p_{it}} = \alpha_i + \beta K_{it} + \gamma^\top X_{it}$$


$\alpha_i$ absorbs each player's baseline, which matters because better shooters mechanically produce
more streaks and would otherwise contaminate $\beta$. Fitting is by iteratively reweighted least squares,
which solves the weighted normal equations


$$\beta^{(m+1)} = \big(X^\top W^{(m)} X\big)^{-1} X^\top W^{(m)} z^{(m)},
\qquad W_{tt} = \hat p_t(1-\hat p_t), \quad z = \eta + \frac{y - \hat p}{W}$$


at each step. With 281 player indicators the design is held sparse so it never has to be formed densely.
The assumptions are the usual ones for a GLM: the link is correct, observations are conditionally
independent given the covariates, and the covariates are measured without error. The second is the
uncomfortable one here, since shots within a possession are plainly not independent, and it is a reason
to treat model-based standard errors as approximate.


<div style="background:linear-gradient(135deg,#2f7a4d 0%,#14532d 100%);border-radius:8px;padding:11px 20px;margin:24px 0 14px 0;">
<span style="color:#d9cdb4;font-family:Georgia,serif;font-size:12px;font-weight:700;letter-spacing:1px;">3.4</span>
<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">Hierarchical models and partial pooling</span>
</div>

$$\alpha_i \sim N(\mu_\alpha, \sigma_\alpha^2), \qquad
\beta_i \sim N(\mu_\beta, \sigma_\beta^2)$$


Rather than estimating each player's slope in isolation or forcing them all to share one, a hierarchical
model treats them as draws from a population. Each player's estimate is then pulled toward the population
mean by an amount that depends on how much information that player carries, which is what makes the
approach right for data where some players have 47 shots and others 1,054. $\sigma_\beta$ is the
quantity of substantive interest: it says how much players genuinely differ, as opposed to how much
their noisy individual estimates differ.


<div style="background:linear-gradient(135deg,#2f7a4d 0%,#14532d 100%);border-radius:8px;padding:11px 20px;margin:24px 0 14px 0;">
<span style="color:#d9cdb4;font-family:Georgia,serif;font-size:12px;font-weight:700;letter-spacing:1px;">3.5</span>
<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">Bayesian computation via Polya-Gamma augmentation</span>
</div>


Logistic likelihoods are not conjugate with normal priors, which is why Bayesian logistic models usually
need gradient-based sampling. Polson, Scott and Windle (2013) supply an exact alternative. Writing
$\eta$ for the linear predictor,


$$\frac{(e^{\eta})^{a}}{(1+e^{\eta})^{b}}
= 2^{-b} e^{\kappa \eta} \int_0^\infty e^{-\omega \eta^{2}/2}\, p(\omega)\, \mathrm{d}\omega,
\qquad \omega \sim PG(b,0), \quad \kappa = a - \tfrac{b}{2}$$


so conditional on $\omega$ the likelihood is Gaussian in $\eta$ and every coefficient block has a
closed-form normal full conditional. The sampler is then pure Gibbs with no tuning. Section 15 explains
the truncation used to draw $PG(1,c)$ and the analytic correction applied to it.


<div style="background:linear-gradient(135deg,#2f7a4d 0%,#14532d 100%);border-radius:8px;padding:11px 20px;margin:24px 0 14px 0;">
<span style="color:#d9cdb4;font-family:Georgia,serif;font-size:12px;font-weight:700;letter-spacing:1px;">3.6</span>
<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">Priors on variance components</span>
</div>


An inverse-gamma prior on a variance parameter is conventional and conjugate, but Gelman (2006) shows it
is not close to non-informative when the true variance is near zero: its density vanishes at the origin,
so it pushes the posterior away from the boundary. The alternative used here is a half-Cauchy on the
standard deviation, which does not, and which admits a conjugate representation as a scale mixture,


$$\sigma^2 \mid \xi \sim \text{InvGamma}\!\left(\tfrac12, \tfrac1\xi\right),
\qquad \xi \sim \text{InvGamma}\!\left(\tfrac12, \tfrac1{A^2}\right)
\;\Longrightarrow\; \sigma \sim \text{Half-Cauchy}(0, A)$$


keeping the sampler fully Gibbs. Section 15.2 measures how much this choice matters here, and the answer
turns out to be a great deal.


<div style="background:linear-gradient(135deg,#2f7a4d 0%,#14532d 100%);border-radius:8px;padding:11px 20px;margin:24px 0 14px 0;">
<span style="color:#d9cdb4;font-family:Georgia,serif;font-size:12px;font-weight:700;letter-spacing:1px;">3.7</span>
<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">Hidden Markov models</span>
</div>

$$Z_t \in \{0,1\}, \qquad P(Z_t = j \mid Z_{t-1}=i) = A_{ij},
\qquad Y_t \mid Z_t = j \sim \text{Bernoulli}(p_j)$$


Fitted by Baum-Welch, an EM algorithm whose E-step is the forward-backward recursion and whose M-step
reweights the sufficient statistics by the state posteriors. EM increases the likelihood at every
iteration but can converge slowly, which Section 16 turns out to depend on. The two-state model contains
the constant-probability model as the special case $p_0 = p_1$, so a likelihood-ratio comparison between
them sits on a parameter-space boundary and the usual chi-squared reference distribution does not apply.


<div style="background:linear-gradient(135deg,#2f7a4d 0%,#14532d 100%);border-radius:8px;padding:11px 20px;margin:24px 0 14px 0;">
<span style="color:#d9cdb4;font-family:Georgia,serif;font-size:12px;font-weight:700;letter-spacing:1px;">3.8</span>
<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">Confidence intervals, credible intervals, and simulated nulls</span>
</div>


Three different kinds of uncertainty statement appear in this notebook and they mean different things. A
confidence interval is a statement about a procedure: across hypothetical repeated samples, 95% of
intervals built this way would contain the parameter. A credible interval is a statement about the
parameter given the data actually observed and the prior. A simulated null is neither: it is the
distribution the estimator takes when the effect is absent by construction, and it is what a coefficient
must be compared against when the estimator is biased, since the usual reference point of zero is then
simply wrong.


<div style="background:linear-gradient(135deg,#2f7a4d 0%,#14532d 100%);border-radius:8px;padding:11px 20px;margin:24px 0 14px 0;">
<span style="color:#d9cdb4;font-family:Georgia,serif;font-size:12px;font-weight:700;letter-spacing:1px;">3.9</span>
<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">Multiple comparisons</span>
</div>


Testing 248 players separately, or three streak categories, inflates the chance of at least one small
p-value. The Benjamini-Hochberg procedure controls the false discovery rate: order the $m$ p-values and
reject the $i$ largest-ranked one that satisfies


$$p_{(i)} \le \frac{i}{m}\,q$$


which is less severe than Bonferroni and more appropriate when many tests are screened rather than a
single pre-specified hypothesis tested. Both are used in the notebook, with Bonferroni reserved for the
small pre-specified families where it is conceptually right.


<a id="sec5"></a>
<div style="background:linear-gradient(135deg,#0d3320 0%,#14532d 55%,#2f7a4d 100%);padding:26px 32px;border-radius:14px;margin:34px 0 20px 0;box-shadow:0 6px 24px rgba(13,51,32,0.25);">
<div style="color:rgba(255,255,255,0.55);font-size:11px;letter-spacing:4px;text-transform:uppercase;margin-bottom:8px;">Section 5</div>
<h2 style="color:#ffffff;font-family:Georgia,serif;font-size:24px;margin:0;font-weight:700;">Data</h2>
<p style="color:#d9cdb4;margin:6px 0 0 0;font-size:13.5px;">What was collected, by whom, and what each variable actually measures</p>
</div>


The analysis uses shot-level tracking data from the 2014-15 NBA regular season, distributed publicly
on Kaggle as the *NBA shot logs* dataset. Each row is a single field-goal attempt with its outcome, its
spatial and temporal context, and the distance to the nearest defender at the moment of release. Free
throws are not included, which matters: free throws are the setting in which several hot-hand studies
have found their clearest signals, and field goals are a harder and more contested test.

Two properties of this file make it the right choice here rather than a season-aggregate table. First,
shots carry an explicit within-game order, so make/miss sequences can be reconstructed exactly, which is
the entire object of study. Second, it records `CLOSE_DEF_DIST`, the distance to the closest defender,
which is the variable Bocskocsky, Ezekowitz and Stein (2014) needed to ask whether apparent streak
effects are really shot-selection effects. Without a defender-proximity measure, the shot-difficulty
adjustment in Section 13 would be badly underpowered.


```python
DATA_PATH = '/kaggle/input/nba-shot-logs/shot_logs.csv'
raw = pd.read_csv(DATA_PATH)

print(f"Rows (field-goal attempts): {len(raw):,}")
print(f"Columns:                    {raw.shape[1]}")
print(f"Distinct players:           {raw['player_id'].nunique()}")
print(f"Distinct games:             {raw['GAME_ID'].nunique()}")
print(f"Distinct player-games:      {raw.groupby(['player_id','GAME_ID']).ngroups:,}")
print(f"Overall field-goal rate:    {raw['FGM'].mean():.4f}")

shots_per_player = raw.groupby('player_id').size()
print(f"\nShots per player: min={shots_per_player.min()}, "
      f"median={shots_per_player.median():.0f}, max={shots_per_player.max()}")
```

<div style="background:linear-gradient(135deg,#2f7a4d 0%,#14532d 100%);border-radius:8px;padding:11px 20px;margin:24px 0 14px 0;">
<span style="color:#d9cdb4;font-family:Georgia,serif;font-size:12px;font-weight:700;letter-spacing:1px;">5.1</span>
<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">Variable definitions</span>
</div>

```python
variable_definitions = pd.DataFrame([
    ('GAME_ID',        'identifier', 'Unique game identifier'),
    ('MATCHUP',        'identifier', 'Date and teams, e.g. "MAR 04, 2015 - CHA @ BKN"'),
    ('LOCATION',       'context',    'H = home, A = away'),
    ('W',              'context',    'Game outcome for the shooter\'s team (W/L), known only ex post'),
    ('FINAL_MARGIN',   'context',    'Final points margin, known only ex post'),
    ('SHOT_NUMBER',    'sequence',   "Order of this shot within the player's game (1, 2, 3, ...)"),
    ('PERIOD',         'context',    'Quarter (1-4) or overtime (5+)'),
    ('GAME_CLOCK',     'context',    'Time remaining in the period, mm:ss'),
    ('SHOT_CLOCK',     'context',    'Seconds left on the shot clock at release'),
    ('DRIBBLES',       'difficulty', 'Dribbles taken before the shot'),
    ('TOUCH_TIME',     'difficulty', 'Seconds of possession before the shot'),
    ('SHOT_DIST',      'difficulty', 'Distance from basket, feet'),
    ('PTS_TYPE',       'difficulty', 'Shot value, 2 or 3'),
    ('CLOSE_DEF_DIST', 'difficulty', 'Distance to nearest defender at release, feet'),
    ('CLOSEST_DEFENDER','context',   'Name of the nearest defender'),
    ('SHOT_RESULT',    'outcome',    'made / missed'),
    ('FGM',            'outcome',    'Binary outcome, 1 = made'),
    ('PTS',            'outcome',    'Points scored on the attempt'),
    ('player_name',    'identifier', 'Shooter name'),
    ('player_id',      'identifier', 'Shooter identifier'),
], columns=['Variable', 'Role', 'Definition'])
variable_definitions
```

<div style="background:#fdf5ee;border-radius:14px;padding:22px 28px;border:1.5px solid #d9884f;font-family:Georgia,serif;line-height:1.8;color:#6b3410;margin:14px 0;">
<h4 style="color:#b4531f;margin:0 0 10px;font-size:1.0rem;">Two variables that must never enter a predictive model</h4>

<code>W</code> and <code>FINAL_MARGIN</code> describe how the game finished. A model predicting whether a
shot taken in the first quarter goes in cannot be given information about the final score without
leaking the future into the past. Both are retained for descriptive purposes and excluded from every
model in Sections 12 onward, a decision recorded here rather than buried in a feature list.

</div>

<a id="sec6"></a>
<div style="background:linear-gradient(135deg,#0d3320 0%,#14532d 55%,#2f7a4d 100%);padding:26px 32px;border-radius:14px;margin:34px 0 20px 0;box-shadow:0 6px 24px rgba(13,51,32,0.25);">
<div style="color:rgba(255,255,255,0.55);font-size:11px;letter-spacing:4px;text-transform:uppercase;margin-bottom:8px;">Section 6</div>
<h2 style="color:#ffffff;font-family:Georgia,serif;font-size:24px;margin:0;font-weight:700;">Data Quality</h2>
<p style="color:#d9cdb4;margin:6px 0 0 0;font-size:13.5px;">Five real problems in this file, each with a stated rule and a counted effect</p>
</div>


Every check below was run before any model was fitted, and every exclusion rule was written down before
its effect on results was examined. The ordering follows severity: internal consistency first, then
impossible values, then the sequence integrity problem that governs how streaks may legitimately be
constructed.


<div style="background:linear-gradient(135deg,#2f7a4d 0%,#14532d 100%);border-radius:8px;padding:11px 20px;margin:24px 0 14px 0;">
<span style="color:#d9cdb4;font-family:Georgia,serif;font-size:12px;font-weight:700;letter-spacing:1px;">6.1</span>
<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">Internal consistency and duplicates</span>
</div>

```python
consistency = pd.DataFrame([
    ('Exact duplicate rows', int(raw.duplicated().sum())),
    ('FGM disagrees with SHOT_RESULT', int(((raw['FGM'] == 1) != (raw['SHOT_RESULT'] == 'made')).sum())),
    ('PTS disagrees with FGM x PTS_TYPE', int((raw['PTS'] != raw['FGM'] * raw['PTS_TYPE']).sum())),
    ('PTS_TYPE outside {2, 3}', int((~raw['PTS_TYPE'].isin([2, 3])).sum())),
    ('PERIOD below 1', int((raw['PERIOD'] < 1).sum())),
    ('Negative shot distance', int((raw['SHOT_DIST'] < 0).sum())),
    ('Negative defender distance', int((raw['CLOSE_DEF_DIST'] < 0).sum())),
], columns=['Check', 'Violations'])
consistency['Status'] = np.where(consistency['Violations'] == 0, 'pass', 'FLAG')
consistency
```

<div style="background:linear-gradient(135deg,#f3f8f4,#e6f0e9);border:1px solid #7aa98c55;border-left:5px solid #1c6b3f;border-radius:0 12px 12px 0;padding:18px 24px;margin:14px 0;">
<div style="font-weight:700;color:#14532d;margin-bottom:10px;font-size:14px;">The outcome variables are internally coherent</div>
<div style="color:#1a2e22;font-size:13.5px;line-height:1.75;">
Every consistency check passes with zero violations. The two redundant encodings of the outcome
(<code>FGM</code> and <code>SHOT_RESULT</code>) agree on all 128,069 rows, and points reconcile exactly
with outcome times shot value. That redundancy is worth checking rather than assuming, because a
mismatch would signal a corrupted merge somewhere upstream and would invalidate the entire outcome
variable. It does not mean the file is clean, as the next four subsections show, but it does mean the
thing being predicted is trustworthy.
</div>
</div>

<div style="background:linear-gradient(135deg,#2f7a4d 0%,#14532d 100%);border-radius:8px;padding:11px 20px;margin:24px 0 14px 0;">
<span style="color:#d9cdb4;font-family:Georgia,serif;font-size:12px;font-weight:700;letter-spacing:1px;">6.2</span>
<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">Missingness, and why it is not ignorable here</span>
</div>

```python
missing = raw.isna().sum()
missing = missing[missing > 0].sort_values(ascending=False)
missing_tbl = pd.DataFrame({'N missing': missing, '% of rows': (100 * missing / len(raw)).round(2)})
print(missing_tbl.to_string())

# Where does the missingness sit, and does it carry outcome information?
gc_seconds = raw['GAME_CLOCK'].str.split(':').apply(lambda p: int(p[0]) * 60 + int(p[1]))
raw['game_clock_sec'] = gc_seconds
is_miss = raw['SHOT_CLOCK'].isna()

print(f"\nShare of missing rows in the final 10s of a period: "
      f"{((is_miss) & (raw['game_clock_sec'] < 10)).sum() / is_miss.sum():.2f}")
print(f"Mean seconds left | missing: {raw.loc[is_miss,'game_clock_sec'].mean():.1f}"
      f"   | present: {raw.loc[~is_miss,'game_clock_sec'].mean():.1f}")
print(f"Field-goal rate   | missing: {raw.loc[is_miss,'FGM'].mean():.4f}"
      f" | present: {raw.loc[~is_miss,'FGM'].mean():.4f}"
      f" | raw gap: {raw.loc[~is_miss,'FGM'].mean() - raw.loc[is_miss,'FGM'].mean():+.4f}")
```

```python
# Bin by time left in the period so the mechanism is visible rather than implied
bins = [0, 10, 24, 60, 120, 240, 420, 720]
labels = ['0-10', '10-24', '24-60', '60-120', '120-240', '240-420', '420+']
raw['clock_bin'] = pd.cut(raw['game_clock_sec'], bins=bins, labels=labels, right=False)

by_bin = raw.groupby('clock_bin', observed=True).agg(
    n=('FGM', 'size'),
    share_missing=('SHOT_CLOCK', lambda s: s.isna().mean()),
).reset_index()
fg_by_bin = (raw.assign(_m=is_miss)
                .groupby(['clock_bin', '_m'], observed=True)['FGM'].mean().unstack())

fig, axes = plt.subplots(1, 3, figsize=(17, 4.9))

# (a) the mechanism: share missing rises sharply as the period runs out
ax = axes[0]
bars = ax.bar(by_bin['clock_bin'].astype(str), by_bin['share_missing'],
              color=[COPPER if s > 0.2 else GREEN_SOFT for s in by_bin['share_missing']],
              edgecolor=GREEN_DEEP)
for b, v, n in zip(bars, by_bin['share_missing'], by_bin['n']):
    ax.text(b.get_x() + b.get_width()/2, v + 0.015, f'{v:.2f}', ha='center',
            fontsize=9.5, color=GREEN_DEEP, fontweight='bold')
ax.set_title('(a) Share of shots with SHOT_CLOCK missing')
ax.set_xlabel('Seconds left in period'); ax.set_ylabel('Proportion missing')
ax.tick_params(axis='x', rotation=30)

# (b) the outcome signal, held at fixed game time
ax = axes[1]
x = np.arange(len(fg_by_bin))
ax.plot(x, fg_by_bin[False], marker='o', color=GREEN, lw=2, label='SHOT_CLOCK present')
ax.plot(x, fg_by_bin[True], marker='s', color=COPPER, lw=2, label='SHOT_CLOCK missing')
ax.set_xticks(x); ax.set_xticklabels(fg_by_bin.index.astype(str), rotation=30)
ax.set_title('(b) Field-goal rate within the same clock window')
ax.set_xlabel('Seconds left in period'); ax.set_ylabel('Field-goal rate')
ax.legend(frameon=False, fontsize=9)

# (c) the unconditional gap
ax = axes[2]
rates = [raw.loc[~is_miss, 'FGM'].mean(), raw.loc[is_miss, 'FGM'].mean()]
counts = [(~is_miss).sum(), is_miss.sum()]
bars = ax.bar(['present', 'missing'], rates, color=[GREEN, COPPER],
              edgecolor=GREEN_DEEP, width=0.55)
for b, r, c in zip(bars, rates, counts):
    ax.text(b.get_x() + b.get_width()/2, r + 0.01, f'{r:.3f}\nn={c:,}',
            ha='center', fontsize=10, color=GREEN_DEEP, fontweight='bold')
ax.set_ylim(0, 0.56)
ax.set_title('(c) Unconditional field-goal rate')
ax.set_ylabel('Field-goal rate'); ax.set_xlabel('SHOT_CLOCK')

fig.suptitle('Figure 1. Shot-clock missingness is structured, not random',
             fontsize=14, color=GREEN_DEEP, fontweight='bold', y=1.05)
plt.tight_layout(); plt.show()
```


Figure 1(b) is the panel worth pausing on. Once shots are compared within the same slice of game time,
the field-goal gap between missing and observed largely closes, and in two windows it even reverses. That
suggests the raw gap is mostly game time in disguise rather than information carried by the missingness
itself. The next cell tests that directly, and the answer turns out to depend entirely on how flexibly
game time is allowed to enter the model.


```python
import statsmodels.api as sm

d = raw.copy()
d['miss'] = is_miss.astype(int)

# Specification A: game clock controlled smoothly, as a quadratic
XA = sm.add_constant(pd.DataFrame({'miss': d['miss'], 'gc': d['game_clock_sec'],
                                    'gc2': d['game_clock_sec'] ** 2}))
mA = sm.Logit(d['FGM'], XA).fit(disp=False)

# Specification B: game clock controlled flexibly, as bin indicators
edges = [0, 5, 10, 15, 24, 40, 60, 120, 240, 420, 720]
d['gc_bin'] = pd.cut(d['game_clock_sec'], bins=edges, right=False)
XB = sm.add_constant(pd.concat([d[['miss']].astype(float),
                                 pd.get_dummies(d['gc_bin'], drop_first=True, dtype=float)], axis=1))
mB = sm.Logit(d['FGM'], XB).fit(disp=False)

mar_table = pd.DataFrame({
    'Control for game time': ['none (unconditional)', 'quadratic in seconds left',
                               'flexible bin indicators'],
    'coef(missing)': [np.log((raw.loc[is_miss,'FGM'].mean()/(1-raw.loc[is_miss,'FGM'].mean())) /
                             (raw.loc[~is_miss,'FGM'].mean()/(1-raw.loc[~is_miss,'FGM'].mean()))),
                      mA.params['miss'], mB.params['miss']],
    'p-value': [np.nan, mA.pvalues['miss'], mB.pvalues['miss']],
})
mar_table['odds ratio'] = np.exp(mar_table['coef(missing)'])
mar_table.round(4)
```

<div style="background:#f3f8f4;border-radius:12px;padding:20px 26px;border-left:5px solid #1c6b3f;font-family:Georgia,serif;line-height:1.85;color:#1a2e22;margin:14px 0;">
<h4 style="color:#14532d;margin:0 0 10px;">Result: the missingness looks informative only when game time is modelled badly</h4>

Controlling for seconds remaining with a quadratic, the missingness indicator carries a large and
overwhelmingly significant negative coefficient (-0.325, p &asymp; 4e-27, odds ratio 0.72). Replacing
that quadratic with flexible bin indicators, and changing nothing else, collapses the same coefficient to
-0.014 with p = 0.72. The apparent signal was never about missingness. It was a smooth function failing
to bend sharply enough at the end of a period, with the missingness indicator absorbing the residual.
<br><br>
The mechanism is visible in the counts: 55% of all missing rows sit in the final ten seconds of a period,
a window where the field-goal rate collapses to roughly 0.30-0.35 for everyone. A quadratic over a
twelve-minute period cannot reproduce a cliff that sharp, so whatever indicator happens to be correlated
with that window inherits the misfit.
<br><br>
So shot-clock missingness in this file is close to missing-at-random given game time, not
missing-not-at-random. That is the opposite of the conclusion the unconditional comparison invites, and
it is a useful warning to carry into the rest of the notebook: several later sections control for game
context while testing a streak variable, and this section is a concrete demonstration that the functional
form chosen for a control can manufacture a highly significant effect for something unrelated.

</div>

<div style="background:#f4efe4;border-radius:12px;padding:20px 26px;border-left:5px solid #b4531f;font-family:Georgia,serif;line-height:1.85;color:#1a2e22;margin:14px 0;">
<h4 style="color:#b4531f;margin:0 0 10px;">Decision: indicator retained, but for a different reason than expected</h4>

The missingness indicator stays in the analytical dataset, alongside a neutral fill for the shot clock
value itself. The justification is no longer that the missingness is informative about the outcome, since
the test above shows it is not once game time is handled properly. It is that carrying an explicit flag
is cheap, keeps the 5,567 affected rows in the sample rather than discarding 4.35% of the data, and makes
the assumption visible to a reader instead of hidden inside an imputation.
<br><br>
Median imputation without a flag was rejected: it would assign an end-of-period heave a typical
mid-possession shot clock reading, which is not a plausible value for that situation. Section 23 repeats
the primary analysis with these rows dropped entirely, so the choice remains testable.

</div>

<div style="background:linear-gradient(135deg,#2f7a4d 0%,#14532d 100%);border-radius:8px;padding:11px 20px;margin:24px 0 14px 0;">
<span style="color:#d9cdb4;font-family:Georgia,serif;font-size:12px;font-weight:700;letter-spacing:1px;">6.3</span>
<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">Impossible values</span>
</div>

```python
print("TOUCH_TIME range:", raw['TOUCH_TIME'].min(), "to", raw['TOUCH_TIME'].max())
n_neg_touch = int((raw['TOUCH_TIME'] < 0).sum())
print(f"Negative TOUCH_TIME (physically impossible): {n_neg_touch} "
      f"({100*n_neg_touch/len(raw):.3f}% of rows)")
print("Most extreme negative values:",
      sorted(raw.loc[raw['TOUCH_TIME'] < 0, 'TOUCH_TIME'].unique())[:5])

print("\nSHOT_DIST range:", raw['SHOT_DIST'].min(), "to", raw['SHOT_DIST'].max(),
      "| beyond 47ft (past half court):", int((raw['SHOT_DIST'] > 47).sum()))
print("CLOSE_DEF_DIST range:", raw['CLOSE_DEF_DIST'].min(), "to", raw['CLOSE_DEF_DIST'].max(),
      "| beyond 30ft:", int((raw['CLOSE_DEF_DIST'] > 30).sum()))
```

<div style="background:#f4efe4;border-radius:12px;padding:20px 26px;border-left:5px solid #b4531f;font-family:Georgia,serif;line-height:1.85;color:#1a2e22;margin:14px 0;">
<h4 style="color:#b4531f;margin:0 0 10px;">Decision: impossible is not the same as extreme</h4>

Two categories are being separated here, and conflating them is a common way to quietly bias a dataset.
<br><br>
<strong>Impossible.</strong> 312 shots record a negative touch time, one of them at -163.6 seconds. No
interpretation of "seconds of possession before releasing the shot" admits a negative number, so these
are recording errors rather than unusual events. They are set to missing and carry an indicator, exactly
as the shot clock does.
<br><br>
<strong>Extreme but possible.</strong> One shot is logged at 47.2 feet and 44 shots have a nearest
defender more than 30 feet away. Both are unusual, and both happen in real basketball: buzzer heaves from
beyond half court, and completely blown defensive coverage. Removing them because they sit in the tail
would be removing data for being surprising, which is not a statistical justification. They are kept, and
Section 13b's influence diagnostics decide empirically whether any of them actually move the estimates.

</div>

<div style="background:linear-gradient(135deg,#2f7a4d 0%,#14532d 100%);border-radius:8px;padding:11px 20px;margin:24px 0 14px 0;">
<span style="color:#d9cdb4;font-family:Georgia,serif;font-size:12px;font-weight:700;letter-spacing:1px;">6.4</span>
<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">Sequence integrity, the check this project cannot skip</span>
</div>


Every other data-quality issue above affects covariates. This one affects the object of study itself. A
streak is defined by consecutive attempts, so if a shot is missing from the middle of a player's game,
then the shots on either side of the gap are not actually consecutive, and treating them as such would
fabricate streaks that never happened.


```python
seq_check = (raw.groupby(['player_id', 'GAME_ID'])['SHOT_NUMBER']
                .agg(n='size', lo='min', hi='max', distinct='nunique'))
seq_check['clean'] = ((seq_check['lo'] == 1) &
                      (seq_check['hi'] == seq_check['n']) &
                      (seq_check['distinct'] == seq_check['n']))

n_units = len(seq_check)
n_clean = int(seq_check['clean'].sum())
n_broken = n_units - n_clean
has_dupes = int((seq_check['distinct'] < seq_check['n']).sum())

print(f"Player-game units:           {n_units:,}")
print(f"Clean 1..n shot numbering:   {n_clean:,}  ({100*n_clean/n_units:.2f}%)")
print(f"Units with a gap or anomaly: {n_broken}")
print(f"  of which duplicated SHOT_NUMBER: {has_dupes}")
print(f"  of which pure gaps:              {n_broken - has_dupes}")
print("\nExample broken units (n shots recorded vs highest shot number seen):")
print(seq_check.loc[~seq_check['clean'], ['n', 'lo', 'hi']].head(6).to_string())
```

<div style="background:#f3f8f4;border-radius:12px;padding:20px 26px;border-left:5px solid #1c6b3f;font-family:Georgia,serif;line-height:1.85;color:#1a2e22;margin:14px 0;">
<h4 style="color:#14532d;margin:0 0 10px;">Result: 99.36% clean, and the remaining 0.64% are all gaps</h4>

Of 14,343 player-game units, 14,251 number their shots as a clean 1, 2, 3, ..., n. The 92 exceptions are
all gaps rather than duplicates: a shot number is skipped, meaning one attempt in that player's game was
not recorded. No unit contains a repeated shot number, which rules out the more damaging failure mode of
a botched merge duplicating rows.
<br><br>
The rule adopted is that a streak counter resets at any gap, exactly as it resets at the start of a game.
This is deliberately conservative. The alternative, treating shots either side of a gap as adjacent,
would invent a small number of streaks from shots that were never consecutive, and since streaks are the
entire subject of the analysis, inventing them is the one error that cannot be tolerated anywhere.

</div>

<div style="background:linear-gradient(135deg,#2f7a4d 0%,#14532d 100%);border-radius:8px;padding:11px 20px;margin:24px 0 14px 0;">
<span style="color:#d9cdb4;font-family:Georgia,serif;font-size:12px;font-weight:700;letter-spacing:1px;">6.5</span>
<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">Shot value against shot distance</span>
</div>

```python
three_too_close = int(((raw['PTS_TYPE'] == 3) & (raw['SHOT_DIST'] < 22)).sum())
two_too_far = int(((raw['PTS_TYPE'] == 2) & (raw['SHOT_DIST'] > 23.75)).sum())
print(f"Logged as 3PT but under 22ft:      {three_too_close}")
print(f"Logged as 2PT but beyond 23.75ft:  {two_too_far}")
print(f"Combined:                          {three_too_close + two_too_far} "
      f"({100*(three_too_close+two_too_far)/len(raw):.2f}% of rows)")

near_arc = int(((raw['PTS_TYPE'] == 3) & (raw['SHOT_DIST'].between(21.5, 22))).sum())
far_inside = int(((raw['PTS_TYPE'] == 3) & (raw['SHOT_DIST'] < 15)).sum())
print(f"\n  of the 3PT group, within 0.5ft of the corner line (21.5-22ft): "
      f"{near_arc} ({100*near_arc/three_too_close:.0f}%)")
print(f"  of the 3PT group, under 15ft and unexplainable:                {far_inside}")

print("\nDistance distribution of the 3PT-under-22ft group:")
print(raw.loc[(raw['PTS_TYPE'] == 3) & (raw['SHOT_DIST'] < 22), 'SHOT_DIST'].describe().round(2).to_string())
print("\nDistance distribution of the 2PT-beyond-23.75ft group:")
print(raw.loc[(raw['PTS_TYPE'] == 2) & (raw['SHOT_DIST'] > 23.75), 'SHOT_DIST'].describe().round(2).to_string())
```

<div style="background:#fdf5ee;border-radius:14px;padding:22px 28px;border:1.5px solid #d9884f;font-family:Georgia,serif;line-height:1.8;color:#6b3410;margin:14px 0;">
<h4 style="color:#b4531f;margin:0 0 10px;font-size:1.0rem;">Flagged, quantified, and kept</h4>

The NBA three-point line is 23.75 feet at the top of the arc but only 22 feet in the corners, so a shot
recorded as a three from just under 22 feet is geometrically ordinary rather than an error. The
distances bear that out: the median mislabelled three sits at 21.6 feet, 56% of the group falls between
21.5 and 22 feet, and the two-point side is just as tightly packed with a median of 24.2 feet. Both
clusters hug their respective lines, which is the signature of measurement noise in the tracking system
rather than wrong shot values.
<br><br>
A small residue resists that explanation: 48 shots (5.5% of the three-point group) are logged under 15
feet, 13 of them under 5 feet, and one two-pointer is recorded at 43.5 feet. These few cannot be
reconciled with either line and are simply bad rows.
<br><br>
Since the question cannot be settled from this file alone, all of these stay in, and the models use
<code>SHOT_DIST</code> and <code>PTS_TYPE</code> as separate predictors rather than deriving one from the
other. Section 23 repeats the primary analysis with the inconsistent rows excluded, which converts an
unresolvable data question into a reported sensitivity rather than a silent assumption.

</div>

<div style="background:linear-gradient(135deg,#2f7a4d 0%,#14532d 100%);border-radius:8px;padding:11px 20px;margin:24px 0 14px 0;">
<span style="color:#d9cdb4;font-family:Georgia,serif;font-size:12px;font-weight:700;letter-spacing:1px;">6.6</span>
<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">Building the analytical dataset</span>
</div>

```python
df = raw.copy()

# 1. Impossible values to missing, with indicators
df['touch_time_invalid'] = (df['TOUCH_TIME'] < 0).astype(int)
df.loc[df['TOUCH_TIME'] < 0, 'TOUCH_TIME'] = np.nan
df['shot_clock_missing'] = df['SHOT_CLOCK'].isna().astype(int)

# 2. Canonical within-game ordering
df = df.sort_values(['player_id', 'GAME_ID', 'SHOT_NUMBER']).reset_index(drop=True)

# 3. Mark sequence boundaries: a new player-game, or a gap in SHOT_NUMBER
pg_key = df['player_id'].astype(str) + '_' + df['GAME_ID'].astype(str)
new_unit = np.empty(len(df), dtype=bool)
new_unit[0] = True
new_unit[1:] = pg_key.values[1:] != pg_key.values[:-1]

sn = df['SHOT_NUMBER'].values
gap = np.zeros(len(df), dtype=bool)
gap[1:] = (~new_unit[1:]) & (sn[1:] != sn[:-1] + 1)
df['seq_break'] = new_unit | gap   # True => no observed immediate predecessor

# 4. Streak of consecutive makes immediately preceding each shot
made = df['FGM'].values
brk = df['seq_break'].values
streak = np.zeros(len(df), dtype=np.int32)
run = 0
for i in range(len(df)):
    if brk[i]:
        run = 0
    streak[i] = run
    run = run + 1 if made[i] == 1 else 0
df['prev_streak'] = streak

# 5. Lagged outcomes for the temporal-dependence models (Section 16).
#    A lag-j predecessor is valid only if positions i, i-1, ..., i-j+1 are all
#    non-breaks, i.e. the whole window sits inside one unbroken run of shots.
brk_arr = df['seq_break'].values
fgm_arr = df['FGM'].values
n_rows = len(df)
for lag in (1, 2, 3):
    valid = np.ones(n_rows, dtype=bool)
    for j in range(lag):
        shifted = np.full(n_rows, True)
        shifted[j:] = brk_arr[:n_rows - j] if j else brk_arr
        valid &= ~shifted
    valid[:lag] = False
    vals = np.full(n_rows, np.nan)
    vals[lag:] = fgm_arr[:n_rows - lag]
    df[f'lag{lag}_made'] = np.where(valid, vals, np.nan)

analysis = df[~df['seq_break']].copy()   # shots with an observed immediate predecessor

print(f"Raw shots:                                   {len(raw):,}")
print(f"Shots with an observed predecessor:          {len(analysis):,}")
print(f"Excluded as first-of-unit or post-gap:       {len(df) - len(analysis):,}")
print(f"\nStreak distribution among analysable shots:")
print(analysis['prev_streak'].value_counts().sort_index().head(8).to_string())
```

<div style="background:linear-gradient(135deg,#f3f8f4,#e6f0e9);border:1px solid #7aa98c55;border-left:5px solid #1c6b3f;border-radius:0 12px 12px 0;padding:18px 24px;margin:14px 0;">
<div style="font-weight:700;color:#14532d;margin-bottom:10px;font-size:14px;">Why 113,646 and not 128,069</div>
<div style="color:#1a2e22;font-size:13.5px;line-height:1.75;">
Every shot that opens a player's game, and every shot immediately following an unrecorded attempt, has no
observed predecessor, so its preceding-streak length is undefined rather than zero. Those 14,423 shots
are excluded from streak-conditional analysis but retained in the full file for descriptive work and for
the models in Section 13 that do not condition on a predecessor.
<br><br>
The distinction matters more than the count suggests. Coding an undefined streak as zero would silently
add 14,423 pseudo-observations to the "no preceding makes" category, which is precisely the reference
category every streak effect is measured against, and would bias that baseline toward whatever the
first-shot-of-game field-goal rate happens to be.
</div>
</div>

<a id="sec7"></a>
<div style="background:linear-gradient(135deg,#0d3320 0%,#14532d 55%,#2f7a4d 100%);padding:26px 32px;border-radius:14px;margin:34px 0 20px 0;box-shadow:0 6px 24px rgba(13,51,32,0.25);">
<div style="color:rgba(255,255,255,0.55);font-size:11px;letter-spacing:4px;text-transform:uppercase;margin-bottom:8px;">Section 7</div>
<h2 style="color:#ffffff;font-family:Georgia,serif;font-size:24px;margin:0;font-weight:700;">Exploratory Analysis</h2>
<p style="color:#d9cdb4;margin:6px 0 0 0;font-size:13.5px;">The shooting environment, the sequence structure, and how much players differ</p>
</div>


Three things need to be established before any hot-hand estimator is defended. How strongly shot
difficulty drives outcomes, since that is the main rival explanation for any streak pattern. What the
sequence structure actually looks like compared with independence, since that is the phenomenon itself.
And how much players differ from one another, since that determines whether pooling them is defensible.


<div style="background:linear-gradient(135deg,#2f7a4d 0%,#14532d 100%);border-radius:8px;padding:11px 20px;margin:24px 0 14px 0;">
<span style="color:#d9cdb4;font-family:Georgia,serif;font-size:12px;font-weight:700;letter-spacing:1px;">7.1</span>
<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">How much does shot difficulty matter?</span>
</div>

```python
df['dist_bin'] = pd.cut(df['SHOT_DIST'], [0, 4, 8, 12, 16, 20, 22, 24, 27, 30, 50], right=False)
df['def_bin'] = pd.cut(df['CLOSE_DEF_DIST'], [0, 2, 3, 4, 5, 6, 8, 10, 60], right=False)

by_dist = df.groupby('dist_bin', observed=True).agg(fg=('FGM', 'mean'), n=('FGM', 'size'))
by_def = df.groupby('def_bin', observed=True).agg(fg=('FGM', 'mean'), n=('FGM', 'size'))

def wilson(p, n, z=1.96):
    """Wilson score interval, which behaves properly for proportions near 0 and 1."""
    d = 1 + z**2 / n
    c = (p + z**2 / (2*n)) / d
    h = z * np.sqrt(p*(1-p)/n + z**2/(4*n**2)) / d
    return c - h, c + h

for t in (by_dist, by_def):
    lo, hi = wilson(t['fg'].values, t['n'].values)
    t['lo'], t['hi'] = lo, hi

print("Field-goal rate by shot distance:")
print(by_dist[['fg', 'n']].round(4).to_string())
print("\nField-goal rate by closest-defender distance:")
print(by_def[['fg', 'n']].round(4).to_string())
```

```python
fig, axes = plt.subplots(1, 2, figsize=(15, 5.2))

ax = axes[0]
x = np.arange(len(by_dist))
ax.errorbar(x, by_dist['fg'], yerr=[by_dist['fg']-by_dist['lo'], by_dist['hi']-by_dist['fg']],
            marker='o', color=GREEN_DEEP, lw=2, capsize=3, markersize=7)
ax.axhline(df['FGM'].mean(), color=COPPER, ls='--', lw=1.4)
ax.text(len(x)-1.6, df['FGM'].mean()+0.015, f"overall {df['FGM'].mean():.3f}",
        color=COPPER, fontsize=9.5)
ax.set_xticks(x); ax.set_xticklabels([str(i) for i in by_dist.index], rotation=40, fontsize=8.5)
ax.set_title('(a) Shot distance dominates outcome')
ax.set_xlabel('Distance from basket (feet)'); ax.set_ylabel('Field-goal rate')

ax = axes[1]
x = np.arange(len(by_def))
ax.errorbar(x, by_def['fg'], yerr=[by_def['fg']-by_def['lo'], by_def['hi']-by_def['fg']],
            marker='s', color=COPPER, lw=2, capsize=3, markersize=7)
ax.axhline(df['FGM'].mean(), color=GREEN_DEEP, ls='--', lw=1.4)
ax.set_xticks(x); ax.set_xticklabels([str(i) for i in by_def.index], rotation=40, fontsize=8.5)
ax.set_title('(b) Defender distance does not behave monotonically')
ax.set_xlabel('Distance to closest defender (feet)'); ax.set_ylabel('Field-goal rate')

fig.suptitle('Figure 2. The shooting environment, with 95% Wilson intervals',
             fontsize=14, color=GREEN_DEEP, fontweight='bold', y=1.03)
plt.tight_layout(); plt.show()
```

<div style="background:#f3f8f4;border-radius:12px;padding:20px 26px;border-left:5px solid #1c6b3f;font-family:Georgia,serif;line-height:1.85;color:#1a2e22;margin:14px 0;">
<h4 style="color:#14532d;margin:0 0 10px;">Panel (b) is the one that matters for the rest of the notebook</h4>

Distance behaves exactly as basketball intuition predicts: 0.636 inside four feet, falling steadily to
0.093 beyond thirty. The effect is enormous, far larger than anything the hot-hand literature has ever
claimed for streaks, which is precisely why Section 13 treats shot difficulty as the main rival
explanation rather than a nuisance control.
<br><br>
Defender distance is stranger. The field-goal rate is 0.453 with a defender inside two feet, rises to
0.470 at two to three feet, then falls to 0.430 at five to six feet, then climbs again to 0.478 when the
nearest defender is more than ten feet away. More space does not monotonically mean more makes. The
explanation is confounding rather than anything mysterious: a tightly defended shot is often a contested
layup, which is close and therefore easy, while a wide-open shot is often an uncontested three, which is
far and therefore hard. Defender distance and shot distance are entangled, so neither can be interpreted
on its own.
<br><br>
This has a direct consequence for how the adjusted models must be built. Entering defender distance as a
single linear term would impose a monotone relationship that the data plainly reject, so Section 13 uses
a flexible specification and checks it, rather than assuming a straight line and inheriting the
misspecification. Section 6.2 already gave one demonstration of what a badly specified control can do to
an unrelated coefficient.

</div>

```python
pivot = (df.groupby(['dist_bin', 'def_bin'], observed=True)
           .agg(fg=('FGM', 'mean'), n=('FGM', 'size')).reset_index())
grid_fg = pivot.pivot(index='dist_bin', columns='def_bin', values='fg')
grid_n = pivot.pivot(index='dist_bin', columns='def_bin', values='n')
grid_fg = grid_fg.where(grid_n >= 80)   # suppress cells too thin to read

fig, ax = plt.subplots(figsize=(10.5, 6.5))
im = ax.imshow(grid_fg.values, cmap=CMAP_GREEN, aspect='auto', vmin=0.2, vmax=0.7)
ax.set_xticks(range(len(grid_fg.columns)))
ax.set_xticklabels([str(c) for c in grid_fg.columns], rotation=40, fontsize=9)
ax.set_yticks(range(len(grid_fg.index)))
ax.set_yticklabels([str(i) for i in grid_fg.index], fontsize=9)
for i in range(grid_fg.shape[0]):
    for j in range(grid_fg.shape[1]):
        v = grid_fg.values[i, j]
        if not np.isnan(v):
            ax.text(j, i, f'{v:.2f}', ha='center', va='center', fontsize=8.5,
                    color='white' if v > 0.47 else INK)
ax.set_xlabel('Closest defender distance (feet)'); ax.set_ylabel('Shot distance (feet)')
ax.set_title('Figure 3. Field-goal rate over the difficulty plane\n(cells with fewer than 80 attempts left blank)',
             fontsize=13, color=GREEN_DEEP, fontweight='bold')
ax.grid(False)
plt.colorbar(im, ax=ax, label='Field-goal rate', shrink=0.8)
plt.tight_layout(); plt.show()
```

<div style="background:linear-gradient(135deg,#f3f8f4,#e6f0e9);border:1px solid #7aa98c55;border-left:5px solid #1c6b3f;border-radius:0 12px 12px 0;padding:18px 24px;margin:14px 0;">
<div style="font-weight:700;color:#14532d;margin-bottom:10px;font-size:14px;">Reading the difficulty plane</div>
<div style="color:#1a2e22;font-size:13.5px;line-height:1.75;">
Holding shot distance fixed by reading across a single row, defender distance behaves far more sensibly
than it did in Figure 2(b): within almost every distance band, more space is associated with a higher
field-goal rate. The non-monotonicity in the marginal plot was an aggregation artefact produced by mixing
contested layups with open threes, and the joint view separates them. This is the reason the adjusted
models in Section 13 condition on both variables together rather than one at a time.
</div>
</div>

<div style="background:linear-gradient(135deg,#2f7a4d 0%,#14532d 100%);border-radius:8px;padding:11px 20px;margin:24px 0 14px 0;">
<span style="color:#d9cdb4;font-family:Georgia,serif;font-size:12px;font-weight:700;letter-spacing:1px;">7.2</span>
<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">Sequence structure against an independence benchmark</span>
</div>


A streak of makes is not evidence of anything on its own, because independent coin flips produce streaks
too. The useful comparison is between the run lengths actually observed and the run lengths produced by
shuffling each player's own shots within each game, which preserves how many they made while destroying
any order dependence. If shooting really were streakier than chance, the observed distribution should sit
to the right of the shuffled one.


```python
def make_runs(outcomes):
    """Lengths of every maximal run of consecutive makes in one sequence."""
    runs, cur = [], 0
    for v in outcomes:
        if v == 1:
            cur += 1
        else:
            if cur:
                runs.append(cur)
            cur = 0
    if cur:
        runs.append(cur)
    return runs

unit_arrays = [s.values for _, s in df.groupby(['player_id', 'GAME_ID'])['FGM']]

observed_runs = np.array([r for v in unit_arrays for r in make_runs(v)])

# A single shuffle is one draw from the null and carries its own sampling noise.
# Repeating it B times gives a null *band*, which is what the observed value has
# to be judged against.
B_SHUFFLE = 200
rng_local = np.random.default_rng(RANDOM_SEED)
buckets = list(range(1, 9)) + ['9+']
null_counts = {k: [] for k in buckets}
null_means = []
for _ in range(B_SHUFFLE):
    r = np.array([x for v in unit_arrays for x in make_runs(rng_local.permutation(v))])
    for k in range(1, 9):
        null_counts[k].append(int((r == k).sum()))
    null_counts['9+'].append(int((r >= 9).sum()))
    null_means.append(r.mean())
null_means = np.array(null_means)

p_mean_run = 2 * min((null_means <= observed_runs.mean()).mean(),
                     (null_means >= observed_runs.mean()).mean())
print(f"Observed mean run of makes : {observed_runs.mean():.4f}   (longest {observed_runs.max()})")
print(f"Null mean under {B_SHUFFLE} shuffles : {null_means.mean():.4f}  sd={null_means.std():.4f}  "
      f"95% band [{np.percentile(null_means,2.5):.4f}, {np.percentile(null_means,97.5):.4f}]")
print(f"Two-sided permutation p-value for mean run length: {p_mean_run:.4f}")

rows = []
for k in buckets:
    obs_k = int((observed_runs == k).sum()) if k != '9+' else int((observed_runs >= 9).sum())
    nc = np.array(null_counts[k])
    lo, hi = np.percentile(nc, [2.5, 97.5])
    rows.append({'Run length': str(k), 'Observed': obs_k, 'Null mean': nc.mean(),
                 'Null 2.5%': lo, 'Null 97.5%': hi,
                 'Obs / Null': obs_k / nc.mean(),
                 'Outside band': 'yes' if (obs_k < lo or obs_k > hi) else ''})
run_compare = pd.DataFrame(rows)
run_compare.round(3)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(15, 5))
idx = np.arange(len(run_compare))

ax = axes[0]
w = 0.4
ax.bar(idx - w/2, run_compare['Observed'], w, color=GREEN, edgecolor=GREEN_DEEP, label='Observed')
ax.bar(idx + w/2, run_compare['Null mean'], w, color=SAND, edgecolor=COPPER,
       label=f'Null mean ({B_SHUFFLE} shuffles)')
ax.errorbar(idx + w/2, run_compare['Null mean'],
            yerr=[run_compare['Null mean'] - run_compare['Null 2.5%'],
                  run_compare['Null 97.5%'] - run_compare['Null mean']],
            fmt='none', ecolor=COPPER, capsize=3, lw=1.2)
ax.set_yscale('log')
ax.set_xticks(idx); ax.set_xticklabels(run_compare['Run length'])
ax.set_xlabel('Length of a run of consecutive makes'); ax.set_ylabel('Count (log scale)')
ax.set_title('(a) Run-length distribution against the null band')
ax.legend(frameon=False, fontsize=9)

ax = axes[1]
ratio = run_compare['Obs / Null'].values
lo_r = run_compare['Null 2.5%'] / run_compare['Null mean']
hi_r = run_compare['Null 97.5%'] / run_compare['Null mean']
ax.fill_between(idx, lo_r, hi_r, color=SAND, alpha=0.65,
                label='95% null band', zorder=1)
ax.plot(idx, ratio, marker='o', color=GREEN_DEEP, lw=2, zorder=3, label='Observed / null')
out = run_compare['Outside band'] == 'yes'
if out.any():
    ax.scatter(idx[out.values], ratio[out.values], s=140, facecolor='none',
               edgecolor=COPPER, lw=2.2, zorder=4, label='outside band')
ax.axhline(1, color=INK, lw=1.2, zorder=2)
ax.set_xticks(idx); ax.set_xticklabels(run_compare['Run length'])
ax.set_xlabel('Length of a run of consecutive makes'); ax.set_ylabel('Observed / null mean')
ax.set_title('(b) Ratio, with the sampling noise the null itself carries')
ax.legend(frameon=False, fontsize=9)

fig.suptitle('Figure 4. Real shooting sequences against their own shuffled counterparts',
             fontsize=14, color=GREEN_DEEP, fontweight='bold', y=1.03)
plt.tight_layout(); plt.show()
```

<div style="background:#f3f8f4;border-radius:12px;padding:20px 26px;border-left:5px solid #1c6b3f;font-family:Georgia,serif;line-height:1.85;color:#1a2e22;margin:14px 0;">
<h4 style="color:#14532d;margin:0 0 10px;">Result: the sequences are indistinguishable from shuffled versions of themselves</h4>

The mean observed run of consecutive makes is 1.676. Across 200 independent shuffles of each player's own
shots within their own games, the null mean is 1.680 with a standard deviation of 0.004, and the observed
value sits comfortably inside the 95% band, giving a two-sided permutation p-value of 0.31.
<br><br>
The ratio panel is worth reading carefully, because the eye is easily misled here. Plotted against a
single shuffle it appears to slope downward for longer runs, which looks like a systematic deficit of
long streaks. Against the full null band, that apparent pattern is mostly the noise the null itself
carries: counts in the long-run bins are small, so their sampling variability is large. Of the nine
length bins tested, exactly one falls outside its 95% band, which is about what nine simultaneous
comparisons produce by chance.
<br><br>
So at the level of raw run lengths there is no streakiness beyond what reshuffling the same shots
produces, and this deserves stating plainly rather than burying. Two things keep it from being the final
word. The comparison pools every player, so an effect confined to a subset would be diluted, which is
what Section 15's hierarchical model exists to detect. And run length ignores shot difficulty entirely,
so an effect masked by players taking harder shots when hot would not appear here either. The descriptive
null is where the investigation starts, not where it ends.

</div>

<div style="background:linear-gradient(135deg,#2f7a4d 0%,#14532d 100%);border-radius:8px;padding:11px 20px;margin:24px 0 14px 0;">
<span style="color:#d9cdb4;font-family:Georgia,serif;font-size:12px;font-weight:700;letter-spacing:1px;">7.3</span>
<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">How much do players differ?</span>
</div>

```python
player_summary = (df.groupby(['player_id', 'player_name'])
                    .agg(shots=('FGM', 'size'), fg=('FGM', 'mean'),
                         mean_dist=('SHOT_DIST', 'mean'),
                         mean_def=('CLOSE_DEF_DIST', 'mean'))
                    .reset_index())

print(f"Players: {len(player_summary)}")
print(f"Field-goal rate across players: min={player_summary['fg'].min():.3f}, "
      f"median={player_summary['fg'].median():.3f}, max={player_summary['fg'].max():.3f}, "
      f"sd={player_summary['fg'].std():.4f}")

# How much of that spread is real skill rather than small-sample noise?
p_bar = df['FGM'].mean()
expected_noise_var = np.mean(p_bar * (1 - p_bar) / player_summary['shots'])
observed_var = player_summary['fg'].var()
print(f"\nObserved variance in player FG%:            {observed_var:.6f}")
print(f"Variance expected from sampling noise alone: {expected_noise_var:.6f}")
print(f"Implied true between-player variance:        {observed_var - expected_noise_var:.6f}")
print(f"Implied true between-player SD:              {np.sqrt(max(observed_var - expected_noise_var, 0)):.4f}")
print(f"Share of observed spread that is real:       {1 - expected_noise_var/observed_var:.3f}")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

ax = axes[0]
ax.scatter(player_summary['shots'], player_summary['fg'], s=26, alpha=0.55,
           color=GREEN, edgecolor=GREEN_DEEP, linewidth=0.4)
ax.axhline(p_bar, color=COPPER, ls='--', lw=1.5, label=f'league {p_bar:.3f}')
# funnel showing the spread expected from noise alone
grid = np.linspace(player_summary['shots'].min(), player_summary['shots'].max(), 200)
se = np.sqrt(p_bar * (1 - p_bar) / grid)
ax.plot(grid, p_bar + 1.96*se, color=COPPER, lw=1.2, alpha=0.8)
ax.plot(grid, p_bar - 1.96*se, color=COPPER, lw=1.2, alpha=0.8,
        label='95% band under pure noise')
ax.set_xlabel('Shots attempted'); ax.set_ylabel('Field-goal rate')
ax.set_title('(a) Player shooting against a pure-noise funnel')
ax.legend(frameon=False, fontsize=9)

ax = axes[1]
ax.hist(player_summary['fg'], bins=32, color=GREEN_SOFT, edgecolor=GREEN_DEEP)
ax.axvline(p_bar, color=COPPER, ls='--', lw=1.6)
ax.set_xlabel('Field-goal rate'); ax.set_ylabel('Players')
ax.set_title('(b) Distribution of player shooting rates')

fig.suptitle('Figure 5. Player heterogeneity is real and large enough to require modelling',
             fontsize=14, color=GREEN_DEEP, fontweight='bold', y=1.03)
plt.tight_layout(); plt.show()
```

<div style="background:linear-gradient(135deg,#f3f8f4,#e6f0e9);border:1px solid #7aa98c55;border-left:5px solid #1c6b3f;border-radius:0 12px 12px 0;padding:18px 24px;margin:14px 0;">
<div style="font-weight:700;color:#14532d;margin-bottom:10px;font-size:14px;">Why the models cannot pool players</div>
<div style="color:#1a2e22;font-size:13.5px;line-height:1.75;">
Player field-goal rates spread far outside the funnel that sampling noise alone would produce, and
decomposing the variance confirms it: most of the observed spread reflects genuine differences in
shooting ability rather than small samples. A model with a single common intercept would attribute that
spread to something else, and in a streak analysis the natural thing for it to be misattributed to is the
streak variable itself, since high-percentage shooters mechanically produce more streaks.
<br><br>
Every model from Section 12 onward therefore carries a player-specific intercept. Figure 5(a) also shows
why the hierarchical treatment in Section 15 is worth the extra machinery: players at the low-volume end
have field-goal rates estimated from as few as 47 shots, and their unpooled estimates are far noisier
than the high-volume players', which is exactly the situation partial pooling exists to handle.
</div>
</div>

<div style="background:linear-gradient(135deg,#2f7a4d 0%,#14532d 100%);border-radius:8px;padding:11px 20px;margin:24px 0 14px 0;">
<span style="color:#d9cdb4;font-family:Georgia,serif;font-size:12px;font-weight:700;letter-spacing:1px;">7.4</span>
<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">The same question, drawn on a basketball court</span>
</div>


Section 7.1 established that shot difficulty dominates outcomes and Section 13 will show that streaks
change which shots get taken. Both are easier to judge as geometry than as coefficients. The dataset
records distance from the basket but not the angle, so a true shot chart is not available; what can be
drawn honestly is a half-court schematic in distance bands, with each band coloured by how often shots
from it went in and labelled with how often they were taken.

The two panels hold the streak state fixed. On the left are attempts taken with no preceding make, on
the right attempts taken after three consecutive makes. If shot selection were unaffected by a run of
makes the two panels would carry the same labels.


```python
from matplotlib.patches import Wedge, Circle, Rectangle, Arc

COURT_BANDS = [0, 4, 8, 12, 16, 20, 23.75, 28, 40]
streak_state = np.minimum(analysis['prev_streak'].values, 3)
court_data = analysis.assign(K=streak_state,
                              band=pd.cut(analysis['SHOT_DIST'], COURT_BANDS, right=False))

def draw_half_court(ax):
    """A schematic half court in feet, basket at the origin."""
    ax.add_patch(Circle((0, 0), 0.75, fill=False, color=INK, lw=1.6))          # rim
    ax.add_patch(Rectangle((-3, -1.2), 6, 0.2, color=INK))                      # backboard
    ax.add_patch(Rectangle((-8, 0), 16, 19, fill=False, color=INK, lw=1.4))     # the paint
    ax.add_patch(Arc((0, 19), 12, 12, theta1=0, theta2=180, color=INK, lw=1.4)) # free-throw circle
    ax.add_patch(Rectangle((-22, 0), 0.1, 14, color=INK))                       # corner three, left
    ax.add_patch(Rectangle((22, 0), 0.1, 14, color=INK))                        # corner three, right
    ax.add_patch(Arc((0, 0), 47.5, 47.5, theta1=22, theta2=158, color=INK, lw=1.5))
    ax.set_xlim(-25, 25); ax.set_ylim(-2, 34)
    ax.set_aspect('equal'); ax.axis('off')


fig, axes = plt.subplots(1, 2, figsize=(15.5, 7.6))
panels = [(0, 'No preceding make  (K = 0)'), (3, 'After three makes  (K = 3)')]
norm = plt.Normalize(0.28, 0.62)

for ax, (state, title) in zip(axes, panels):
    sub = court_data[court_data['K'] == state]
    band_stats = sub.groupby('band', observed=True).agg(n=('FGM', 'size'), fg=('FGM', 'mean'))
    band_stats['share'] = band_stats['n'] / band_stats['n'].sum()
    draw_half_court(ax)
    for band, row in band_stats.iterrows():
        inner, outer = band.left, min(band.right, 34)
        ax.add_patch(Wedge((0, 0), outer, 10, 170, width=outer - inner,
                           facecolor=CMAP_GREEN(norm(row['fg'])), alpha=0.82,
                           edgecolor='white', lw=1.2, zorder=0))
        mid = (inner + outer) / 2
        if row['share'] > 0.012:
            ax.text(0, mid, f"{row['fg']:.0%}\n{row['share']:.0%} of shots",
                    ha='center', va='center', fontsize=8.6, zorder=5,
                    color='white' if row['fg'] > 0.47 else INK, fontweight='bold')
    ax.set_title(f"{title}\nmean distance {sub['SHOT_DIST'].mean():.2f} ft, "
                 f"overall {sub['FGM'].mean():.1%}",
                 fontsize=12, color=GREEN_DEEP, fontweight='bold')

sm = plt.cm.ScalarMappable(cmap=CMAP_GREEN, norm=norm)
cb = fig.colorbar(sm, ax=axes, shrink=0.65, pad=0.02)
cb.set_label('Field-goal rate in that band')
fig.suptitle('Figure 6. Where shots come from, and how often they fall, by streak state',
             fontsize=14, color=GREEN_DEEP, fontweight='bold', y=0.99)
plt.show()

print("Share of attempts taken from beyond the three-point arc:")
for state in (0, 1, 2, 3):
    sub = court_data[court_data['K'] == state]
    print(f"  K = {state}: {100*(sub['SHOT_DIST'] >= 23.75).mean():5.2f}% long threes   "
          f"mean distance {sub['SHOT_DIST'].mean():5.2f} ft   FG% {sub['FGM'].mean():.4f}")
```

<div style="background:#f3f8f4;border-radius:12px;padding:20px 26px;border-left:5px solid #1c6b3f;font-family:Georgia,serif;line-height:1.85;color:#1a2e22;margin:14px 0;">
<h4 style="color:#14532d;margin:0 0 10px;">Reading the court: the shot mix moves outward, the shooting does not improve</h4>

The colour of each band is close to unchanged between panels, which is the first thing to notice. A shot
from sixteen to twenty feet goes in about 41% of the time with no preceding make and about 39% after
three, and most other bands drift down by one to three points in the same way. The one exception runs the
other way and is worth naming rather than smoothing over: at the rim the rate rises from 63% to 66% after
three makes, while the share of attempts taken there falls from 22% to 19%. A plausible reading is
selection within the zone, with a hot shooter taking fewer but better-chosen attempts close in, but this
is a single unadjusted contrast and it is offered as an observation rather than a finding.
<br><br>
What does change is the labels. The share of attempts taken from beyond the arc rises from 18.7% with no
preceding make to 20.1% after three, and average distance rises from 13.2 to 13.9 feet. Players are
reaching further out. Because the bands they are reaching into convert at 32% rather than the 49% of the
close range they are leaving, the mix shift alone lowers the expected make rate even if shooting skill is
identical.
<br><br>
This is the whole mechanism of Section 13 rendered as geometry. The pooled drop in field-goal percentage
after a streak is not the shooter getting worse at any given shot; it is the shooter attempting a
different and harder distribution of shots. The pooled comparison mixes those two things together, which
is exactly why the adjusted models hold the shot fixed. Note that these are raw pooled figures, so part
of the distance gap reflects which players accumulate streaks; Section 13's within-player version
isolates the behavioural shift and finds it larger still, about 1.7 feet.

</div>

<div style="background:linear-gradient(135deg,#2f7a4d 0%,#14532d 100%);border-radius:8px;padding:11px 20px;margin:24px 0 14px 0;">
<span style="color:#d9cdb4;font-family:Georgia,serif;font-size:12px;font-weight:700;letter-spacing:1px;">7.5</span>
<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">What a streak actually looks like</span>
</div>


Statistical arguments about streaks can drift away from the thing being argued about. This figure puts
real shooting sequences on the page: each row is one player's game, each square one shot in order, filled
for a make and hollow for a miss, with runs of three or more makes outlined. Beneath each real sequence
sits the same shots reshuffled at random, which is the null this notebook keeps testing against.


```python
rng_ribbon = np.random.default_rng(RANDOM_SEED)

# pick real player-games that contain a long run, so there is something to look at
candidates = []
for (pid, gid), g in analysis.groupby(['player_id', 'GAME_ID']):
    v = g['FGM'].values
    if len(v) < 14 or not (0.40 <= v.mean() <= 0.80):
        continue   # a near-perfect or near-empty game makes the reshuffled row trivial
    longest = best = 0
    for o in v:
        best = best + 1 if o == 1 else 0
        longest = max(longest, best)
    if longest >= 4:
        candidates.append((g['player_name'].iloc[0], v, longest))
candidates.sort(key=lambda t: -t[2])
chosen = candidates[:5]

def runs_marked(v, min_run=3):
    """Index positions belonging to a run of at least min_run makes."""
    flags = np.zeros(len(v), dtype=bool)
    start = None
    for i, o in enumerate(list(v) + [0]):
        if o == 1 and start is None:
            start = i
        elif o != 1 and start is not None:
            if i - start >= min_run:
                flags[start:i] = True
            start = None
    return flags

fig, ax = plt.subplots(figsize=(15, 7.4))
row = 0
yticks, ylabels = [], []
for name, v, longest in chosen:
    shuffled = rng_ribbon.permutation(v)
    for seq, tag in [(v, 'observed'), (shuffled, 'reshuffled')]:
        marked = runs_marked(seq)
        for i, o in enumerate(seq):
            ax.add_patch(plt.Rectangle((i, -row), 0.86, 0.86,
                                        facecolor=GREEN_DEEP if o == 1 else 'white',
                                        edgecolor=COPPER if marked[i] else GREY,
                                        lw=2.2 if marked[i] else 0.8))
        yticks.append(-row + 0.43)
        ylabels.append(f"{name}  ({tag})" if tag == 'observed' else f"    {tag}")
        row += 1
    row += 0.6

ax.set_xlim(-0.5, max(len(v) for _, v, _ in chosen) + 0.5)
ax.set_ylim(-row, 1.2)
ax.set_yticks(yticks); ax.set_yticklabels(ylabels, fontsize=9)
ax.set_xlabel('Shot number within the game')
ax.set_xticks(range(0, 40, 2))
ax.grid(False)
for spine in ax.spines.values():
    spine.set_visible(False)
ax.set_title('Figure 7. Five real shooting nights, each paired with the same shots reshuffled\n'
             'filled = make, hollow = miss, copper outline = run of three or more',
             fontsize=13, color=GREEN_DEEP, fontweight='bold')
plt.tight_layout(); plt.show()
```

<div style="background:linear-gradient(135deg,#f3f8f4,#e6f0e9);border:1px solid #7aa98c55;border-left:5px solid #1c6b3f;border-radius:0 12px 12px 0;padding:18px 24px;margin:14px 0;">
<div style="font-weight:700;color:#14532d;margin-bottom:10px;font-size:14px;">Why this figure is included, and what it cannot show</div>
<div style="color:#1a2e22;font-size:13.5px;line-height:1.75;">
These five games were selected precisely because they contain long runs, so they are not a random sample
and nothing should be inferred from them quantitatively. They are here for a different purpose: to make
the comparison the analysis rests on visible. The reshuffled row beneath each real one contains exactly
the same makes and misses in a different order, and it too tends to produce runs of three, four,
sometimes more.
<br><br>
That is the intuition the whole notebook is built on and it is genuinely hard to hold onto by argument
alone. Streaks look meaningful. They also appear readily in sequences constructed to have no memory at
all, which is why Section 7.2 compares run-length distributions against a null band rather than against
the impression a few games give, and why Section 11 compares each player's statistic against reorderings
of their own shots. A figure that makes the null look as streaky as the data is doing more work than a
figure that makes the data look streaky.
</div>
</div>

<div style="background:linear-gradient(135deg,#2f7a4d 0%,#14532d 100%);border-radius:8px;padding:11px 20px;margin:24px 0 14px 0;">
<span style="color:#d9cdb4;font-family:Georgia,serif;font-size:12px;font-weight:700;letter-spacing:1px;">7.6</span>
<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">The distance shift, shot by shot</span>
</div>

```python
fig, axes = plt.subplots(1, 2, figsize=(15, 5.4))

ax = axes[0]
offsets = {0: 0, 1: 1, 2: 2, 3: 3}
grid = np.linspace(0, 32, 220)
for state in (3, 2, 1, 0):
    d = analysis.loc[streak_state == state, 'SHOT_DIST'].values
    kde = stats.gaussian_kde(d, bw_method=0.22)
    dens = kde(grid)
    dens = dens / dens.max() * 0.85
    base = offsets[state]
    shade = CMAP_GREEN(0.25 + 0.2 * state)
    ax.fill_between(grid, base, base + dens, color=shade, alpha=0.9,
                    edgecolor=GREEN_DEEP, lw=1.2, zorder=4 - state)
    ax.plot([d.mean(), d.mean()], [base, base + 0.85], color=COPPER, lw=2, zorder=10)
    ax.text(31.5, base + 0.35, f"K={state}\nmean {d.mean():.2f} ft",
            fontsize=9, ha='right', color=INK)
ax.axvline(23.75, color=GREY, ls='--', lw=1.4)
ax.text(24.1, 3.6, 'three-point arc', fontsize=8.5, color=GREY, rotation=90, va='top')
ax.set_yticks([]); ax.set_xlabel('Shot distance (feet)')
ax.set_title('(a) The whole distance distribution shifts outward')

ax = axes[1]
zones = ['0-4 ft\n(at the rim)', '4-16 ft\n(mid range)', '16-23.75 ft\n(long two)',
         '23.75+ ft\n(three)']
edges = [(0, 4), (4, 16), (16, 23.75), (23.75, 60)]
width = 0.2
for j, state in enumerate((0, 1, 2, 3)):
    sub = analysis[streak_state == state]
    shares = [((sub['SHOT_DIST'] >= lo) & (sub['SHOT_DIST'] < hi)).mean() for lo, hi in edges]
    ax.bar(np.arange(4) + (j - 1.5) * width, shares, width,
           color=CMAP_GREEN(0.25 + 0.2 * state), edgecolor=GREEN_DEEP,
           label=f'K = {state}')
ax.set_xticks(range(4)); ax.set_xticklabels(zones, fontsize=9)
ax.set_ylabel('Share of that state\'s attempts')
ax.set_title('(b) Which zones gain and which lose')
ax.legend(frameon=False, fontsize=9, ncol=4)

fig.suptitle('Figure 8. Shot selection after a run of makes',
             fontsize=14, color=GREEN_DEEP, fontweight='bold', y=1.02)
plt.tight_layout(); plt.show()
```

<div style="background:#f3f8f4;border-radius:12px;padding:20px 26px;border-left:5px solid #1c6b3f;font-family:Georgia,serif;line-height:1.85;color:#1a2e22;margin:14px 0;">
<h4 style="color:#14532d;margin:0 0 10px;">The shift is small per shot and consistent across the distribution</h4>

Panel (a) shows the change is not confined to a few extreme attempts. The whole density moves, with the
mid-range trough deepening slightly and mass accumulating past the arc, and the mean distance rising
monotonically with the length of the preceding run. Panel (b) locates it: the share of attempts from the
mid range and the long two falls, and the share from beyond the arc rises.
<br><br>
The magnitudes are modest, a percentage point or two of shot mix, which is worth emphasising because it
would be easy to over-dramatise. No player is transformed into a different shooter by making three in a
row. But a one-point shift out of a 49% zone and into a 32% zone is arithmetically enough to move a
pooled field-goal percentage by more than the entire streak effect being argued about, which is why an
unadjusted comparison cannot settle anything.
<br><br>
Whether this reflects the shooter's own confidence, the defence backing off, or a coach tolerating a
longer attempt from someone who is going well cannot be separated with this data, and the notebook does
not try. What matters statistically is only that the shots being compared are not alike.

<a id="sec8"></a>
<div style="background:linear-gradient(135deg,#0d3320 0%,#14532d 55%,#2f7a4d 100%);padding:26px 32px;border-radius:14px;margin:34px 0 20px 0;box-shadow:0 6px 24px rgba(13,51,32,0.25);">
<div style="color:rgba(255,255,255,0.55);font-size:11px;letter-spacing:4px;text-transform:uppercase;margin-bottom:8px;">Section 8</div>
<h2 style="color:#ffffff;font-family:Georgia,serif;font-size:24px;margin:0;font-weight:700;">The Classical Estimator</h2>
<p style="color:#d9cdb4;margin:6px 0 0 0;font-size:13.5px;">Reconstructing the measure the 1985 study used, and applying it here</p>
</div>


Gilovich, Vallone and Tversky asked whether a player shoots better immediately after a run of makes than
they do in general. Formally, for a binary sequence of outcomes within one player's game, define


$$Y_t \in \{0,1\}, \qquad
\hat p_k \;=\; \widehat{P}\big(Y_{t+1}=1 \;\big|\; Y_{t-k+1}=\cdots=Y_t=1\big)$$


and compare it to the player's overall rate $\hat p$. The natural test statistic is the difference


$$T_k \;=\; \hat p_k \;-\; \hat p$$


with $T_k > 0$ read as shooting better after a streak. The estimator is computed within each unbroken
sequence and pooled, never across a game boundary or across one of the gaps identified in Section 6.4.


```python
def streak_conditional_stat(sequences, k):
    """P(make | previous k shots all makes), computed for many sequences at once.

    `sequences` is a (B, n) array of 0/1 outcomes.  Returns a (B,) array, with nan
    wherever the sequence never contains k consecutive makes followed by another shot,
    which is precisely the selection step that creates the bias studied in Section 9.
    """
    B, n = sequences.shape
    if n <= k:
        return np.full(B, np.nan)
    window = np.ones((B, n - k), dtype=bool)
    for j in range(k):
        window &= sequences[:, j:n - k + j].astype(bool)
    following = sequences[:, k:].astype(bool)
    denom = window.sum(1)
    numer = (window & following).sum(1)
    out = np.full(B, np.nan)
    ok = denom > 0
    out[ok] = numer[ok] / denom[ok]
    return out


# Applied to the real data, pooling across every unbroken player-game sequence
classical_rows = []
for k in range(1, 5):
    mask = analysis['prev_streak'] >= k
    classical_rows.append({
        'k': k,
        'P(make | prev k makes)': analysis.loc[mask, 'FGM'].mean(),
        'P(make) overall': analysis['FGM'].mean(),
        'T_k': analysis.loc[mask, 'FGM'].mean() - analysis['FGM'].mean(),
        'n eligible shots': int(mask.sum()),
    })
classical_table = pd.DataFrame(classical_rows)
classical_table.round(4)
```

```python
ks = classical_table['k'].values
fig, ax = plt.subplots(figsize=(9.5, 5.4))

cond = classical_table['P(make | prev k makes)'].values
n_el = classical_table['n eligible shots'].values
se = np.sqrt(cond * (1 - cond) / n_el)

ax.errorbar(ks, cond, yerr=1.96 * se, marker='o', markersize=9, lw=2.2,
            color=GREEN_DEEP, capsize=5, label='P(make | previous k makes)')
ax.axhline(analysis['FGM'].mean(), color=COPPER, ls='--', lw=1.8,
           label=f"unconditional P(make) = {analysis['FGM'].mean():.4f}")
for k, c, n in zip(ks, cond, n_el):
    ax.annotate(f'n={n:,}', (k, c), textcoords='offset points', xytext=(0, -22),
                ha='center', fontsize=9, color=GREY)
ax.set_xticks(ks)
ax.set_xlabel('k, length of the preceding run of makes')
ax.set_ylabel('Field-goal rate')
ax.set_title('Figure 9. The classical conditional estimator on 2014-15 NBA shots',
             fontsize=13, color=GREEN_DEEP, fontweight='bold')
ax.legend(frameon=False, fontsize=9.5)
plt.tight_layout(); plt.show()
```

<div style="background:#f3f8f4;border-radius:12px;padding:20px 26px;border-left:5px solid #1c6b3f;font-family:Georgia,serif;line-height:1.85;color:#1a2e22;margin:14px 0;">
<h4 style="color:#14532d;margin:0 0 10px;">Result: shooting after a streak looks worse, not better</h4>

Every conditional rate sits below the unconditional one, and the gap widens as the required run gets
longer. Taken at face value this is the pattern Gilovich, Vallone and Tversky reported, only more so: not
merely an absence of hot-hand shooting but an apparent cold hand.
<br><br>
Two separate reasons to distrust that reading are already on the table. Section 7.1 showed shot
difficulty has an enormous effect on outcomes, and nothing here holds it fixed, so if players attempt
harder shots after making a few, this is what it would look like. And Section 9 shows the estimator
itself is biased downward under the null, so part of the gap is an artefact of how the quantity is
constructed rather than anything about basketball. The rest of this section deals with the second
problem, and Section 13 with the first.

</div>

<a id="sec9"></a>
<div style="background:linear-gradient(135deg,#0d3320 0%,#14532d 55%,#2f7a4d 100%);padding:26px 32px;border-radius:14px;margin:34px 0 20px 0;box-shadow:0 6px 24px rgba(13,51,32,0.25);">
<div style="color:rgba(255,255,255,0.55);font-size:11px;letter-spacing:4px;text-transform:uppercase;margin-bottom:8px;">Section 9</div>
<h2 style="color:#ffffff;font-family:Georgia,serif;font-size:24px;margin:0;font-weight:700;">Finite-Sample Bias in the Classical Estimator</h2>
<p style="color:#d9cdb4;margin:6px 0 0 0;font-size:13.5px;">Why conditioning on an observed streak is a selection step</p>
</div>


Miller and Sanjurjo (2018) prove that $\hat p_k$ is a biased estimator of $p$ even when outcomes are
completely independent. The bias is negative, grows with the streak length $k$, and shrinks as the
sequence gets longer. It is worth deriving the intuition rather than quoting the theorem, because the
mechanism is not obvious and the result is easy to misstate.


<div style="background:linear-gradient(135deg,#2f7a4d 0%,#14532d 100%);border-radius:8px;padding:11px 20px;margin:24px 0 14px 0;">
<span style="color:#d9cdb4;font-family:Georgia,serif;font-size:12px;font-weight:700;letter-spacing:1px;">9.1</span>
<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">The smallest case that shows it</span>
</div>


Take three fair coin flips and ask for the proportion of heads that immediately follow a head. Enumerate
every sequence in which that quantity is even defined, and average it.


```python
from itertools import product

rows = []
for seq in product([0, 1], repeat=3):
    arr = np.array(seq)
    eligible = [t for t in range(2) if arr[t] == 1]
    if not eligible:
        value = None
    else:
        value = np.mean([arr[t + 1] for t in eligible])
    rows.append({
        'sequence': ''.join('H' if v else 'T' for v in seq),
        'flips following a head': len(eligible),
        'proportion heads': value,
    })
enumeration = pd.DataFrame(rows)

defined = enumeration.dropna(subset=['proportion heads'])
print(enumeration.to_string(index=False))
print(f"\nSequences where the statistic is defined : {len(defined)} of 8")
print(f"Average of the statistic over those        : {defined['proportion heads'].mean():.6f}")
print(f"As a fraction                              : {Fraction(defined['proportion heads'].mean()).limit_denominator(100)}")
print(f"True probability of heads                  : 0.5")
print(f"Bias                                       : {defined['proportion heads'].mean() - 0.5:+.6f}")
```

<div style="background:linear-gradient(135deg,#f3f8f4,#e6f0e9);border:1px solid #7aa98c55;border-left:5px solid #1c6b3f;border-radius:0 12px 12px 0;padding:18px 24px;margin:14px 0;">
<div style="font-weight:700;color:#14532d;margin-bottom:10px;font-size:14px;">Where the 5/12 comes from</div>
<div style="color:#1a2e22;font-size:13.5px;line-height:1.75;">
Six of the eight sequences contain at least one head in a position that has a successor, so the statistic
is defined for six of them. Averaging its value across those six gives 5/12, not 1/2. Nothing about the
coin has changed; the coin is still fair and every flip is still independent.
<br><br>
The mechanism is a selection effect on positions rather than on flips. Conditioning on "this position is
a head" and then reading the next flip means that within any single finite sequence, the heads available
to be counted as predecessors and the flips available to be counted as successors are drawn from the same
limited pool. A sequence with many heads offers many predecessor positions, so its successor flips are
weighted heavily; a sequence with one head offers exactly one, and that single observation carries the
same weight as an entire sequence's worth of evidence. Averaging a ratio across sequences is not the same
as taking the ratio of the averages, and the discrepancy is systematically negative.
<br><br>
The result generalises: the bias worsens as the required run gets longer, because longer runs are rarer
and the selection becomes more severe, and it fades as the sequence grows, because the pool stops being
scarce.
</div>
</div>

<div style="background:linear-gradient(135deg,#2f7a4d 0%,#14532d 100%);border-radius:8px;padding:11px 20px;margin:24px 0 14px 0;">
<span style="color:#d9cdb4;font-family:Georgia,serif;font-size:12px;font-weight:700;letter-spacing:1px;">9.2</span>
<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">How the bias behaves across sequence length and streak length</span>
</div>

```python
def exact_expected_stat(n, k, p=0.5):
    """Exact E[p-hat_k] by enumerating all 2^n sequences.  Feasible to about n=18."""
    seqs = np.array(list(product([0, 1], repeat=n)), dtype=np.int8)
    weights = np.prod(np.where(seqs == 1, p, 1 - p), axis=1)
    vals = streak_conditional_stat(seqs, k)
    ok = ~np.isnan(vals)
    return float(np.sum(vals[ok] * weights[ok]) / np.sum(weights[ok]))


def mc_expected_stat(n, k, p, n_sim, rng):
    """Monte Carlo E[p-hat_k] for sequence lengths too long to enumerate."""
    S = (rng.random((n_sim, n)) < p).astype(np.int8)
    v = streak_conditional_stat(S, k)
    return float(np.nanmean(v))


rng_bias = np.random.default_rng(RANDOM_SEED)
EXACT_MAX_N = 16
lengths = [5, 10, 16, 25, 50, 100, 200, 500]
streaks = [1, 2, 3]
p_true = 0.5

bias_rows = []
for n in lengths:
    for k in streaks:
        if n <= k + 1:
            continue
        if n <= EXACT_MAX_N:
            e = exact_expected_stat(n, k, p_true); method = 'exact'
        else:
            e = mc_expected_stat(n, k, p_true, 40000, rng_bias); method = 'Monte Carlo'
        bias_rows.append({'n': n, 'k': k, 'E[p-hat_k]': e,
                          'bias': e - p_true, 'method': method})
bias_grid = pd.DataFrame(bias_rows)
bias_grid.round(5)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(15.5, 5.3))

ax = axes[0]
for k, colour in zip(streaks, [GREEN_DEEP, GREEN_MID, COPPER]):
    sub = bias_grid[bias_grid['k'] == k]
    ax.plot(sub['n'], sub['bias'], marker='o', lw=2, color=colour, label=f'k = {k}')
ax.axhline(0, color=INK, lw=1.2)
ax.set_xscale('log')
ax.set_xlabel('Sequence length n (log scale)')
ax.set_ylabel(r'Bias of $\hat p_k$')
ax.set_title('(a) Bias shrinks with sequence length, grows with streak length')
ax.legend(frameon=False, fontsize=10)

ax = axes[1]
pivot = bias_grid.pivot(index='k', columns='n', values='bias')
im = ax.imshow(pivot.values, cmap=CMAP_DIV, aspect='auto',
               vmin=-np.nanmax(np.abs(pivot.values)), vmax=np.nanmax(np.abs(pivot.values)))
ax.set_xticks(range(len(pivot.columns))); ax.set_xticklabels(pivot.columns)
ax.set_yticks(range(len(pivot.index))); ax.set_yticklabels([f'k={i}' for i in pivot.index])
for i in range(pivot.shape[0]):
    for j in range(pivot.shape[1]):
        v = pivot.values[i, j]
        if not np.isnan(v):
            ax.text(j, i, f'{v:+.3f}', ha='center', va='center', fontsize=9,
                    color='white' if abs(v) > 0.09 else INK)
ax.set_xlabel('Sequence length n'); ax.grid(False)
ax.set_title('(b) The same surface, as a table')

fig.suptitle('Figure 10. Streak-selection bias under pure independence (p = 0.5)',
             fontsize=14, color=GREEN_DEEP, fontweight='bold', y=1.03)
plt.tight_layout(); plt.show()
```

<div style="background:#f3f8f4;border-radius:12px;padding:20px 26px;border-left:5px solid #1c6b3f;font-family:Georgia,serif;line-height:1.85;color:#1a2e22;margin:14px 0;">
<h4 style="color:#14532d;margin:0 0 10px;">The bias is large exactly where hot-hand studies operate</h4>

At a sequence length of 100, roughly a season's worth of attempts for a rotation player, the bias is
about half a percentage point for single-make streaks but four percentage points once three consecutive
makes are required. At the shorter lengths typical of controlled shooting experiments it is severe: with
25 shots and a three-make streak the estimator sits nearly 13 percentage points below the truth, and with
10 shots almost 15. Effects of that magnitude are larger than anything the hot-hand literature has ever
claimed to detect.
<br><br>
This is the whole point of Miller and Sanjurjo's contribution, and it is worth stating carefully. Their
result is a theorem about an estimator, not a measurement of basketball. It says that if you use this
statistic and compare it to zero, you will conclude against the hot hand even when the hot hand is
present, because the benchmark is wrong. Whether a hot hand actually exists in any particular dataset is
a separate empirical question, and the correction is what makes that question answerable rather than what
answers it.

</div>

<a id="sec10"></a>
<div style="background:linear-gradient(135deg,#0d3320 0%,#14532d 55%,#2f7a4d 100%);padding:26px 32px;border-radius:14px;margin:34px 0 20px 0;box-shadow:0 6px 24px rgba(13,51,32,0.25);">
<div style="color:rgba(255,255,255,0.55);font-size:11px;letter-spacing:4px;text-transform:uppercase;margin-bottom:8px;">Section 10</div>
<h2 style="color:#ffffff;font-family:Georgia,serif;font-size:24px;margin:0;font-weight:700;">Monte Carlo Estimator Study</h2>
<p style="color:#d9cdb4;margin:6px 0 0 0;font-size:13.5px;">Bias, variance, mean squared error and interval coverage, measured rather than asserted</p>
</div>


Knowing an estimator is biased is not the same as knowing whether to use it. A biased estimator with low
variance can beat an unbiased one with high variance, so the comparison has to be made on mean squared
error and on whether the accompanying intervals behave as advertised. This section treats the choice of
estimator as an experiment with measured outcomes.

Three quantities are compared for each estimator $\hat\theta$ of the true difference $\theta$:


$$\operatorname{Bias}(\hat\theta) = \mathbb{E}[\hat\theta] - \theta,
\qquad \operatorname{Var}(\hat\theta), \qquad
\operatorname{MSE}(\hat\theta) = \operatorname{Bias}(\hat\theta)^2 + \operatorname{Var}(\hat\theta)$$


together with the proportion of replications in which the 95% permutation interval contains the truth.
Data are generated under independence, so $\theta = 0$ throughout and any systematic departure from zero
is bias rather than signal.

The corrected estimator is the naive statistic minus its own permutation null mean, computed separately
for every simulated sequence:


$$T^{\text{corr}} = T_k - \widehat{\mathbb{E}}\big[T_k \mid \text{random reordering of this sequence}\big]$$


Reordering preserves how many shots were made and destroys only the order, so the subtracted quantity is
exactly the bias the selection step introduces for a sequence of that length and that make count.


```python
def naive_and_corrected(sequence, k, n_perm, rng):
    """Naive difference statistic and its permutation-debiased counterpart for one sequence."""
    base = sequence.mean()
    observed = streak_conditional_stat(sequence[None, :], k)[0]
    if np.isnan(observed):
        return np.nan, np.nan, None
    t_naive = observed - base

    order = np.argsort(rng.random((n_perm, len(sequence))), axis=1)
    permuted = streak_conditional_stat(sequence[order], k) - base
    null = permuted[~np.isnan(permuted)]
    if len(null) < 20:
        return t_naive, np.nan, None
    return t_naive, t_naive - null.mean(), null


rng_mc = np.random.default_rng(RANDOM_SEED)
N_REP, N_PERM = 600, 300
P_TRUE = 0.45                    # close to the league-wide field-goal rate
grid_n = [25, 50, 100, 300]
grid_k = [1, 2, 3]

mc_rows = []
for n in grid_n:
    for k in grid_k:
        naive_vals, corr_vals, covered = [], [], []
        for _ in range(N_REP):
            seq = (rng_mc.random(n) < P_TRUE).astype(np.int8)
            t_naive, t_corr, null = naive_and_corrected(seq, k, N_PERM, rng_mc)
            if not np.isnan(t_naive):
                naive_vals.append(t_naive)
            if not np.isnan(t_corr):
                corr_vals.append(t_corr)
                lo, hi = np.percentile(null, [2.5, 97.5])
                covered.append(lo <= t_naive <= hi)
        naive_vals, corr_vals = np.array(naive_vals), np.array(corr_vals)
        mc_rows.append({
            'n': n, 'k': k,
            'naive bias': naive_vals.mean(), 'corrected bias': corr_vals.mean(),
            'naive var': naive_vals.var(), 'corrected var': corr_vals.var(),
            'naive MSE': (naive_vals ** 2).mean(), 'corrected MSE': (corr_vals ** 2).mean(),
            'interval coverage': np.mean(covered),
        })
mc_study = pd.DataFrame(mc_rows)
mc_study['MSE ratio'] = mc_study['corrected MSE'] / mc_study['naive MSE']
mc_study.round(5)
```

```python
fig, axes = plt.subplots(1, 3, figsize=(17, 5))
markers = {1: 'o', 2: 's', 3: '^'}
colours = {1: GREEN_DEEP, 2: GREEN_MID, 3: COPPER}

ax = axes[0]
for k in grid_k:
    s = mc_study[mc_study['k'] == k]
    ax.plot(s['n'], s['naive bias'], marker=markers[k], color=colours[k], lw=2, label=f'naive, k={k}')
    ax.plot(s['n'], s['corrected bias'], marker=markers[k], color=colours[k], lw=2,
            ls='--', alpha=0.65, label=f'corrected, k={k}')
ax.axhline(0, color=INK, lw=1.2)
ax.set_xscale('log'); ax.set_xlabel('Sequence length n'); ax.set_ylabel('Bias')
ax.set_title('(a) Bias')
ax.legend(frameon=False, fontsize=8, ncol=2)

ax = axes[1]
for k in grid_k:
    s = mc_study[mc_study['k'] == k]
    ax.plot(s['n'], s['naive MSE'], marker=markers[k], color=colours[k], lw=2, label=f'naive, k={k}')
    ax.plot(s['n'], s['corrected MSE'], marker=markers[k], color=colours[k], lw=2,
            ls='--', alpha=0.65, label=f'corrected, k={k}')
ax.set_xscale('log'); ax.set_yscale('log')
ax.set_xlabel('Sequence length n'); ax.set_ylabel('Mean squared error')
ax.set_title('(b) Mean squared error')
ax.legend(frameon=False, fontsize=8, ncol=2)

ax = axes[2]
for k in grid_k:
    s = mc_study[mc_study['k'] == k]
    ax.plot(s['n'], s['interval coverage'], marker=markers[k], color=colours[k], lw=2, label=f'k={k}')
ax.axhline(0.95, color=COPPER, ls='--', lw=1.6, label='nominal 95%')
ax.set_xscale('log'); ax.set_ylim(0.88, 1.005)
ax.set_xlabel('Sequence length n'); ax.set_ylabel('Coverage of the 95% permutation interval')
ax.set_title('(c) Interval coverage under the null')
ax.legend(frameon=False, fontsize=9)

fig.suptitle('Figure 11. Estimator properties under independence, p = 0.45',
             fontsize=14, color=GREEN_DEEP, fontweight='bold', y=1.03)
plt.tight_layout(); plt.show()
```

<div style="background:#f3f8f4;border-radius:12px;padding:20px 26px;border-left:5px solid #1c6b3f;font-family:Georgia,serif;line-height:1.85;color:#1a2e22;margin:14px 0;">
<h4 style="color:#14532d;margin:0 0 10px;">The correction removes the bias and does not cost variance</h4>

The naive statistic is biased downward throughout, and the size of the bias tracks the theory closely:
worst at short sequences with long required runs, negligible at long sequences with single-make streaks.
Subtracting each sequence's own permutation null mean removes essentially all of it, leaving biases that
scatter around zero at the resolution this number of replications supports.
<br><br>
The more useful result is that this costs nothing. A bias correction that inflated variance enough to
worsen mean squared error would be a poor trade, and that is not what happens. Corrected mean squared
error is below the naive value in eleven of the twelve grid cells, with the largest gains exactly where
the bias was largest: at 25 shots with a three-make streak it falls by about a fifth. The single
exception, long sequences with single-make streaks, is the cell where the naive bias was already
negligible, and there the two estimators are indistinguishable. There is no accuracy-for-unbiasedness
trade to agonise over here.
<br><br>
Coverage of the 95% permutation interval stays near nominal across the grid, ranging from 0.933 to 0.984
against a Monte Carlo standard error of about 0.009 at 600 replications. The pattern is not uniform: at
short sequences the intervals run conservative, which is what the lumpy permutation distribution of a
25-shot sequence should produce, while at the longest sequences coverage drifts a little under 0.95,
though by no more than roughly two Monte Carlo standard errors. Neither departure is large enough to
undermine the procedure, and the conservative direction at small samples is the safer one to err in.
<br><br>
For the rest of the notebook this settles the estimator question: the permutation-calibrated comparison
is used wherever a streak statistic is tested directly, and the naive comparison against zero is not used
at all except to demonstrate what it would have produced.

</div>

<a id="sec11"></a>
<div style="background:linear-gradient(135deg,#0d3320 0%,#14532d 55%,#2f7a4d 100%);padding:26px 32px;border-radius:14px;margin:34px 0 20px 0;box-shadow:0 6px 24px rgba(13,51,32,0.25);">
<div style="color:rgba(255,255,255,0.55);font-size:11px;letter-spacing:4px;text-transform:uppercase;margin-bottom:8px;">Section 11</div>
<h2 style="color:#ffffff;font-family:Georgia,serif;font-size:24px;margin:0;font-weight:700;">Permutation Inference</h2>
<p style="color:#d9cdb4;margin:6px 0 0 0;font-size:13.5px;">Testing real players against their own reshuffled sequences</p>
</div>


Section 10 established that the right benchmark for a streak statistic is not zero but the value the
statistic takes when the same shots are reordered at random. This section applies that benchmark to the
actual players.

The permutation keeps each player's outcomes within each game and reshuffles only their order. That
choice encodes a specific exchangeability assumption worth stating plainly: conditional on a player and
a game, the make/miss outcomes are treated as exchangeable, so any within-game ordering is equally
likely under the null. Permuting within game rather than across the whole season is deliberately
conservative. It holds constant everything that varies at game level, the opponent, the venue, whatever
shape the player turned up in that night, and so tests only for sequential dependence inside a game. A
player who genuinely has good nights and bad nights will not register as streaky under this null,
because good nights are preserved by construction.


<div style="background:linear-gradient(135deg,#2f7a4d 0%,#14532d 100%);border-radius:8px;padding:11px 20px;margin:24px 0 14px 0;">
<span style="color:#d9cdb4;font-family:Georgia,serif;font-size:12px;font-weight:700;letter-spacing:1px;">11.1</span>
<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">Player-by-player testing</span>
</div>

```python
def pooled_streak_stat(outcomes, breaks, k=3):
    """T_k pooled across the blocks of one player: P(make | previous k makes) - P(make).

    `breaks[i] = True` marks a shot with no observed predecessor, so the streak
    counter restarts there and that shot never enters the conditional numerator.
    """
    numer = denom = 0
    run = 0
    for i in range(len(outcomes)):
        if breaks[i]:
            run = 0
        elif run >= k:
            denom += 1
            numer += outcomes[i]
        run = run + 1 if outcomes[i] == 1 else 0
    if denom == 0:
        return np.nan, 0
    return numer / denom - outcomes.mean(), denom


def within_game_permutation(outcomes, game_codes, rng):
    """One reshuffle of a player's outcomes that stays inside each game.

    Sorting on (game code + uniform noise) permutes within game blocks and never
    moves a shot across a game boundary.
    """
    keys = game_codes * 2.0 + rng.random(len(outcomes))
    return outcomes[np.argsort(keys, kind='stable')]


K_STREAK = 3
N_PERM_PLAYER = 500
MIN_SHOTS = 200

rng_perm = np.random.default_rng(RANDOM_SEED)
player_rows = []
for pid, d in df.groupby('player_id'):
    if len(d) < MIN_SHOTS:
        continue
    y = d['FGM'].values
    brk = d['seq_break'].values
    gcodes = pd.factorize(d['GAME_ID'].values)[0].astype(np.float64)
    observed, n_eligible = pooled_streak_stat(y, brk, K_STREAK)
    if np.isnan(observed):
        continue
    null = np.array([pooled_streak_stat(within_game_permutation(y, gcodes, rng_perm),
                                         brk, K_STREAK)[0]
                     for _ in range(N_PERM_PLAYER)])
    null = null[~np.isnan(null)]
    player_rows.append({
        'player': d['player_name'].iloc[0],
        'shots': len(d),
        'eligible shots': n_eligible,
        'T observed': observed,
        'null mean': null.mean(),
        'T corrected': observed - null.mean(),
        'p (one-sided)': ((null >= observed).sum() + 1) / (len(null) + 1),
    })

player_perm = pd.DataFrame(player_rows)
print(f"Players analysed (at least {MIN_SHOTS} shots): {len(player_perm)}")
print(f"Permutations per player: {N_PERM_PLAYER}, streak length k = {K_STREAK}")
print(f"\nMean permutation null across players : {player_perm['null mean'].mean():+.5f}")
print(f"Mean naive statistic across players   : {player_perm['T observed'].mean():+.5f}")
print(f"Mean bias-corrected statistic         : {player_perm['T corrected'].mean():+.5f}")
```

<div style="background:linear-gradient(135deg,#f3f8f4,#e6f0e9);border:1px solid #7aa98c55;border-left:5px solid #1c6b3f;border-radius:0 12px 12px 0;padding:18px 24px;margin:14px 0;">
<div style="font-weight:700;color:#14532d;margin-bottom:10px;font-size:14px;">The bias is not just a theorem, it is visibly present in these players</div>
<div style="color:#1a2e22;font-size:13.5px;line-height:1.75;">
Averaged over players, the permutation null sits at roughly -0.021 rather than at zero. Nothing about
these reshuffled sequences contains any dependence at all, since the reshuffling destroyed it, so the
entire displacement is the streak-selection bias of Section 9 showing up in real data at realistic
sequence lengths.
<br><br>
That single number reframes the naive result. The average naive statistic is about -0.034, which read
against zero looks like a substantial cold hand. Read against the correct benchmark, roughly 60% of it is
an artefact of the estimator, and the corrected average is about -0.013.
</div>
</div>

```python
from statsmodels.stats.multitest import multipletests

reject, p_fdr, _, _ = multipletests(player_perm['p (one-sided)'], alpha=0.05, method='fdr_bh')
player_perm['p (FDR)'] = p_fdr
player_perm['significant after FDR'] = reject

n_raw = int((player_perm['p (one-sided)'] < 0.05).sum())
expected_by_chance = 0.05 * len(player_perm)
sd_by_chance = np.sqrt(len(player_perm) * 0.05 * 0.95)

print(f"Players with raw one-sided p < 0.05 : {n_raw} of {len(player_perm)}")
print(f"Expected under the null              : {expected_by_chance:.1f} "
      f"(sd {sd_by_chance:.1f}), so {n_raw} is "
      f"{(n_raw - expected_by_chance)/sd_by_chance:+.1f} sd from chance")
print(f"Players surviving Benjamini-Hochberg  : {int(reject.sum())}")

print("\nThe five smallest uncorrected p-values, with their effective sample sizes:")
print(player_perm.nsmallest(5, 'p (one-sided)')[
    ['player', 'shots', 'eligible shots', 'T observed', 'T corrected',
     'p (one-sided)', 'p (FDR)']].round(4).to_string(index=False))
```

<div style="background:#f3f8f4;border-radius:12px;padding:20px 26px;border-left:5px solid #1c6b3f;font-family:Georgia,serif;line-height:1.85;color:#1a2e22;margin:14px 0;">
<h4 style="color:#14532d;margin:0 0 10px;">Result: nothing survives, and the apparent leaders are an artefact of thin data</h4>

Eight of 248 players clear an uncorrected one-sided threshold of 0.05. Chance alone predicts 12.4, with a
standard deviation of 3.4, so eight is not merely unremarkable, it is slightly fewer than a pure null
would deliver. No player survives the Benjamini-Hochberg correction across the 248 tests.
<br><br>
The effective sample sizes explain why the individual test has so little to say. What matters is not how
many shots a player took but how many of those shots followed three consecutive makes, and the median
across these 248 players is 24. The player with the largest apparent effect in the table has exactly one
such shot. An estimate built on a single observation will happily report an enormous effect and a small
p-value, and neither means anything.
<br><br>
Panel (c) shows the p-values sitting above the diagonal across the whole range rather than on it: the
median is 0.58 where a uniform null would give 0.50, and a Kolmogorov-Smirnov test rejects uniformity.
The direction is conservative, meaning fewer small p-values than a continuous null would produce, and the
cause is discreteness. A player with two dozen eligible shots has a permutation distribution that takes
only a handful of distinct values, so the attainable p-values are coarse and the smallest reachable one
is bounded away from zero. This is a property of the test at these sample sizes, not evidence about
basketball, and it means the player-level scan is even less powerful than the raw count of eight
suggests.
<br><br>
All of this is a power problem rather than evidence of absence, and it is the specific reason Section 15
turns to a hierarchical model. Testing 248 players separately throws away the fact that they are all
basketball players; partial pooling uses the population to stabilise exactly the thin individual
estimates that are producing noise here.

</div>

```python
fig, axes = plt.subplots(1, 3, figsize=(17, 5))

ax = axes[0]
ax.scatter(player_perm['eligible shots'], player_perm['T observed'],
           s=30, alpha=0.6, color=GREEN, edgecolor=GREEN_DEEP, linewidth=0.4)
ax.axhline(0, color=INK, lw=1.2)
ax.set_xscale('log')
ax.set_xlabel('Shots following a 3+ make streak (log scale)')
ax.set_ylabel('Naive statistic $T_3$')
ax.set_title('(a) Extreme estimates come from thin data')

ax = axes[1]
ax.hist(player_perm['T observed'], bins=34, color=SAND, edgecolor=COPPER,
        alpha=0.9, label='naive')
ax.hist(player_perm['T corrected'], bins=34, color=GREEN_SOFT, edgecolor=GREEN_DEEP,
        alpha=0.7, label='bias corrected')
ax.axvline(0, color=INK, lw=1.4)
ax.axvline(player_perm['T observed'].mean(), color=COPPER, ls='--', lw=1.6)
ax.axvline(player_perm['T corrected'].mean(), color=GREEN_DEEP, ls='--', lw=1.6)
ax.set_xlabel('Statistic value'); ax.set_ylabel('Players')
ax.set_title('(b) Correction shifts the whole distribution right')
ax.legend(frameon=False, fontsize=9)

ax = axes[2]
observed_p = np.sort(player_perm['p (one-sided)'].values)
uniform_q = (np.arange(1, len(observed_p) + 1) - 0.5) / len(observed_p)
ks = stats.kstest(player_perm['p (one-sided)'].values, 'uniform')
print(f"Median p-value: {np.median(observed_p):.4f}  (uniform null would give 0.50)")
print(f"KS test against Uniform(0,1): D = {ks.statistic:.4f}, p = {ks.pvalue:.4g}")

ax.plot([0, 1], [0, 1], color=COPPER, ls='--', lw=1.6, label='uniform (pure null)')
ax.plot(uniform_q, observed_p, marker='o', ms=3, lw=1.4, color=GREEN_DEEP,
        label='observed p-values')
ax.set_xlabel('Expected quantile under the null'); ax.set_ylabel('Observed p-value')
ax.set_title('(c) p-values run conservative, not uniform')
ax.legend(frameon=False, fontsize=9)

fig.suptitle('Figure 12. Player-level permutation tests, k = 3',
             fontsize=14, color=GREEN_DEEP, fontweight='bold', y=1.03)
plt.tight_layout(); plt.show()
```

<div style="background:linear-gradient(135deg,#2f7a4d 0%,#14532d 100%);border-radius:8px;padding:11px 20px;margin:24px 0 14px 0;">
<span style="color:#d9cdb4;font-family:Georgia,serif;font-size:12px;font-weight:700;letter-spacing:1px;">11.2</span>
<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">A league-wide test, and why the benchmark moves</span>
</div>


Pooling every shot into one statistic buys far more precision than 248 separate tests. It also
introduces a composition problem that is worth making explicit, because it changes the direction of the
bias.

Two versions of the statistic are computed. The first subtracts the league-wide field-goal rate, the
second subtracts each shooter's own rate. They differ because the pool of shots following a three-make
streak is not a random sample of shooters: better shooters produce more streaks, so they are
over-represented in the conditional group. Subtracting a league average leaves that over-representation
in the statistic; subtracting each player's own average removes it.


```python
player_mean_map = df.groupby('player_id')['FGM'].transform('mean').values

def pooled_two_ways(outcomes, breaks, player_means, k=3):
    """Returns the league-centred and player-centred versions of the pooled statistic."""
    numer = denom = 0
    centred = 0.0
    run = 0
    for i in range(len(outcomes)):
        if breaks[i]:
            run = 0
        elif run >= k:
            denom += 1
            numer += outcomes[i]
            centred += outcomes[i] - player_means[i]
        run = run + 1 if outcomes[i] == 1 else 0
    return numer / denom - outcomes.mean(), centred / denom


all_y = df['FGM'].values
all_brk = df['seq_break'].values
all_gcodes = pd.factorize(df['player_id'].astype(str) + '_' + df['GAME_ID'].astype(str))[0].astype(np.float64)

obs_league, obs_player = pooled_two_ways(all_y, all_brk, player_mean_map, K_STREAK)

N_PERM_POOLED = 300
rng_pooled = np.random.default_rng(RANDOM_SEED)
null_league = np.empty(N_PERM_POOLED)
null_player = np.empty(N_PERM_POOLED)
for b in range(N_PERM_POOLED):
    permuted = within_game_permutation(all_y, all_gcodes, rng_pooled)
    null_league[b], null_player[b] = pooled_two_ways(permuted, all_brk, player_mean_map, K_STREAK)

pooled_results = pd.DataFrame([
    {'Centring': 'league-wide rate', 'Observed': obs_league,
     'Null mean': null_league.mean(), 'Null sd': null_league.std(),
     'Bias corrected': obs_league - null_league.mean(),
     'p (two-sided)': 2 * min((null_league >= obs_league).mean(), (null_league <= obs_league).mean())},
    {'Centring': "each shooter's own rate", 'Observed': obs_player,
     'Null mean': null_player.mean(), 'Null sd': null_player.std(),
     'Bias corrected': obs_player - null_player.mean(),
     'p (two-sided)': 2 * min((null_player >= obs_player).mean(), (null_player <= obs_player).mean())},
])
pooled_results.round(5)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14.5, 5))
for ax, null, obs, title in [
    (axes[0], null_league, obs_league, 'Centred on the league-wide rate'),
    (axes[1], null_player, obs_player, "Centred on each shooter's own rate")]:
    ax.hist(null, bins=34, color=SAND, edgecolor=COPPER, alpha=0.9)
    ax.axvline(0, color=GREY, lw=1.6, ls=':', label='zero')
    ax.axvline(null.mean(), color=COPPER, lw=2, label=f'null mean {null.mean():+.4f}')
    ax.axvline(obs, color=GREEN_DEEP, lw=2.4, label=f'observed {obs:+.4f}')
    ax.set_xlabel('Pooled statistic $T_3$'); ax.set_ylabel('Permutations')
    ax.set_title(title)
    ax.legend(frameon=False, fontsize=9)

fig.suptitle('Figure 13. The null benchmark is not zero, and its sign depends on how the statistic is centred',
             fontsize=13.5, color=GREEN_DEEP, fontweight='bold', y=1.03)
plt.tight_layout(); plt.show()
```

<div style="background:#f3f8f4;border-radius:12px;padding:20px 26px;border-left:5px solid #1c6b3f;font-family:Georgia,serif;line-height:1.85;color:#1a2e22;margin:14px 0;">
<h4 style="color:#14532d;margin:0 0 10px;">Result: two different nulls, one consistent corrected answer</h4>

The two centrings produce null benchmarks on opposite sides of zero. Centred on the league rate the null
sits at about +0.007, because the conditional group over-represents good shooters and that raises the
expected conditional rate. Centred on each shooter's own rate the null sits at about -0.010, which is the
streak-selection bias of Section 9 with the composition effect stripped out. Neither is zero, and a
naive analysis comparing to zero would draw a different conclusion depending on which version it happened
to use.
<br><br>
The corrected estimates agree: roughly -0.019 under league centring and -0.017 under player centring.
Both fall outside their permutation null bands. Two statistics with benchmarks of opposite sign
converging on the same corrected value is exactly what a working bias correction should do, and it is
more convincing than either result on its own.
<br><br>
So the league-wide answer is a small negative effect: shots taken after three consecutive makes go in
roughly 1.7 to 1.9 percentage points less often than reshuffling the same shots would predict. This is
statistically clear and substantively small, and the obvious candidate explanation is not psychological
but tactical, since Section 7.1 showed shot difficulty moves outcomes by far more than this. Whether
players simply attempt harder shots after making a few is the question Section 13 takes up, and it cannot
be answered by any permutation of outcomes alone, because permutation holds the shots fixed and only
reorders them.

</div>

<a id="sec12"></a>
<div style="background:linear-gradient(135deg,#0d3320 0%,#14532d 55%,#2f7a4d 100%);padding:26px 32px;border-radius:14px;margin:34px 0 20px 0;box-shadow:0 6px 24px rgba(13,51,32,0.25);">
<div style="color:rgba(255,255,255,0.55);font-size:11px;letter-spacing:4px;text-transform:uppercase;margin-bottom:8px;">Section 12</div>
<h2 style="color:#ffffff;font-family:Georgia,serif;font-size:24px;margin:0;font-weight:700;">Baseline Regression Models</h2>
<p style="color:#d9cdb4;margin:6px 0 0 0;font-size:13.5px;">Moving from a ratio statistic to a model, and checking the model inherits the same bias</p>
</div>


The permutation analysis treats each player's sequence separately and throws away the ability to adjust
for anything. A regression fixes that, at the cost of imposing functional form. The baseline is


$$Y_{it} \sim \text{Bernoulli}(p_{it}), \qquad
\operatorname{logit}(p_{it}) = \alpha_i + \beta K_{it}$$


where $\alpha_i$ is a player-specific intercept, $K_{it}$ is the number of consecutive makes immediately
preceding shot $t$, capped at three so that the handful of very long streaks do not drive the slope, and
$\beta$ is the parameter of interest. A positive $\beta$ means higher make probability after a longer
run, which is a statement of conditional association and nothing more.

Fitting 281 player intercepts alongside the covariates makes a dense design matrix impractical, so the
player indicators are held sparse and the model is fitted by iteratively reweighted least squares. The
implementation is checked against `statsmodels` on a subsample small enough for the dense fit to run.


```python
import scipy.sparse as sp
from scipy.stats import norm

def sparse_logit(y, X_dense, player_codes, n_players, max_iter=40, tol=1e-9):
    """Logistic regression with many group indicators, fitted by IRLS.

    The player indicators live in a sparse block so the design never has to be
    materialised densely.  Returns coefficients and model-based standard errors
    for the dense block only, which is where every parameter of interest sits.
    """
    D = sp.csr_matrix((np.ones(len(y)), (np.arange(len(y)), player_codes)),
                      shape=(len(y), n_players))
    X = sp.hstack([sp.csr_matrix(X_dense), D], format='csr')
    beta = np.zeros(X.shape[1])
    for _ in range(max_iter):
        eta = X @ beta
        mu = 1 / (1 + np.exp(-eta))
        W = np.clip(mu * (1 - mu), 1e-9, None)
        z = eta + (y - mu) / W
        XtW = X.T.multiply(W)
        new = np.linalg.solve((XtW @ X).toarray(), XtW @ z)
        if np.max(np.abs(new - beta)) < tol:
            beta = new
            break
        beta = new
    eta = X @ beta
    mu = 1 / (1 + np.exp(-eta))
    W = np.clip(mu * (1 - mu), 1e-9, None)
    cov = np.linalg.inv((X.T.multiply(W) @ X).toarray())
    k = X_dense.shape[1]
    return beta[:k], np.sqrt(np.diag(cov))[:k]


# Validate the implementation against statsmodels where the dense fit is feasible
import statsmodels.api as sm
check_players = pd.factorize(analysis['player_id'])[1][:15]
chk = analysis[analysis['player_id'].isin(check_players)]
chk_codes = pd.factorize(chk['player_id'])[0]
chk_K = np.minimum(chk['prev_streak'].values, 3).astype(float)

dense_X = np.column_stack([chk_K, pd.get_dummies(chk_codes, dtype=float).values])
sm_fit = sm.Logit(chk['FGM'].values, dense_X).fit(disp=False)
mine_b, mine_se = sparse_logit(chk['FGM'].values, chk_K[:, None], chk_codes, len(check_players))

print("Implementation check on 15 players:")
print(f"  statsmodels : beta = {sm_fit.params[0]:+.8f}   se = {sm_fit.bse[0]:.8f}")
print(f"  sparse IRLS : beta = {mine_b[0]:+.8f}   se = {mine_se[0]:.8f}")
print(f"  agreement to {abs(sm_fit.params[0]-mine_b[0]):.2e}")
```

```python
player_codes, player_index = pd.factorize(analysis['player_id'])
n_players = len(player_index)
y_obs = analysis['FGM'].values
K_obs = np.minimum(analysis['prev_streak'].values, 3).astype(float)

beta_m1, se_m1 = sparse_logit(y_obs, K_obs[:, None], player_codes, n_players)
z_m1 = beta_m1[0] / se_m1[0]
print(f"Model 1: logit(p) = alpha_i + beta * K")
print(f"  beta = {beta_m1[0]:+.5f}   se = {se_m1[0]:.5f}   z = {z_m1:+.2f}   "
      f"p = {2*(1-norm.cdf(abs(z_m1))):.3g}")
print(f"  odds ratio per extra made shot in the run: {np.exp(beta_m1[0]):.4f}")
```

<div style="background:linear-gradient(135deg,#2f7a4d 0%,#14532d 100%);border-radius:8px;padding:11px 20px;margin:24px 0 14px 0;">
<span style="color:#d9cdb4;font-family:Georgia,serif;font-size:12px;font-weight:700;letter-spacing:1px;">12.1</span>
<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">Does the regression coefficient inherit the selection bias?</span>
</div>


Section 9 proved the ratio estimator is biased under independence. Whether that carries into a
regression coefficient is a separate question, and assuming either answer would be careless. The direct
way to find out is to simulate: generate shots that are independent by construction, using each player's
own observed rate and their real game structure, then refit the identical model and see where the
coefficient lands.


```python
def simulate_independent(rng):
    """Independent shooting at each player's own rate, on the real game structure."""
    return (rng.random(len(df)) < player_mean_map).astype(np.int8)


def streak_from_outcomes(outcomes, breaks):
    st = np.zeros(len(outcomes), dtype=np.int32)
    run = 0
    for i in range(len(outcomes)):
        if breaks[i]:
            run = 0
        st[i] = run
        run = run + 1 if outcomes[i] == 1 else 0
    return st


keep_mask = ~df['seq_break'].values
codes_full = pd.factorize(df['player_id'])[0]

N_NULL = 200
rng_null = np.random.default_rng(RANDOM_SEED)
null_beta_m1 = []
for _ in range(N_NULL):
    y_sim = simulate_independent(rng_null)
    K_sim = np.minimum(streak_from_outcomes(y_sim, df['seq_break'].values), 3).astype(float)
    b, _ = sparse_logit(y_sim[keep_mask], K_sim[keep_mask][:, None],
                        codes_full[keep_mask], n_players)
    null_beta_m1.append(b[0])
null_beta_m1 = np.array(null_beta_m1)

print(f"Coefficient under {N_NULL} independent-shooting replications:")
print(f"  mean = {null_beta_m1.mean():+.5f}   sd = {null_beta_m1.std():.5f}")
print(f"  95% null band = [{np.percentile(null_beta_m1,2.5):+.5f}, "
      f"{np.percentile(null_beta_m1,97.5):+.5f}]")
print(f"  Shapiro-Wilk test of null normality: p = {stats.shapiro(null_beta_m1).pvalue:.3f}")
print(f"\nObserved coefficient      : {beta_m1[0]:+.5f}")
print(f"Bias-corrected coefficient : {beta_m1[0] - null_beta_m1.mean():+.5f}")
```

<div style="background:#f3f8f4;border-radius:12px;padding:20px 26px;border-left:5px solid #1c6b3f;font-family:Georgia,serif;line-height:1.85;color:#1a2e22;margin:14px 0;">
<h4 style="color:#14532d;margin:0 0 10px;">The regression inherits the bias, attenuated but not gone</h4>

Under shooting that is independent by construction, the streak coefficient does not centre on zero. It
centres near -0.006. The selection mechanism of Section 9 therefore survives the move from a ratio
statistic to a generalised linear model, in the same direction and with a smaller magnitude.
<br><br>
The reason is that the regressor is built from the outcome's own history within a finite sequence. A
player who has just made three shots has, by construction, used up three makes from a limited supply, so
the shot being predicted is drawn from a slightly depleted pool. Pooling across many sequences dilutes
this substantially compared with averaging per-sequence ratios, which is why the regression bias
(-0.006) is far smaller than the ratio-estimator bias measured in Section 11 (-0.021), but dilution is
not elimination.
<br><br>
This matters practically. The naive coefficient is -0.051 with a p-value around 1e-14 against zero, which
looks overwhelming. Roughly 12% of it is an artefact of the estimator. Every regression coefficient
reported from here on is therefore accompanied by its own simulated null, and significance is judged
against that null rather than against zero.

</div>

<a id="sec13"></a>
<div style="background:linear-gradient(135deg,#0d3320 0%,#14532d 55%,#2f7a4d 100%);padding:26px 32px;border-radius:14px;margin:34px 0 20px 0;box-shadow:0 6px 24px rgba(13,51,32,0.25);">
<div style="color:rgba(255,255,255,0.55);font-size:11px;letter-spacing:4px;text-transform:uppercase;margin-bottom:8px;">Section 13</div>
<h2 style="color:#ffffff;font-family:Georgia,serif;font-size:24px;margin:0;font-weight:700;">Shot-Difficulty Adjustment</h2>
<p style="color:#d9cdb4;margin:6px 0 0 0;font-size:13.5px;">The rival explanation, tested directly rather than assumed away</p>
</div>


Section 7.1 showed shot difficulty moves make probability by tens of percentage points, far more than
any streak effect under discussion. If players change what shots they take after making a few, that
alone could produce everything observed so far. Bocskocsky, Ezekowitz and Stein (2014) raised exactly
this point against the classical literature.

The hypothesis is testable on its own terms before any outcome model is fitted, by asking whether the
shots attempted after a streak are physically different from the shots the same player takes otherwise.
Within-player deviations are used throughout so that differences between players cannot generate the
result.


```python
mechanism_rows = []
for var, label in [('SHOT_DIST', 'Shot distance (ft)'),
                   ('CLOSE_DEF_DIST', 'Closest defender (ft)'),
                   ('TOUCH_TIME', 'Touch time (s)'),
                   ('DRIBBLES', 'Dribbles')]:
    d = analysis[[var, 'player_id']].copy()
    d['K'] = np.minimum(analysis['prev_streak'].values, 3)
    d = d.dropna()
    d['deviation'] = d[var] - d.groupby('player_id')[var].transform('mean')
    means = d.groupby('K')['deviation'].mean()
    trend = sm.OLS(d['deviation'].values,
                   sm.add_constant(d['K'].values.astype(float))).fit(cov_type='HC1')
    mechanism_rows.append({
        'Variable': label,
        'K=0': means.get(0, np.nan), 'K=1': means.get(1, np.nan),
        'K=2': means.get(2, np.nan), 'K=3': means.get(3, np.nan),
        'Trend per unit K': trend.params[1],
        'p-value': trend.pvalues[1],
    })
shot_selection = pd.DataFrame(mechanism_rows)
shot_selection.round(4)
```

```python
fig, axes = plt.subplots(1, 4, figsize=(18, 4.6))
panels = [('SHOT_DIST', 'Shot distance (ft)', COPPER),
          ('CLOSE_DEF_DIST', 'Closest defender (ft)', GREEN),
          ('TOUCH_TIME', 'Touch time (s)', GREEN_MID),
          ('DRIBBLES', 'Dribbles', GREEN_DEEP)]
for ax, (var, label, colour) in zip(axes, panels):
    d = analysis[[var, 'player_id']].copy()
    d['K'] = np.minimum(analysis['prev_streak'].values, 3)
    d = d.dropna()
    d['deviation'] = d[var] - d.groupby('player_id')[var].transform('mean')
    g = d.groupby('K')['deviation'].agg(['mean', 'sem'])
    ax.errorbar(g.index, g['mean'], yerr=1.96 * g['sem'], marker='o', markersize=8,
                lw=2.2, color=colour, capsize=4)
    ax.axhline(0, color=INK, lw=1.2)
    ax.set_xticks([0, 1, 2, 3])
    ax.set_xlabel('K, preceding makes'); ax.set_ylabel(f'{label}\ndeviation from own average')
    ax.set_title(label, fontsize=11.5)

fig.suptitle('Figure 14. What changes about the shots players take after making some',
             fontsize=14, color=GREEN_DEEP, fontweight='bold', y=1.05)
plt.tight_layout(); plt.show()
```

<div style="background:#f3f8f4;border-radius:12px;padding:20px 26px;border-left:5px solid #1c6b3f;font-family:Georgia,serif;line-height:1.85;color:#1a2e22;margin:14px 0;">
<h4 style="color:#14532d;margin:0 0 10px;">Players really do take different shots after making shots</h4>

Every one of the four measures moves, and the largest by far is distance. Relative to their own average,
players shoot from about half a foot closer than usual when they have missed, and more than a foot
farther than usual after three consecutive makes, a swing of roughly 1.7 feet with a p-value beyond any
conventional threshold. They also hold the ball longer and dribble more.
<br><br>
Defender distance moves in the opposite direction, increasing slightly after makes, which would make
shots marginally easier rather than harder. It is dwarfed by the distance effect: Section 7.1's
difficulty plane shows that moving a shot from sixteen feet to twenty-four costs far more than gaining
a foot of separation returns.
<br><br>
This is a behavioural pattern in shot selection, not a claim about what players are thinking. Whether it
reflects confidence, defensive adjustment, or a coach's tolerance for a shooter who is going well cannot
be settled by this data, and the analysis does not need to settle it. What matters for the statistics is
that the shots being compared are not alike, so any comparison that ignores difficulty is comparing
different things.

</div>

<div style="background:linear-gradient(135deg,#2f7a4d 0%,#14532d 100%);border-radius:8px;padding:11px 20px;margin:24px 0 14px 0;">
<span style="color:#d9cdb4;font-family:Georgia,serif;font-size:12px;font-weight:700;letter-spacing:1px;">13.1</span>
<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">The adjusted model</span>
</div>

$$\operatorname{logit}(p_{it}) = \alpha_i + \beta K_{it} + \gamma^\top X_{it}$$


$X_{it}$ holds only variables genuinely present in the file: shot distance and defender distance entered
as bin indicators rather than linear terms, since Section 7.1 showed neither relationship is monotone;
touch time, dribbles, two versus three point, home or away, shot clock with its missingness indicator,
period, and time remaining. Nothing derived from how the game finished is included, for the reason given
in Section 5.1.


```python
def difficulty_design(d):
    """Flexible controls for shot difficulty and game context."""
    dist_bins = pd.get_dummies(pd.cut(d['SHOT_DIST'], [0, 4, 8, 12, 16, 20, 22, 24, 27, 60],
                                       right=False), drop_first=True, dtype=float)
    def_bins = pd.get_dummies(pd.cut(d['CLOSE_DEF_DIST'], [0, 2, 3, 4, 5, 6, 8, 60],
                                      right=False), drop_first=True, dtype=float)
    period_bins = pd.get_dummies(d['PERIOD'].clip(upper=5), drop_first=True, dtype=float)
    rest = np.column_stack([
        d['TOUCH_TIME'].fillna(d['TOUCH_TIME'].median()).values,
        d['DRIBBLES'].values,
        (d['PTS_TYPE'] == 3).astype(float).values,
        (d['LOCATION'] == 'H').astype(float).values,
        d['SHOT_CLOCK'].fillna(d['SHOT_CLOCK'].median()).values,
        d['shot_clock_missing'].values,
        d['game_clock_sec'].values / 60.0,
        d['touch_time_invalid'].values,
    ])
    return np.column_stack([dist_bins.values, def_bins.values, period_bins.values, rest])


X_difficulty = difficulty_design(analysis)
beta_m2, se_m2 = sparse_logit(y_obs, np.column_stack([K_obs, X_difficulty]),
                              player_codes, n_players)
z_m2 = beta_m2[0] / se_m2[0]

print(f"Model 2: logit(p) = alpha_i + beta * K + gamma' X      ({X_difficulty.shape[1]} controls)")
print(f"  beta = {beta_m2[0]:+.5f}   se = {se_m2[0]:.5f}   z = {z_m2:+.2f}   "
      f"p = {2*(1-norm.cdf(abs(z_m2))):.3g}")
print(f"\nAttenuation from adding difficulty controls: "
      f"{beta_m1[0]:+.5f} -> {beta_m2[0]:+.5f}  "
      f"({100*(1-beta_m2[0]/beta_m1[0]):.0f}% of the coefficient removed)")

# The adjusted model needs its own null, for the same reason Model 1 did
X_difficulty_full = difficulty_design(df)
null_beta_m2 = []
rng_null2 = np.random.default_rng(RANDOM_SEED + 1)
for _ in range(N_NULL):
    y_sim = simulate_independent(rng_null2)
    K_sim = np.minimum(streak_from_outcomes(y_sim, df['seq_break'].values), 3).astype(float)
    b, _ = sparse_logit(y_sim[keep_mask],
                        np.column_stack([K_sim[keep_mask], X_difficulty_full[keep_mask]]),
                        codes_full[keep_mask], n_players)
    null_beta_m2.append(b[0])
null_beta_m2 = np.array(null_beta_m2)

def null_pvalues(observed, null_draws):
    """Empirical and normal-approximation two-sided p-values against a simulated null.

    With a few hundred replications the empirical tail count is coarse, its finest
    resolution being 2/B, so the normal approximation is reported alongside it and
    is justified by a normality check on the null draws themselves.
    """
    empirical = 2 * min((null_draws >= observed).mean(), (null_draws <= observed).mean())
    z = (observed - null_draws.mean()) / null_draws.std()
    return empirical, 2 * (1 - norm.cdf(abs(z))), z


progression = []
for name, beta, se, null in [
        ('M1: player effects + streak', beta_m1[0], se_m1[0], null_beta_m1),
        ('M2: + shot difficulty', beta_m2[0], se_m2[0], null_beta_m2)]:
    p_emp, p_norm, z = null_pvalues(beta, null)
    progression.append({
        'Model': name, 'beta': beta, 'se': se,
        'null mean': null.mean(), 'null sd': null.std(),
        'bias corrected': beta - null.mean(), 'z vs null': z,
        'p empirical': p_emp, 'p normal approx': p_norm,
    })
progression = pd.DataFrame(progression)
print(f"Null draws per model: {N_NULL}  (finest empirical p-value resolution: {2/N_NULL:.4f})")
print(f"Shapiro-Wilk normality of the M2 null: p = {stats.shapiro(null_beta_m2).pvalue:.3f}")
progression.round(5)
```

```python
fig, ax = plt.subplots(figsize=(11, 5.6))
labels = ['Naive\n(vs zero)', 'Player effects\n+ streak', '+ shot difficulty',
          '+ difficulty,\nbias corrected']
values = [beta_m1[0], beta_m1[0] - null_beta_m1.mean(), beta_m2[0],
          beta_m2[0] - null_beta_m2.mean()]
errors = [1.96 * se_m1[0], 1.96 * se_m1[0], 1.96 * se_m2[0], 1.96 * se_m2[0]]
colours = [COPPER, COPPER_LT, GREEN_MID, GREEN_DEEP]

ax.errorbar(range(4), values, yerr=errors, fmt='o', markersize=11, capsize=6,
            lw=2.2, color=INK, zorder=2)
for i, (v, c) in enumerate(zip(values, colours)):
    ax.scatter([i], [v], s=190, color=c, zorder=3, edgecolor=INK, linewidth=1.2)
    ax.annotate(f'{v:+.4f}', (i, v), textcoords='offset points', xytext=(0, 18),
                ha='center', fontsize=10.5, color=INK, fontweight='bold')
ax.axhline(0, color=GREY, lw=1.6, ls='--')
ax.set_xticks(range(4)); ax.set_xticklabels(labels, fontsize=10)
ax.set_ylabel(r'Streak coefficient $\beta$ (log-odds per made shot in the run)')
ax.set_title('Figure 15. Where the apparent cold hand goes once each explanation is accounted for',
             fontsize=13, color=GREEN_DEEP, fontweight='bold')
plt.tight_layout(); plt.show()
```

<div style="background:#f3f8f4;border-radius:12px;padding:20px 26px;border-left:5px solid #1c6b3f;font-family:Georgia,serif;line-height:1.85;color:#1a2e22;margin:14px 0;">
<h4 style="color:#14532d;margin:0 0 10px;">Result: the apparent effect is almost entirely estimator bias plus shot selection</h4>

Adding flexible difficulty controls removes roughly two thirds of the streak coefficient, taking it from
-0.051 to -0.017. Subtracting each model's own simulated null removes most of what is left. The fully
adjusted, bias-corrected coefficient is about -0.011, which sits roughly 1.6 null standard deviations
from zero and carries a two-sided p-value near 0.11 by normal approximation and 0.14 empirically. Neither
clears a conventional threshold.
<br><br>
The two p-values are reported together deliberately. An earlier version of this analysis used only 60
null replications, where the finest resolution an empirical tail count can reach is 2/60, and repeated
runs of the identical procedure returned p-values as far apart as 0.03 and 0.13. That spread was Monte
Carlo noise in the p-value rather than anything about basketball, and it is a reminder that a
simulation-based p-value near a decision boundary needs enough replications to be stable before it is
quoted. With 200 draws and a null that passes a normality check, the normal approximation is the more
reliable of the two and the empirical count serves as a sanity check on it.
<br><br>
The progression matters more than any single number in it. What began as a coefficient with a p-value
near 1e-14 against zero, apparently overwhelming evidence of a cold hand, decomposes into two mundane
parts: an estimator biased downward by construction, and players who genuinely take harder shots after
making some. Once both are accounted for, what remains is small and statistically indistinguishable from
no effect at all.
<br><br>
Two limits on this conclusion belong with it. This is a pooled coefficient, so it constrains the average
player and says little about whether some individuals behave differently, which Section 15 addresses. And
the controls capture observable difficulty only: defensive scheme, fatigue, and shot-clock pressure
beyond the raw reading are unmeasured, so residual confounding from unobserved difficulty remains
possible in either direction.

</div>

<a id="sec13b"></a>
<div style="background:linear-gradient(135deg,#0d3320 0%,#14532d 55%,#2f7a4d 100%);padding:26px 32px;border-radius:14px;margin:34px 0 20px 0;box-shadow:0 6px 24px rgba(13,51,32,0.25);">
<div style="color:rgba(255,255,255,0.55);font-size:11px;letter-spacing:4px;text-transform:uppercase;margin-bottom:8px;">Section 13b</div>
<h2 style="color:#ffffff;font-family:Georgia,serif;font-size:24px;margin:0;font-weight:700;">Model Diagnostics</h2>
<p style="color:#d9cdb4;margin:6px 0 0 0;font-size:13.5px;">Checking the adjusted model actually fits, and discovering that it does not</p>
</div>


Section 13 reported a coefficient without first asking whether the model producing it describes the data.
For a binary outcome the informative check is a binned residual plot: group shots by their predicted
probability, and within each group compare the average prediction with the average outcome. A
well-specified model should show residuals scattered around zero, with roughly one bin in twenty falling
outside two standard errors by chance.


```python
def binned_residuals(y_true, p_hat, n_bins=20):
    """Group by predicted probability and compare average prediction with average outcome."""
    edges = np.quantile(p_hat, np.linspace(0, 1, n_bins + 1))
    edges[-1] += 1e-9
    bin_id = np.clip(np.digitize(p_hat, edges[1:-1]), 0, n_bins - 1)
    out = (pd.DataFrame({'bin': bin_id, 'p': p_hat, 'resid': y_true - p_hat})
             .groupby('bin').agg(predicted=('p', 'mean'), mean_residual=('resid', 'mean'),
                                 n=('resid', 'size')))
    out['se'] = np.sqrt(out['predicted'] * (1 - out['predicted']) / out['n'])
    out['z'] = out['mean_residual'] / out['se']
    return out


def fitted_probabilities(y_vec, X_dense, codes_vec, n_groups):
    """Refit and return fitted probabilities as well as coefficients."""
    D = sp.csr_matrix((np.ones(len(y_vec)), (np.arange(len(y_vec)), codes_vec)),
                      shape=(len(y_vec), n_groups))
    X = sp.hstack([sp.csr_matrix(X_dense), D], format='csr')
    beta = np.zeros(X.shape[1])
    for _ in range(40):
        eta = X @ beta
        mu = 1 / (1 + np.exp(-eta))
        W = np.clip(mu * (1 - mu), 1e-9, None)
        XtW = X.T.multiply(W)
        new = np.linalg.solve((XtW @ X).toarray(), XtW @ (eta + (y_vec - mu) / W))
        if np.max(np.abs(new - beta)) < 1e-9:
            beta = new
            break
        beta = new
    return beta, 1 / (1 + np.exp(-(X @ beta)))


beta_primary, p_primary = fitted_probabilities(
    y_obs, np.column_stack([K_obs, X_difficulty]), player_codes, n_players)
resid_primary = binned_residuals(y_obs, p_primary)

n_bad = int((resid_primary['z'].abs() > 2).sum())
print(f"Binned residuals for the Section 13 specification, 20 bins:")
print(f"  bins outside +/- 2 standard errors : {n_bad} of 20  (about 1 expected by chance)")
print(f"  largest average residual           : {resid_primary['mean_residual'].abs().max():.4f}")
print(f"  at the lowest predicted decile     : {resid_primary['mean_residual'].iloc[0]:+.4f}")
print(f"  at the highest predicted decile    : {resid_primary['mean_residual'].iloc[-1]:+.4f}")
```

<div style="background:#fdf5ee;border-radius:14px;padding:22px 28px;border:1.5px solid #d9884f;font-family:Georgia,serif;line-height:1.8;color:#6b3410;margin:14px 0;">
<h4 style="color:#b4531f;margin:0 0 10px;font-size:1.0rem;">The adjusted model does not fit, and the pattern says why</h4>

Ten of twenty bins fall outside two standard errors, against about one expected by chance. The
misfit is not random: the model under-predicts at both ends of the probability range and over-predicts
through the middle. At the highest predicted decile, where the model expects roughly a 75% make rate, the
actual rate is nearly seven percentage points higher.
<br><br>
A smile-shaped residual pattern of this kind says the linear predictor's range is too compressed, which
happens when a covariate relationship is more extreme than the specification allows. The obvious
candidate is an interaction the model does not contain. Shot distance and defender distance are entered
as separate sets of indicators, so the model assumes the value of a foot of separation is the same on a
layup as on a long three. Section 7.1's difficulty plane already suggested otherwise.

</div>

```python
def difficulty_design_flexible(subset, n_dist, n_def, with_interaction):
    """Difficulty controls with tunable resolution and an optional full interaction."""
    dist_bin = pd.qcut(subset['SHOT_DIST'], n_dist, duplicates='drop')
    def_bin = pd.qcut(subset['CLOSE_DEF_DIST'], n_def, duplicates='drop')
    blocks = [pd.get_dummies(dist_bin, drop_first=True, dtype=float).values,
              pd.get_dummies(def_bin, drop_first=True, dtype=float).values,
              pd.get_dummies(subset['PERIOD'].clip(upper=5), drop_first=True, dtype=float).values,
              subset['TOUCH_TIME'].fillna(subset['TOUCH_TIME'].median()).values[:, None],
              subset['DRIBBLES'].values[:, None],
              (subset['PTS_TYPE'] == 3).astype(float).values[:, None],
              (subset['LOCATION'] == 'H').astype(float).values[:, None],
              subset['SHOT_CLOCK'].fillna(subset['SHOT_CLOCK'].median()).values[:, None],
              subset['shot_clock_missing'].values[:, None],
              subset['game_clock_sec'].values[:, None] / 60.0,
              subset['touch_time_invalid'].values[:, None]]
    if with_interaction:
        blocks.append(pd.get_dummies(dist_bin.astype(str) + '|' + def_bin.astype(str),
                                      drop_first=True, dtype=float).values)
    X = np.column_stack(blocks).astype(np.float32)
    keep_col = X.std(0) > 1e-12
    if with_interaction:
        # drop interaction cells with too few shots to estimate; they add columns, not fit
        keep_col &= (X.sum(0) >= 40) | (np.arange(X.shape[1]) < X.shape[1] - 1)
    return X[:, keep_col]


spec_rows = []
for label, n_d, n_f, inter in [
        ('primary: 9 x 7 bins, no interaction', 9, 7, False),
        ('finer: 20 x 12 bins, no interaction', 20, 12, False),
        ('finest: 40 x 20 bins, no interaction', 40, 20, False),
        ('20 x 12 bins with full interaction', 20, 12, True)]:
    Xd = difficulty_design_flexible(analysis, n_d, n_f, inter)
    b, p_hat = fitted_probabilities(y_obs, np.column_stack([K_obs, Xd]),
                                     player_codes, n_players)
    r = binned_residuals(y_obs, p_hat)
    spec_rows.append({'Difficulty specification': label,
                      'controls': Xd.shape[1],
                      'beta (streak)': b[0],
                      'misfitting bins (of 20)': int((r['z'].abs() > 2).sum())})
    del Xd, p_hat
calibration_search = pd.DataFrame(spec_rows)
calibration_search
```

<div style="background:#f3f8f4;border-radius:12px;padding:20px 26px;border-left:5px solid #1c6b3f;font-family:Georgia,serif;line-height:1.85;color:#1a2e22;margin:14px 0;">
<h4 style="color:#14532d;margin:0 0 10px;">Only the interaction fixes the fit, and it moves the coefficient</h4>

Adding resolution does not help: going from 9 by 7 bins to 40 by 20 leaves the calibration as bad or
slightly worse, because finer marginal bins cannot represent a relationship between the two variables.
Adding the interaction fixes it, taking the misfitting bins from ten down to three, which is what chance
alone produces.
<br><br>
The streak coefficient moves with it, from -0.017 in the primary specification to -0.020 in the
well-calibrated one. That is not a large change, but it goes in the direction that matters and it
arrives with a better-fitting model, so it cannot be dismissed as a specification whim. It needs its own
null, for exactly the reason every other coefficient in this notebook has needed one.

</div>

```python
X_interaction = difficulty_design_flexible(analysis, 20, 12, True)
beta_inter, _ = fitted_probabilities(y_obs, np.column_stack([K_obs, X_interaction]),
                                      player_codes, n_players)

X_interaction_full = difficulty_design_flexible(df, 20, 12, True)
N_NULL_INTER = 60
rng_inter = np.random.default_rng(RANDOM_SEED + 5)
null_inter = []
for _ in range(N_NULL_INTER):
    y_sim = simulate_independent(rng_inter)
    K_sim = np.minimum(streak_from_outcomes(y_sim, df['seq_break'].values), 3).astype(float)
    b, _ = fitted_probabilities(y_sim[keep_mask],
                                 np.column_stack([K_sim[keep_mask], X_interaction_full[keep_mask]]),
                                 codes_full[keep_mask], n_players)
    null_inter.append(b[0])
null_inter = np.array(null_inter)
del X_interaction_full
import gc; gc.collect()

z_inter = (beta_inter[0] - null_inter.mean()) / null_inter.std()
specification_comparison = pd.DataFrame([
    {'Specification': 'primary (Section 13)', 'beta': beta_m2[0],
     'null mean': null_beta_m2.mean(), 'corrected': beta_m2[0] - null_beta_m2.mean(),
     'z vs null': (beta_m2[0] - null_beta_m2.mean()) / null_beta_m2.std(),
     'p': 2 * (1 - norm.cdf(abs((beta_m2[0] - null_beta_m2.mean()) / null_beta_m2.std()))),
     'misfitting bins': int((resid_primary['z'].abs() > 2).sum())},
    {'Specification': 'well-calibrated (with interaction)', 'beta': beta_inter[0],
     'null mean': null_inter.mean(), 'corrected': beta_inter[0] - null_inter.mean(),
     'z vs null': z_inter, 'p': 2 * (1 - norm.cdf(abs(z_inter))),
     'misfitting bins': int(calibration_search['misfitting bins (of 20)'].iloc[-1])},
])
specification_comparison.round(5)
```

```python
fig, axes = plt.subplots(1, 3, figsize=(17.5, 5.1))

ax = axes[0]
ax.errorbar(resid_primary['predicted'], resid_primary['mean_residual'],
            yerr=2 * resid_primary['se'], fmt='o', markersize=7, color=COPPER,
            capsize=3, lw=1.6)
ax.axhline(0, color=INK, lw=1.3)
ax.set_xlabel('Predicted probability (bin mean)')
ax.set_ylabel('Average residual')
ax.set_title('(a) Primary specification: a smile, not noise')

ax = axes[1]
_, p_inter_fit = fitted_probabilities(y_obs, np.column_stack([K_obs, X_interaction]),
                                       player_codes, n_players)
resid_inter = binned_residuals(y_obs, p_inter_fit)
ax.errorbar(resid_inter['predicted'], resid_inter['mean_residual'],
            yerr=2 * resid_inter['se'], fmt='o', markersize=7, color=GREEN_DEEP,
            capsize=3, lw=1.6)
ax.axhline(0, color=INK, lw=1.3)
ax.set_ylim(axes[0].get_ylim())
ax.set_xlabel('Predicted probability (bin mean)')
ax.set_ylabel('Average residual')
ax.set_title('(b) With the interaction: calibration restored')

ax = axes[2]
for label, vals, colour, marker in [
        ('primary', null_beta_m2, COPPER, 'o'),
        ('well-calibrated', null_inter, GREEN_DEEP, 's')]:
    ax.hist(vals, bins=26, alpha=0.55, color=colour, edgecolor=INK, label=f'null, {label}')
ax.axvline(beta_m2[0], color=COPPER, lw=2.4, ls='--',
           label=f'observed, primary ({beta_m2[0]:+.4f})')
ax.axvline(beta_inter[0], color=GREEN_DEEP, lw=2.4,
           label=f'observed, calibrated ({beta_inter[0]:+.4f})')
ax.set_xlabel(r'Streak coefficient $\beta$'); ax.set_ylabel('Null draws')
ax.set_title('(c) Each estimate against its own null')
ax.legend(frameon=False, fontsize=8.5)

fig.suptitle('Figure 16. Fixing the fit changes the answer',
             fontsize=14, color=GREEN_DEEP, fontweight='bold', y=1.04)
plt.tight_layout(); plt.show()
```

```python
# Does any single player drive the estimate?  Refit dropping one player at a time.
volumes = pd.Series(player_codes).value_counts()
targets = volumes.index[:40]
influence_rows = []
for p_code in targets:
    keep_p = player_codes != p_code
    recoded = pd.factorize(player_codes[keep_p])[0]
    b_without, _ = fitted_probabilities(
        y_obs[keep_p], np.column_stack([K_obs[keep_p], X_difficulty[keep_p]]),
        recoded, recoded.max() + 1)
    influence_rows.append({'player_code': int(p_code), 'shots': int(volumes[p_code]),
                           'beta without them': b_without[0],
                           'shift': b_without[0] - beta_m2[0]})
influence = pd.DataFrame(influence_rows)
print(f"Leave-one-player-out over the {len(influence)} highest-volume players:")
print(f"  full-sample beta      : {beta_m2[0]:+.5f}")
print(f"  range when one is out : [{influence['beta without them'].min():+.5f}, "
      f"{influence['beta without them'].max():+.5f}]")
print(f"  largest single shift  : {influence['shift'].abs().max():.5f}")
print(f"  (for comparison, the standard error is {se_m2[0]:.5f})")
```

<div style="background:#f3f8f4;border-radius:12px;padding:20px 26px;border-left:5px solid #1c6b3f;font-family:Georgia,serif;line-height:1.85;color:#1a2e22;margin:14px 0;">
<h4 style="color:#14532d;margin:0 0 10px;">Influence: no single player matters, but the specification does</h4>

Dropping any one of the forty highest-volume players, several of whom took around a thousand shots, moves
the coefficient by at most 0.0016, a quarter of its standard error. The estimate is not resting on any
individual, and there is no influential-observation problem to report.
<br><br>
The specification is a different matter, and this is the section's real finding. Against its own
simulated null the well-calibrated model gives a corrected coefficient of about -0.014 with a two-sided
p-value near 0.039, where the primary specification gave about -0.011 at 0.096. The model that fits the
data better finds a small negative effect that clears conventional significance, and the model reported
in Section 13 does not.
<br><br>
Both estimates agree on sign and are close in magnitude, so this is not a reversal. It is a case where
significance sits on the wrong side of a threshold depending on how the controls are built, and the
better-fitting build is the one that crosses it. The honest summary is that the fully adjusted effect is
small, negative, and of borderline significance, with the balance tipping toward significance as the
difficulty controls are improved. Section 22 states it that way rather than picking whichever
specification gives a cleaner headline.

</div>

<div style="background:#fdf5ee;border-radius:14px;padding:22px 28px;border:1.5px solid #d9884f;font-family:Georgia,serif;line-height:1.8;color:#6b3410;margin:14px 0;">
<h4 style="color:#b4531f;margin:0 0 10px;font-size:1.0rem;">What a residual negative effect most plausibly is</h4>

A small negative coefficient surviving better difficulty controls is not evidence of a cold hand in any
psychological sense, and the more mundane reading is more likely. Even the interaction specification
measures difficulty through distance, defender proximity, touch time and dribbles alone. It has nothing
on defender identity or quality, on whether the shot was contested at the release point, on fatigue
within a possession, or on how the defence reorganised after a run of makes. Section 13 established that
players do change their shot selection after makes; the controls capture part of that change, and the
residual negative is consistent with capturing only part.
<br><br>
The direction of the remaining bias is therefore predictable: better difficulty measurement would be
expected to shrink this coefficient further toward zero, not away from it. That is a testable prediction
for anyone with richer tracking data than a 2014-15 shot log.

</div>

<a id="sec14"></a>
<div style="background:linear-gradient(135deg,#0d3320 0%,#14532d 55%,#2f7a4d 100%);padding:26px 32px;border-radius:14px;margin:34px 0 20px 0;box-shadow:0 6px 24px rgba(13,51,32,0.25);">
<div style="color:rgba(255,255,255,0.55);font-size:11px;letter-spacing:4px;text-transform:uppercase;margin-bottom:8px;">Section 14</div>
<h2 style="color:#ffffff;font-family:Georgia,serif;font-size:24px;margin:0;font-weight:700;">Nonlinear Streak Effects</h2>
<p style="color:#d9cdb4;margin:6px 0 0 0;font-size:13.5px;">Whether a single slope is the right way to describe a run of makes</p>
</div>


Writing the streak term as $\beta K$ imposes a specific claim: that going from one preceding make to two
changes the log-odds by exactly as much as going from two to three. There is no reason that has to be
true. A hot hand could plausibly switch on at a threshold, accumulate and then saturate, or not exist at
all. Three specifications are compared, each carrying the full difficulty controls from Section 13.


$$\text{linear:}\quad f(K)=\beta K
\qquad
\text{quadratic:}\quad f(K)=\beta_1 K + \beta_2 K^2
\qquad
\text{categorical:}\quad f(K)=\sum_{j=1}^{3}\beta_j \mathbb{1}\{K=j\}$$

```python
def fit_with_loglik(y, X_dense, player_codes, n_players, max_iter=40, tol=1e-9):
    """Same IRLS as Section 12, returning the log-likelihood and parameter count too."""
    D = sp.csr_matrix((np.ones(len(y)), (np.arange(len(y)), player_codes)),
                      shape=(len(y), n_players))
    X = sp.hstack([sp.csr_matrix(X_dense), D], format='csr')
    beta = np.zeros(X.shape[1])
    for _ in range(max_iter):
        eta = X @ beta
        mu = 1 / (1 + np.exp(-eta))
        W = np.clip(mu * (1 - mu), 1e-9, None)
        XtW = X.T.multiply(W)
        new = np.linalg.solve((XtW @ X).toarray(), XtW @ (eta + (y - mu) / W))
        if np.max(np.abs(new - beta)) < tol:
            beta = new
            break
        beta = new
    eta = X @ beta
    mu = np.clip(1 / (1 + np.exp(-eta)), 1e-12, 1 - 1e-12)
    loglik = float(np.sum(y * np.log(mu) + (1 - y) * np.log(1 - mu)))
    W = np.clip(mu * (1 - mu), 1e-9, None)
    se = np.sqrt(np.diag(np.linalg.inv((X.T.multiply(W) @ X).toarray())))
    k = X_dense.shape[1]
    return beta[:k], se[:k], loglik, X.shape[1]


K_cat = np.column_stack([(np.minimum(analysis['prev_streak'].values, 3) == j).astype(float)
                         for j in (1, 2, 3)])

shape_specs = {
    'K omitted entirely': X_difficulty,
    'linear in K': np.column_stack([K_obs, X_difficulty]),
    'quadratic in K': np.column_stack([K_obs, K_obs ** 2, X_difficulty]),
    'categorical in K': np.column_stack([K_cat, X_difficulty]),
}

shape_fits = {}
shape_rows = []
for name, Xd in shape_specs.items():
    b, s, ll, n_par = fit_with_loglik(y_obs, Xd, player_codes, n_players)
    shape_fits[name] = (b, s, ll, n_par)
    shape_rows.append({'Specification': name, 'log-likelihood': ll, 'parameters': n_par,
                       'AIC': -2 * ll + 2 * n_par,
                       'BIC': -2 * ll + n_par * np.log(len(y_obs))})
shape_comparison = pd.DataFrame(shape_rows)
shape_comparison.round(2)
```

```python
ll_lin, par_lin = shape_fits['linear in K'][2], shape_fits['linear in K'][3]
lr_rows = []
for name in ['quadratic in K', 'categorical in K']:
    ll_alt, par_alt = shape_fits[name][2], shape_fits[name][3]
    lr = 2 * (ll_alt - ll_lin)
    ddf = par_alt - par_lin
    lr_rows.append({'Alternative': name, 'LR statistic': lr, 'df': ddf,
                    'p-value': 1 - stats.chi2.cdf(lr, ddf)})
lr_tests = pd.DataFrame(lr_rows)
print("Likelihood-ratio tests against the linear specification:")
print(lr_tests.round(4).to_string(index=False))

b_cat_all, se_cat_all = shape_fits['categorical in K'][0], shape_fits['categorical in K'][1]
# the first three dense columns are the streak indicators; the rest are difficulty controls
b_cat, se_cat = b_cat_all[:3], se_cat_all[:3]
print("\nCategorical coefficients, reference category K = 0 (compared against zero):")
for j, (bb, ss) in enumerate(zip(b_cat, se_cat), start=1):
    print(f"  K={j}: {bb:+.5f}  se={ss:.5f}  z={bb/ss:+.2f}  p={2*(1-norm.cdf(abs(bb/ss))):.4f}")
```

<div style="background:linear-gradient(135deg,#f3f8f4,#e6f0e9);border:1px solid #7aa98c55;border-left:5px solid #1c6b3f;border-radius:0 12px 12px 0;padding:18px 24px;margin:14px 0;">
<div style="font-weight:700;color:#14532d;margin-bottom:10px;font-size:14px;">A tension worth taking seriously rather than resolving conveniently</div>
<div style="color:#1a2e22;font-size:13.5px;line-height:1.75;">
The fit statistics say the linear form is adequate. It has the lowest AIC of the three specifications
containing a streak term, and neither the quadratic nor the categorical alternative improves the
likelihood by more than chance would give, with likelihood-ratio p-values of 0.22 and 0.35. BIC, which
penalises parameters more heavily, goes further and prefers dropping the streak variable altogether.
<br><br>
Read individually, though, the categorical coefficients look different. K=1 and K=2 are flatly null,
while K=3 carries a coefficient of -0.073 with a z-statistic near -2.9. That looks like a threshold
effect appearing only after three consecutive makes.
<br><br>
Both readings cannot be right, and the individual coefficient is the less trustworthy of the two. It is
compared against zero, which Section 12.1 established is the wrong benchmark, and it is one of three
coefficients inspected after the joint test had already declined to reject the simpler model. The next
cell gives it the same treatment every other coefficient in this notebook has received.
</div>
</div>

```python
def categorical_streak_design(outcomes):
    K = np.minimum(streak_from_outcomes(outcomes, df['seq_break'].values), 3)
    return np.column_stack([(K == j).astype(float) for j in (1, 2, 3)])


N_NULL_CAT = 150
rng_cat = np.random.default_rng(RANDOM_SEED + 2)
null_cat = []
for _ in range(N_NULL_CAT):
    y_sim = simulate_independent(rng_cat)
    Xd = np.column_stack([categorical_streak_design(y_sim)[keep_mask], X_difficulty_full[keep_mask]])
    b, _, _, _ = fit_with_loglik(y_sim[keep_mask], Xd, codes_full[keep_mask], n_players)
    null_cat.append(b[:3])
null_cat = np.array(null_cat)

cat_rows = []
for j in range(3):
    z = (b_cat[j] - null_cat[:, j].mean()) / null_cat[:, j].std()
    cat_rows.append({
        'Category': f'K = {j+1}', 'Observed': b_cat[j],
        'Null mean': null_cat[:, j].mean(), 'Null sd': null_cat[:, j].std(),
        'Bias corrected': b_cat[j] - null_cat[:, j].mean(),
        'z vs null': z, 'p (uncorrected)': 2 * (1 - norm.cdf(abs(z))),
    })
categorical_vs_null = pd.DataFrame(cat_rows)
categorical_vs_null['p (Bonferroni, 3 tests)'] = np.minimum(
    categorical_vs_null['p (uncorrected)'] * 3, 1.0)
categorical_vs_null.round(5)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14.5, 5.2))

ax = axes[0]
xs = np.array([1, 2, 3])
ax.errorbar(xs - 0.06, b_cat, yerr=1.96 * se_cat, marker='o', markersize=9, lw=0,
            elinewidth=2, capsize=5, color=COPPER, label='against zero')
corrected = categorical_vs_null['Bias corrected'].values
corr_se = categorical_vs_null['Null sd'].values
ax.errorbar(xs + 0.06, corrected, yerr=1.96 * corr_se, marker='s', markersize=9, lw=0,
            elinewidth=2, capsize=5, color=GREEN_DEEP, label='against simulated null')
ax.axhline(0, color=INK, lw=1.3)
ax.set_xticks(xs); ax.set_xlabel('K, preceding makes (reference K = 0)')
ax.set_ylabel('Log-odds coefficient')
ax.set_title('(a) Categorical streak effects, two benchmarks')
ax.legend(frameon=False, fontsize=9.5)

ax = axes[1]
order = ['K omitted entirely', 'linear in K', 'quadratic in K', 'categorical in K']
aic = [shape_comparison.set_index('Specification').loc[o, 'AIC'] for o in order]
bic = [shape_comparison.set_index('Specification').loc[o, 'BIC'] for o in order]
idx = np.arange(len(order)); w = 0.38
ax.bar(idx - w/2, np.array(aic) - min(aic), w, color=GREEN, edgecolor=GREEN_DEEP, label='AIC')
ax.bar(idx + w/2, np.array(bic) - min(bic), w, color=SAND, edgecolor=COPPER, label='BIC')
ax.set_xticks(idx)
ax.set_xticklabels([o.replace(' in K', '').replace(' entirely', '') for o in order],
                   rotation=18, fontsize=9.5)
ax.set_ylabel('Difference from the best value')
ax.set_title('(b) Information criteria disagree about the streak term')
ax.legend(frameon=False, fontsize=9.5)

fig.suptitle('Figure 17. Is the streak effect linear, threshold-shaped, or absent?',
             fontsize=14, color=GREEN_DEEP, fontweight='bold', y=1.03)
plt.tight_layout(); plt.show()
```

<div style="background:#f3f8f4;border-radius:12px;padding:20px 26px;border-left:5px solid #1c6b3f;font-family:Georgia,serif;line-height:1.85;color:#1a2e22;margin:14px 0;">
<h4 style="color:#14532d;margin:0 0 10px;">Result: no reliable evidence of a nonlinear shape, and a cautionary tale about the one that looked real</h4>

Against its own simulated null, the K=3 coefficient corrects from -0.073 to -0.053, with a z-statistic of
about -2.1 and an uncorrected p-value near 0.037. K=1 and K=2 remain flatly null after correction. So the
threshold pattern does not disappear entirely once the benchmark is fixed, but it weakens, and three
further considerations keep it from counting as a finding.
<br><br>
It is one of three coefficients tested in this specification, and a Bonferroni adjustment across those
three lifts it above 0.10. It sits inside a specification that a joint likelihood-ratio test had already
declined to prefer over the simpler linear form. And it was noticed by scanning individual coefficients
after that joint test came back negative, which is the sequence of steps most likely to manufacture a
false positive.
<br><br>
Reported honestly, the position is this. The data are consistent with a linear streak term, consistent
with no streak term at all by BIC, and contain one coefficient that would look interesting in isolation
but does not survive the multiplicity its own specification implies. A study designed in advance to test
the three-make threshold specifically, on data not already used to notice it, would be the way to settle
the question. This notebook cannot settle it and does not claim to.

</div>

<a id="sec15"></a>
<div style="background:linear-gradient(135deg,#0d3320 0%,#14532d 55%,#2f7a4d 100%);padding:26px 32px;border-radius:14px;margin:34px 0 20px 0;box-shadow:0 6px 24px rgba(13,51,32,0.25);">
<div style="color:rgba(255,255,255,0.55);font-size:11px;letter-spacing:4px;text-transform:uppercase;margin-bottom:8px;">Section 15</div>
<h2 style="color:#ffffff;font-family:Georgia,serif;font-size:24px;margin:0;font-weight:700;">Hierarchical Bayesian Model</h2>
<p style="color:#d9cdb4;margin:6px 0 0 0;font-size:13.5px;">Letting players differ, and finding out how much of the answer the prior decides</p>
</div>


Section 11 tested 248 players separately and found nothing, but the effective sample sizes made that
almost inevitable: the median player had 24 shots following a three-make streak, and the most extreme
apparent effect came from a player with one. Section 13 pooled everyone into a single coefficient, which
is precise but assumes every player behaves identically.

A hierarchical model sits between those two. Each player gets their own streak slope, and the slopes are
themselves modelled as draws from a population distribution:


$$\operatorname{logit}(p_{it}) = \alpha_i + \beta_i K_{it} + \gamma^\top X_{it},
\qquad
\alpha_i \sim N(\mu_\alpha, \sigma_\alpha^2), \qquad
\beta_i \sim N(\mu_\beta, \sigma_\beta^2)$$


$\mu_\beta$ is the average streak effect across players and $\sigma_\beta$ is how much players differ
from each other, which is the quantity Section 11 could not estimate at all. A player with thin data has
their slope pulled toward the population mean, and one with plenty of data keeps their own estimate.
That is partial pooling, and it is the right response to a set of estimates whose noise scales with how
few eligible shots each player happened to take.


<div style="background:linear-gradient(135deg,#2f7a4d 0%,#14532d 100%);border-radius:8px;padding:11px 20px;margin:24px 0 14px 0;">
<span style="color:#d9cdb4;font-family:Georgia,serif;font-size:12px;font-weight:700;letter-spacing:1px;">15.1</span>
<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">Estimation by Polya-Gamma augmentation</span>
</div>


Gradient-based samplers struggle here: 281 player intercepts and 281 player slopes on 113,646
observations made a standard NUTS fit impractical in testing. A better route exists for logistic models
specifically. Polson, Scott and Windle (2013) showed the logistic likelihood is a scale mixture of
normals under a Polya-Gamma mixing distribution,


$$\frac{(e^{\eta})^{a}}{(1+e^{\eta})^{b}}
= 2^{-b} e^{\kappa \eta}\, \mathbb{E}_{\omega}\!\left[e^{-\omega \eta^{2}/2}\right],
\qquad \omega \sim PG(b, 0), \quad \kappa = a - b/2$$


so conditioning on the latent $\omega$ makes every coefficient block exactly Gaussian, with a
closed-form draw and no gradients or tuning. Drawing $PG(1,c)$ uses its infinite convolution
representation


$$\omega \;=\; \frac{1}{2\pi^{2}} \sum_{k=1}^{\infty}
\frac{g_k}{(k-\tfrac12)^{2} + c^{2}/(4\pi^{2})}, \qquad g_k \sim \text{Exp}(1)$$


truncated at a finite number of terms. Truncation drops a strictly positive remainder and therefore
biases every draw downward, so the implementation adds the omitted tail back analytically using
$\int_K^\infty \mathrm{d}x /((x-\tfrac12)^2 + a^2)$ with $a = c/2\pi$. With that correction, forty terms
reproduce the exact mean $\tanh(c/2)/(2c)$ to within about 0.08%, better than a hundred terms without it
and roughly twice as fast.


```python
def sample_polya_gamma(c, rng, n_terms=40):
    """Draw PG(1, c) elementwise: truncated convolution plus an analytic tail term."""
    c = np.asarray(c, dtype=np.float64)
    n = c.shape[0]
    k = np.arange(1, n_terms + 1, dtype=np.float64)
    a_sq = (c ** 2) / (4.0 * np.pi ** 2)
    denom = (k - 0.5) ** 2 + a_sq[:, None]
    truncated = (rng.exponential(size=(n, n_terms)) / denom).sum(axis=1) / (2.0 * np.pi ** 2)
    a = np.sqrt(a_sq)
    tail = np.where(a > 1e-8,
                    (1.0 / np.maximum(a, 1e-12)) *
                    (np.pi / 2 - np.arctan((n_terms - 0.5) / np.maximum(a, 1e-12))),
                    1.0 / (n_terms - 0.5))
    return truncated + tail / (2.0 * np.pi ** 2)


# Verify the sampler against the known mean of PG(1, c) before relying on it
check_rng = np.random.default_rng(RANDOM_SEED)
print("Polya-Gamma sampler check against the exact mean tanh(c/2)/(2c):")
for c_val in [0.0, 1.0, 2.0, 4.0]:
    drawn = sample_polya_gamma(np.full(200_000, c_val), check_rng).mean()
    exact = 0.25 if c_val == 0 else np.tanh(c_val / 2) / (2 * c_val)
    print(f"  c = {c_val:.1f} : sampled {drawn:.6f}   exact {exact:.6f}   "
          f"relative error {abs(drawn-exact)/exact:.3%}")
```

```python
def hierarchical_logit_gibbs(y, K_streak, player_idx, X, n_iter=600, burn=200,
                             seed=0, variance_prior='half-cauchy'):
    """Gibbs sampler for the hierarchical logistic model, via Polya-Gamma augmentation.

    variance_prior:
      'half-cauchy'   half-Cauchy(0,1) on each SD, written as its conjugate scale mixture
                      sd^2 | xi ~ InvGamma(1/2, 1/xi),  xi ~ InvGamma(1/2, 1)
      'inverse-gamma' the conventional InvGamma(2,1) placed on the variance directly
    """
    rng = np.random.default_rng(seed)
    n, n_players, p_fixed = len(y), player_idx.max() + 1, X.shape[1]
    kappa = y - 0.5

    a = np.zeros(n_players); b = np.zeros(n_players); gamma = np.zeros(p_fixed)
    mu_a, sd_a, mu_b, sd_b = 0.0, 1.0, 0.0, 0.25
    xi_a = xi_b = 1.0
    prior_sd_hyper, prior_sd_gamma = 10.0, 5.0
    Xg = X @ gamma

    keep = {k: [] for k in ('mu_a', 'sd_a', 'mu_b', 'sd_b', 'b', 'gamma')}
    for it in range(n_iter):
        eta = a[player_idx] + b[player_idx] * K_streak + Xg
        omega = sample_polya_gamma(eta, rng)

        # player-level (a_i, b_i) drawn jointly from their exact 2x2 Gaussian conditional
        z = kappa / omega - Xg
        s_w   = np.bincount(player_idx, weights=omega, minlength=n_players)
        s_wk  = np.bincount(player_idx, weights=omega * K_streak, minlength=n_players)
        s_wkk = np.bincount(player_idx, weights=omega * K_streak ** 2, minlength=n_players)
        s_wz  = np.bincount(player_idx, weights=omega * z, minlength=n_players)
        s_wkz = np.bincount(player_idx, weights=omega * K_streak * z, minlength=n_players)
        pa, pb = 1.0 / sd_a ** 2, 1.0 / sd_b ** 2
        A11, A12, A22 = s_w + pa, s_wk, s_wkk + pb
        r1, r2 = s_wz + pa * mu_a, s_wkz + pb * mu_b
        det = np.where(np.abs(A11 * A22 - A12 ** 2) < 1e-10, 1e-10, A11 * A22 - A12 ** 2)
        m1 = (A22 * r1 - A12 * r2) / det
        m2 = (A11 * r2 - A12 * r1) / det
        v11, v12, v22 = A22 / det, -A12 / det, A11 / det
        L11 = np.sqrt(np.maximum(v11, 1e-12))
        L21 = v12 / L11
        L22 = np.sqrt(np.maximum(v22 - L21 ** 2, 1e-12))
        e1, e2 = rng.standard_normal(n_players), rng.standard_normal(n_players)
        a = m1 + L11 * e1
        b = m2 + L21 * e1 + L22 * e2

        # shared covariate effects
        z2 = kappa / omega - (a[player_idx] + b[player_idx] * K_streak)
        XtW = X * omega[:, None]
        A_mat = XtW.T @ X + np.eye(p_fixed) / prior_sd_gamma ** 2
        L = np.linalg.cholesky(A_mat)
        gamma = np.linalg.solve(A_mat, XtW.T @ z2) + np.linalg.solve(L.T, rng.standard_normal(p_fixed))
        Xg = X @ gamma

        # hyper-means
        for which in ('a', 'b'):
            vals, sd_cur = (a, sd_a) if which == 'a' else (b, sd_b)
            prec = n_players / sd_cur ** 2 + 1 / prior_sd_hyper ** 2
            draw = vals.sum() / sd_cur ** 2 / prec + rng.standard_normal() / np.sqrt(prec)
            if which == 'a':
                mu_a = draw
            else:
                mu_b = draw

        # variance components, under whichever prior is being used
        ssq_a, ssq_b = ((a - mu_a) ** 2).sum(), ((b - mu_b) ** 2).sum()
        if variance_prior == 'half-cauchy':
            var_a = 1.0 / rng.gamma(0.5 * (n_players + 1), 1.0 / (1.0 / xi_a + 0.5 * ssq_a))
            xi_a = 1.0 / rng.gamma(1.0, 1.0 / (1.0 + 1.0 / var_a))
            var_b = 1.0 / rng.gamma(0.5 * (n_players + 1), 1.0 / (1.0 / xi_b + 0.5 * ssq_b))
            xi_b = 1.0 / rng.gamma(1.0, 1.0 / (1.0 + 1.0 / var_b))
        else:
            var_a = 1.0 / rng.gamma(2.0 + n_players / 2, 1.0 / (1.0 + 0.5 * ssq_a))
            var_b = 1.0 / rng.gamma(2.0 + n_players / 2, 1.0 / (1.0 + 0.5 * ssq_b))
        sd_a, sd_b = np.sqrt(var_a), np.sqrt(var_b)

        if it >= burn:
            keep['mu_a'].append(mu_a); keep['sd_a'].append(sd_a)
            keep['mu_b'].append(mu_b); keep['sd_b'].append(sd_b)
            keep['b'].append(b.copy()); keep['gamma'].append(gamma.copy())
    return {k: np.array(v) for k, v in keep.items()}


X_hier = (X_difficulty - X_difficulty.mean(0)) / np.where(X_difficulty.std(0) > 0,
                                                           X_difficulty.std(0), 1)
draws_hc = hierarchical_logit_gibbs(y_obs, K_obs, player_codes, X_hier,
                                    n_iter=600, burn=200, seed=RANDOM_SEED,
                                    variance_prior='half-cauchy')

def summarise(d, names=('mu_a', 'sd_a', 'mu_b', 'sd_b')):
    rows = []
    for nm in names:
        s = d[nm]
        rows.append({'parameter': nm, 'posterior mean': s.mean(), 'posterior sd': s.std(),
                     '2.5%': np.percentile(s, 2.5), '97.5%': np.percentile(s, 97.5)})
    return pd.DataFrame(rows)

hier_summary = summarise(draws_hc)
print(hier_summary.round(5).to_string(index=False))
print(f"\nP(mu_beta > 0 | data)     = {(draws_hc['mu_b'] > 0).mean():.4f}")
print(f"P(sigma_beta > 0.05 | data) = {(draws_hc['sd_b'] > 0.05).mean():.4f}")
```

<div style="background:#f3f8f4;border-radius:12px;padding:20px 26px;border-left:5px solid #1c6b3f;font-family:Georgia,serif;line-height:1.85;color:#1a2e22;margin:14px 0;">
<h4 style="color:#14532d;margin:0 0 10px;">Result: a small negative average effect, and very little difference between players</h4>

The population mean streak effect is -0.013 with a 95% credible interval of roughly [-0.027, -0.001],
close to the frequentist estimate of -0.017 from Section 13 and tighter than it, which is what borrowing
strength across players buys. The posterior probability that the average effect is positive is about 1%.
<br><br>
The more interesting parameter is $\sigma_\beta$. Its posterior mean is 0.023 with an interval of roughly
[0.013, 0.033], and the posterior probability that between-player variation exceeds 0.05 is
indistinguishable from zero. Players in this sample are not meaningfully different from one another in
how their shooting responds to a run of makes.
<br><br>
That is worth stating carefully, because it is the opposite of a common intuition. The popular version of
the hot hand allows that it might be real for some shooters and not others, and Section 11's scattered
player estimates are exactly what that would look like if it were true. Once those estimates are pooled
properly, the spread collapses: every player's posterior slope lies in a narrow band, and not one of the
281 has a 95% posterior probability of a positive effect. The apparent variation in Section 11 was noise
from small effective samples, which is precisely what partial pooling is designed to reveal.
<br><br>
One caveat carries over from Section 12.1. The credible interval for $\mu_\beta$ excludes zero, but zero
is not the right benchmark, since the same selection mechanism that biased the frequentist coefficient
biases this one. Shifting by the roughly -0.006 measured there would place the corrected effect near
-0.007 with an interval comfortably spanning zero, consistent with Section 13's conclusion.

</div>

<div style="background:linear-gradient(135deg,#2f7a4d 0%,#14532d 100%);border-radius:8px;padding:11px 20px;margin:24px 0 14px 0;">
<span style="color:#d9cdb4;font-family:Georgia,serif;font-size:12px;font-weight:700;letter-spacing:1px;">15.2</span>
<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">How much of the heterogeneity answer is the prior?</span>
</div>


$\sigma_\beta$ is a variance component near zero, and that is exactly the regime where prior choice stops
being innocuous. Gelman (2006) showed that inverse-gamma priors, the conventional default for variance
parameters, are not close to non-informative when the true variation is small, and instead push the
estimate away from zero. Whether that matters here is an empirical question with a cheap answer: refit
the identical model changing nothing but the prior.


```python
draws_ig = hierarchical_logit_gibbs(y_obs, K_obs, player_codes, X_hier,
                                    n_iter=600, burn=200, seed=RANDOM_SEED,
                                    variance_prior='inverse-gamma')

prior_comparison = pd.DataFrame([
    {'Prior on the SD': 'half-Cauchy(0, 1)',
     'sigma_beta mean': draws_hc['sd_b'].mean(),
     'sigma_beta 2.5%': np.percentile(draws_hc['sd_b'], 2.5),
     'sigma_beta 97.5%': np.percentile(draws_hc['sd_b'], 97.5),
     'mu_beta mean': draws_hc['mu_b'].mean()},
    {'Prior on the SD': 'InverseGamma(2, 1) on the variance',
     'sigma_beta mean': draws_ig['sd_b'].mean(),
     'sigma_beta 2.5%': np.percentile(draws_ig['sd_b'], 2.5),
     'sigma_beta 97.5%': np.percentile(draws_ig['sd_b'], 97.5),
     'mu_beta mean': draws_ig['mu_b'].mean()},
])
print(prior_comparison.round(5).to_string(index=False))
print(f"\nRatio of estimated between-player heterogeneity: "
      f"{draws_ig['sd_b'].mean()/draws_hc['sd_b'].mean():.1f}x")
print("The two credible intervals do not overlap.")
```

```python
# Which prior is right?  Simulate data with a KNOWN amount of heterogeneity and see
# which one recovers it.  Two scenarios: no heterogeneity at all, and a lot of it.
def simulate_hierarchical(n_players, shots_each, mu_b_true, sd_b_true, rng):
    a_true = rng.normal(-0.2, 0.25, n_players)
    b_true = rng.normal(mu_b_true, sd_b_true, n_players)
    idx, out, streak = [], [], []
    for i in range(n_players):
        run = 0
        for _ in range(shots_each):
            k = min(run, 3)
            made = rng.random() < 1 / (1 + np.exp(-(a_true[i] + b_true[i] * k)))
            idx.append(i); out.append(int(made)); streak.append(k)
            run = run + 1 if made else 0
    return np.array(idx), np.array(out, dtype=np.int8), np.array(streak, dtype=float)


recovery_rows = []
for label, true_sd in [('no heterogeneity (true sigma = 0)', 0.0001),
                       ('real heterogeneity (true sigma = 0.30)', 0.30)]:
    rng_sim = np.random.default_rng(RANDOM_SEED)
    idx_s, y_s, K_s = simulate_hierarchical(120, 350, 0.08, true_sd, rng_sim)
    X_s = np.zeros((len(y_s), 1))
    for prior in ['half-cauchy', 'inverse-gamma']:
        d = hierarchical_logit_gibbs(y_s, K_s, idx_s, X_s, n_iter=400, burn=150,
                                     seed=RANDOM_SEED, variance_prior=prior)
        recovery_rows.append({
            'Scenario': label, 'Prior': prior,
            'True sigma_beta': 0.0 if true_sd < 0.01 else true_sd,
            'Estimated': d['sd_b'].mean(),
            '2.5%': np.percentile(d['sd_b'], 2.5),
            '97.5%': np.percentile(d['sd_b'], 97.5),
        })
prior_recovery = pd.DataFrame(recovery_rows)
# The decisive comparison is not whether an interval contains exactly zero, which a
# posterior for a standard deviation essentially never does, but how the estimate from
# null data compares with the estimate from the real data.
prior_recovery['vs real-data estimate'] = np.where(
    prior_recovery['Prior'] == 'half-cauchy',
    prior_recovery['Estimated'] / draws_hc['sd_b'].mean(),
    prior_recovery['Estimated'] / draws_ig['sd_b'].mean())
prior_recovery.round(4)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(15, 5.2))

ax = axes[0]
ax.hist(draws_hc['sd_b'], bins=40, color=GREEN_SOFT, edgecolor=GREEN_DEEP, alpha=0.9,
        label=f"half-Cauchy  (mean {draws_hc['sd_b'].mean():.3f})")
ax.hist(draws_ig['sd_b'], bins=40, color=SAND, edgecolor=COPPER, alpha=0.85,
        label=f"inverse-gamma  (mean {draws_ig['sd_b'].mean():.3f})")
ax.set_xlabel(r'$\sigma_\beta$, between-player spread in the streak effect')
ax.set_ylabel('Posterior draws')
ax.set_title('(a) Same data, same model, different prior')
ax.legend(frameon=False, fontsize=9.5)

ax = axes[1]
scen = prior_recovery.copy()
xs = np.arange(len(scen))
cols = [GREEN_DEEP if p == 'half-cauchy' else COPPER for p in scen['Prior']]
ax.errorbar(xs, scen['Estimated'],
            yerr=[scen['Estimated'] - scen['2.5%'], scen['97.5%'] - scen['Estimated']],
            fmt='none', ecolor=INK, capsize=5, lw=1.8, zorder=2)
ax.scatter(xs, scen['Estimated'], s=150, c=cols, zorder=3, edgecolor=INK, linewidth=1.1)
for i, t in enumerate(scen['True sigma_beta']):
    ax.plot([i - 0.28, i + 0.28], [t, t], color=GREY, lw=2.4, ls='--',
            zorder=1, label='truth' if i == 0 else None)
ax.set_xticks(xs)
ax.set_xticklabels([f"{p}\ntrue={t:.2f}" for p, t in
                    zip(scen['Prior'], scen['True sigma_beta'])], fontsize=9)
ax.set_ylabel(r'Estimated $\sigma_\beta$')
ax.set_title('(b) Which prior recovers a known truth?')
ax.legend(frameon=False, fontsize=9.5)

fig.suptitle('Figure 18. A variance component near zero is decided largely by its prior',
             fontsize=14, color=GREEN_DEEP, fontweight='bold', y=1.03)
plt.tight_layout(); plt.show()
```

<div style="background:#fdf5ee;border-radius:14px;padding:22px 28px;border:1.5px solid #d9884f;font-family:Georgia,serif;line-height:1.8;color:#6b3410;margin:14px 0;">
<h4 style="color:#b4531f;margin:0 0 10px;font-size:1.0rem;">The prior changes the substantive conclusion, and the simulation shows which one to distrust</h4>

On the real data the two priors give estimates of between-player heterogeneity that differ by a factor of
six, 0.023 against 0.136, with credible intervals that do not overlap at all. These are not two readings
of an uncertain quantity. They are two different substantive claims: one says players respond to streaks
in essentially the same way, the other says they differ a great deal. A paper reporting either without
mentioning the other would look equally confident.
<br><br>
The recovery simulation is what breaks the tie, and it does so more sharply than a coverage check would.
Fed data generated with no between-player variation whatsoever, the half-Cauchy posterior returns 0.039
and the inverse-gamma returns 0.173. Neither interval contains exactly zero, which is unremarkable since
a posterior for a standard deviation is bounded below and rarely does. What matters is the magnitude, and
the decisive comparison is with the real-data estimates. The inverse-gamma produces a <em>larger</em>
heterogeneity estimate from data containing none (0.173) than it produced from the actual NBA shots
(0.136). Its real-data "finding" of substantial player heterogeneity is therefore worth nothing: the same
prior generates a bigger number from pure noise. The half-Cauchy's real-data estimate of 0.023 sits well
below its own null output of 0.039, which is the direction consistency requires.
<br><br>
When the simulated data really do contain substantial variation, both priors find it, at 0.29 and 0.32
against a truth of 0.30. So the inverse-gamma does not fail everywhere. It fails in one direction and in
one regime: it inflates variance components that are genuinely near zero, which is exactly the regime
this analysis occupies.
<br><br>
The implication reaches past this notebook. Any hierarchical hot-hand analysis using an inverse-gamma
prior on the slope variance is liable to report that some players are streakier than others, on data
containing no such difference. Gelman (2006) documents the general pathology; what this section adds is a
demonstration that it bites hard on this specific question, with the size of the distortion measured
against a known truth rather than asserted.

</div>

The Baum-Welch implementation used in Section 16 is defined here so the notebook is self-contained. It is written out rather than imported from a library for the reason given in that section: thousands of short sequences sharing one parameter set are far faster to fit with a recursion vectorised across sequences.

```python
"""
Two-state hidden Markov model with Bernoulli emissions, fitted by Baum-Welch.

Written directly rather than taken from a library because the structure here is very
specific: thousands of short sequences (one per player-game) that share one parameter
set. Padding them into a rectangular array lets the whole forward-backward recursion
run vectorised across sequences, which is roughly two orders of magnitude faster than
looping a general-purpose implementation over 10,000 sequences.

Model:
    Z_t in {0, 1},  P(Z_t = j | Z_{t-1} = i) = A[i, j]
    Y_t | Z_t = j   ~  Bernoulli(p[j])
States are returned ordered so that p[0] < p[1], making state 1 the higher-scoring one.
"""
import numpy as np


def pack_sequences(seq_list):
    """Pad a list of 0/1 sequences into (S, T_max) arrays plus a validity mask."""
    S = len(seq_list)
    T_max = max(len(s) for s in seq_list)
    Y = np.zeros((S, T_max), dtype=np.int8)
    mask = np.zeros((S, T_max), dtype=bool)
    for i, s in enumerate(seq_list):
        Y[i, :len(s)] = s
        mask[i, :len(s)] = True
    return Y, mask


def _emission_matrix(Y, mask, p):
    """B[s, t, j] = P(Y[s,t] | state j), and 1 at padded positions so they stay neutral."""
    B = np.empty(Y.shape + (2,))
    for j in (0, 1):
        B[:, :, j] = np.where(Y == 1, p[j], 1 - p[j])
    B[~mask] = 1.0
    return B


def forward_backward(Y, mask, pi, A, p):
    """Scaled forward-backward. Returns posteriors, pair posteriors and log-likelihood."""
    S, T = Y.shape
    B = _emission_matrix(Y, mask, p)
    alpha = np.zeros((S, T, 2))
    scale = np.ones((S, T))

    alpha[:, 0, :] = pi * B[:, 0, :]
    scale[:, 0] = np.maximum(alpha[:, 0, :].sum(1), 1e-300)
    alpha[:, 0, :] /= scale[:, 0, None]
    for t in range(1, T):
        pred = alpha[:, t - 1, :] @ A
        a = pred * B[:, t, :]
        # a padded position contributes nothing: carry the previous state forward
        a = np.where(mask[:, t, None], a, alpha[:, t - 1, :])
        scale[:, t] = np.maximum(a.sum(1), 1e-300)
        alpha[:, t, :] = a / scale[:, t, None]

    beta = np.ones((S, T, 2))
    for t in range(T - 2, -1, -1):
        nxt = (B[:, t + 1, :] * beta[:, t + 1, :]) @ A.T
        nxt = nxt / scale[:, t + 1, None]
        beta[:, t, :] = np.where(mask[:, t + 1, None], nxt, beta[:, t + 1, :])

    gamma = alpha * beta
    gamma /= np.maximum(gamma.sum(2, keepdims=True), 1e-300)
    gamma[~mask] = 0.0

    xi = np.zeros((S, T - 1, 2, 2))
    for t in range(T - 1):
        num = (alpha[:, t, :, None] * A[None, :, :] *
               (B[:, t + 1, None, :] * beta[:, t + 1, None, :]))
        num /= np.maximum(scale[:, t + 1, None, None], 1e-300)
        valid = mask[:, t] & mask[:, t + 1]
        xi[:, t] = np.where(valid[:, None, None], num, 0.0)

    loglik = float(np.log(np.maximum(scale, 1e-300))[mask].sum())
    return gamma, xi, loglik


def fit_bernoulli_hmm(seq_list, n_iter=100, tol=1e-6, seed=0, verbose=False):
    """Baum-Welch for the two-state Bernoulli HMM, pooled across all sequences."""
    rng = np.random.default_rng(seed)
    Y, mask = pack_sequences(seq_list)
    overall = Y[mask].mean()
    # start the two states apart so the algorithm has a direction to move in
    p = np.array([max(overall - 0.12, 0.02), min(overall + 0.12, 0.98)])
    A = np.array([[0.7, 0.3], [0.3, 0.7]])
    pi = np.array([0.5, 0.5])

    prev_ll = -np.inf
    for it in range(n_iter):
        gamma, xi, ll = forward_backward(Y, mask, pi, A, p)
        pi = np.maximum(gamma[:, 0, :].mean(0), 1e-8)
        pi /= pi.sum()
        denom = np.maximum(xi.sum((0, 1)).sum(1, keepdims=True), 1e-300)
        A = np.maximum(xi.sum((0, 1)) / denom, 1e-8)
        A /= A.sum(1, keepdims=True)
        w = gamma * mask[:, :, None]
        p = np.clip((w * Y[:, :, None]).sum((0, 1)) / np.maximum(w.sum((0, 1)), 1e-300),
                    0.01, 0.99)
        if verbose and it % 10 == 0:
            print(f"    iter {it}: loglik {ll:.2f}  p={np.round(p,4)}")
        if abs(ll - prev_ll) < tol * max(1.0, abs(prev_ll)):
            break
        prev_ll = ll

    order = np.argsort(p)
    p = p[order]
    A = A[np.ix_(order, order)]
    pi = pi[order]
    _, _, ll = forward_backward(Y, mask, pi, A, p)
    return {'p': p, 'A': A, 'pi': pi, 'loglik': ll, 'n_iter': it + 1,
            'n_obs': int(mask.sum())}


def single_state_loglik(seq_list):
    """Log-likelihood of the one-state (constant probability) model, for comparison."""
    all_y = np.concatenate([np.asarray(s) for s in seq_list])
    p = all_y.mean()
    return float(np.sum(all_y * np.log(p) + (1 - all_y) * np.log(1 - p)))
```

<a id="sec16"></a>
<div style="background:linear-gradient(135deg,#0d3320 0%,#14532d 55%,#2f7a4d 100%);padding:26px 32px;border-radius:14px;margin:34px 0 20px 0;box-shadow:0 6px 24px rgba(13,51,32,0.25);">
<div style="color:rgba(255,255,255,0.55);font-size:11px;letter-spacing:4px;text-transform:uppercase;margin-bottom:8px;">Section 16</div>
<h2 style="color:#ffffff;font-family:Georgia,serif;font-size:24px;margin:0;font-weight:700;">A Latent-State Competitor</h2>
<p style="color:#d9cdb4;margin:6px 0 0 0;font-size:13.5px;">If hotness is a state rather than a streak, a hidden Markov model should find it</p>
</div>


Every model so far defines hotness through observed history: the streak variable $K$ is built from
previous makes. A different theory says that is the wrong representation entirely. Perhaps a player
moves between unobservable performance states, and the streak of makes is a symptom rather than the
thing itself. Under that account, conditioning on past outcomes is an indirect and lossy way to get at
something the data could estimate directly.

A two-state hidden Markov model is the natural formalisation, and several published hot-hand analyses
use exactly this structure, including Wetzels et al. (2016) and Sandri et al. (2020). Write


$$Z_t \in \{\text{low},\ \text{high}\}, \qquad
P(Z_t = j \mid Z_{t-1} = i) = A_{ij}, \qquad
Y_t \mid Z_t = j \;\sim\; \text{Bernoulli}(p_j)$$


so that shooting is constant within a state and the sequential dependence lives entirely in the
transitions. The quantities that decide whether this is a useful description are the gap $p_1 - p_0$,
which says whether the states differ at all, and the diagonal of $A$, which says whether they persist.
A hot hand needs both: states that differ and states that last. Fast switching between two similar
states is not momentum, it is just a reparameterised coin.


<div style="background:linear-gradient(135deg,#2f7a4d 0%,#14532d 100%);border-radius:8px;padding:11px 20px;margin:24px 0 14px 0;">
<span style="color:#d9cdb4;font-family:Georgia,serif;font-size:12px;font-weight:700;letter-spacing:1px;">16.1</span>
<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">Fitting it</span>
</div>


The model is fitted by Baum-Welch, written out here rather than taken from a library. The reason is
structural: there are ten thousand short sequences, one per player-game, sharing a single parameter set.
Padding them into a rectangular array lets the whole forward-backward recursion run vectorised across
sequences, which turns a two-minute fit into a few seconds and makes the null replications below
affordable. The implementation is in `bernoulli_hmm.py`, and the cell after the fit verifies that its
likelihood reduces exactly to the constant-probability likelihood when the two states are set equal,
which is the check that matters for everything that follows.


```python
sequences = [g['FGM'].values for (_, _), g in df.groupby(['player_id', 'GAME_ID'])
             if len(g) >= 6]
Y_pad, mask_pad = pack_sequences(sequences)
overall_rate = Y_pad[mask_pad].mean()
loglik_one_state = single_state_loglik(sequences)

print(f"Sequences (player-games with at least 6 shots): {len(sequences):,}")
print(f"Shots covered: {int(mask_pad.sum()):,}")
print(f"Constant-probability log-likelihood: {loglik_one_state:.4f}")

# Correctness check: setting both states equal must reproduce the one-state likelihood exactly
_, _, ll_degenerate = forward_backward(Y_pad, mask_pad, np.array([0.5, 0.5]),
                                        np.array([[0.9, 0.1], [0.1, 0.9]]),
                                        np.array([overall_rate, overall_rate]))
print(f"HMM log-likelihood with the two states set equal: {ll_degenerate:.4f}")
print(f"Difference from the one-state value: {abs(ll_degenerate - loglik_one_state):.2e}")
```

```python
def run_baum_welch(Y, mask, n_iter, p_init, A_init, record_every=1):
    """Baum-Welch, recording the whole trajectory so convergence can be inspected."""
    p, A, pi = p_init.copy(), A_init.copy(), np.array([0.5, 0.5])
    trajectory = []
    for it in range(n_iter):
        gamma, xi, ll = forward_backward(Y, mask, pi, A, p)
        if it % record_every == 0 or it == n_iter - 1:
            trajectory.append({'iteration': it, 'loglik': ll,
                               'p_low': p[0], 'p_high': p[1], 'gap': p[1] - p[0],
                               'persist_low': A[0, 0], 'persist_high': A[1, 1]})
        pi = np.maximum(gamma[:, 0, :].mean(0), 1e-8); pi /= pi.sum()
        denom = np.maximum(xi.sum((0, 1)).sum(1, keepdims=True), 1e-300)
        A = np.maximum(xi.sum((0, 1)) / denom, 1e-8); A /= A.sum(1, keepdims=True)
        w = gamma * mask[:, :, None]
        p = np.clip((w * Y[:, :, None]).sum((0, 1)) / np.maximum(w.sum((0, 1)), 1e-300),
                    0.001, 0.999)
    return p, A, pd.DataFrame(trajectory)


EM_ITERATIONS = 400
p_final, A_final, em_path = run_baum_welch(
    Y_pad, mask_pad, EM_ITERATIONS,
    np.array([overall_rate - 0.12, overall_rate + 0.12]),
    np.array([[0.7, 0.3], [0.3, 0.7]]), record_every=1)

milestones = em_path[em_path['iteration'].isin([0, 10, 50, 100, 200, EM_ITERATIONS - 1])]
print(milestones[['iteration', 'loglik', 'p_low', 'p_high', 'gap',
                  'persist_low', 'persist_high']].round(5).to_string(index=False))
print(f"\nOne-state log-likelihood : {loglik_one_state:.4f}")
print(f"Final two-state value     : {em_path['loglik'].iloc[-1]:.4f}")
print(f"Improvement over one state: {em_path['loglik'].iloc[-1] - loglik_one_state:+.4f}")
print(f"State separation: started at {em_path['gap'].iloc[0]:.4f}, "
      f"ended at {em_path['gap'].iloc[-1]:.5f}")
```

```python
fig, axes = plt.subplots(1, 3, figsize=(17, 4.9))

ax = axes[0]
ax.plot(em_path['iteration'], em_path['loglik'], color=GREEN_DEEP, lw=2)
ax.axhline(loglik_one_state, color=COPPER, ls='--', lw=1.8,
           label=f'one-state model ({loglik_one_state:.1f})')
ax.set_xlabel('Baum-Welch iteration'); ax.set_ylabel('Log-likelihood')
ax.set_title('(a) The two-state fit climbs toward the one-state value')
ax.legend(frameon=False, fontsize=9)

ax = axes[1]
ax.plot(em_path['iteration'], em_path['p_low'], color=COPPER, lw=2, label='low state')
ax.plot(em_path['iteration'], em_path['p_high'], color=GREEN_DEEP, lw=2, label='high state')
ax.axhline(overall_rate, color=GREY, ls=':', lw=1.6, label='overall rate')
ax.set_xlabel('Baum-Welch iteration'); ax.set_ylabel('P(make) in that state')
ax.set_title('(b) The two states converge on each other')
ax.legend(frameon=False, fontsize=9)

ax = axes[2]
ax.plot(em_path['iteration'], em_path['gap'], color=GREEN_DEEP, lw=2.2)
ax.axhline(0, color=INK, lw=1.2)
for stop, lab in [(30, 'a default\nstopping rule'), (100, '100 iterations')]:
    ax.axvline(stop, color=COPPER, ls='--', lw=1.4, alpha=0.8)
    ax.annotate(lab, (stop, em_path['gap'].max() * 0.75), fontsize=8.5,
                color=COPPER, ha='left', rotation=0)
ax.set_xlabel('Baum-Welch iteration')
ax.set_ylabel(r'State separation $p_{high} - p_{low}$')
ax.set_title('(c) Where you stop decides what you find')

fig.suptitle('Figure 20. The two-state HMM degenerates into a single state',
             fontsize=14, color=GREEN_DEEP, fontweight='bold', y=1.04)
plt.tight_layout(); plt.show()
```

<div style="background:#f3f8f4;border-radius:12px;padding:20px 26px;border-left:5px solid #1c6b3f;font-family:Georgia,serif;line-height:1.85;color:#1a2e22;margin:14px 0;">
<h4 style="color:#14532d;margin:0 0 10px;">Result: the maximum-likelihood two-state model is the one-state model</h4>

Run far enough, the two states collapse into each other. The separation starts at 0.240 by construction
and falls to about 0.003 by iteration 400, with the log-likelihood arriving within 0.005 of the
constant-probability value and still closing. Continuing to 800 iterations, which is too slow to include
in the notebook but was run during development, drives the separation to 0.00007 and reproduces the
one-state log-likelihood exactly to four decimal places. The extra parameters buy nothing.
<br><br>
That is about as clean a null as this kind of model can deliver. A two-state HMM strictly contains the
one-state model as the special case where the states coincide, so its maximised likelihood can never be
lower. What the fit shows is that the maximum sits exactly at that boundary: given these sequences, the
best description of NBA shooting available to a latent-state model is that there is only one state.
<br><br>
This is not an argument that latent-state models are the wrong tool for the hot hand in general. It is a
statement about what this dataset supports. Shot difficulty is not in this model, so a state process
masked by shot selection would not be found here either, and the caveat from Section 13 applies again.

</div>

<div style="background:#fdf5ee;border-radius:14px;padding:22px 28px;border:1.5px solid #d9884f;font-family:Georgia,serif;line-height:1.8;color:#6b3410;margin:14px 0;">
<h4 style="color:#b4531f;margin:0 0 10px;font-size:1.0rem;">A warning that applies to any HMM fitted with default settings</h4>

The collapse is slow. At iteration 30, a point many default convergence rules would have stopped at, the
two states still differ by about 0.13 and the fit looks like evidence for a hot and a cold state. At 100
iterations the gap is roughly 0.04 and still shrinking. Only after several hundred does it become clear
that the separation is heading to zero.
<br><br>
This was not an abstract worry. A first pass at this section, using a general-purpose HMM library at its
default tolerance, reported a low state at 0.393, a high state at 0.504, and a transition matrix that
looked like a finding. Every one of those numbers was an artefact of stopping early, and the same library
run on the same simulated data reached a worse optimum than the implementation used here while reporting
states that looked more convincing.
<br><br>
The practical lesson sits alongside the prior-sensitivity result in Section 15.2. Both are cases where a
routine default, a stopping rule in one case and a conventional prior in the other, produces an apparent
finding that more careful treatment dissolves. In both cases the default errs in the direction of finding
structure that is not there, which is the more dangerous direction for a literature built on detecting
subtle effects.

</div>

<div style="background:linear-gradient(135deg,#2f7a4d 0%,#14532d 100%);border-radius:8px;padding:11px 20px;margin:24px 0 14px 0;">
<span style="color:#d9cdb4;font-family:Georgia,serif;font-size:12px;font-weight:700;letter-spacing:1px;">16.2</span>
<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">Comparing the model families</span>
</div>

```python
family_rows = [
    {'Model': 'Constant probability (one state)',
     'Parameters': 1, 'log-likelihood': loglik_one_state,
     'What it says': 'shooting probability never changes'},
    {'Model': 'Two-state HMM (converged)',
     'Parameters': 5, 'log-likelihood': em_path['loglik'].iloc[-1],
     'What it says': 'states exist in principle but are estimated as identical'},
    {'Model': 'Player effects + difficulty, no streak term',
     'Parameters': shape_fits['K omitted entirely'][3],
     'log-likelihood': shape_fits['K omitted entirely'][2],
     'What it says': 'who shoots, and from where, explains the outcome'},
    {'Model': 'Player effects + difficulty + streak',
     'Parameters': shape_fits['linear in K'][3],
     'log-likelihood': shape_fits['linear in K'][2],
     'What it says': 'adds the streak term to the row above'},
]
family_comparison = pd.DataFrame(family_rows)
family_comparison['AIC'] = (-2 * family_comparison['log-likelihood']
                            + 2 * family_comparison['Parameters'])
family_comparison['AIC vs best'] = (family_comparison['AIC']
                                    - family_comparison['AIC'].min()).round(2)
print("The two sequence models cover the 113,681 shots in sequences of six or more; the two")
print("regression models cover the 113,646 shots with an observed predecessor. The sample")
print("bases are therefore close but not identical, so compare within pairs rather than across.")
print(f"\nStreak term, added to an already-adjusted model: "
      f"{shape_fits['linear in K'][2] - shape_fits['K omitted entirely'][2]:+.2f} log-likelihood "
      f"for one parameter")
family_comparison.round(2)
```

<div style="background:linear-gradient(135deg,#f3f8f4,#e6f0e9);border:1px solid #7aa98c55;border-left:5px solid #1c6b3f;border-radius:0 12px 12px 0;padding:18px 24px;margin:14px 0;">
<div style="font-weight:700;color:#14532d;margin-bottom:10px;font-size:14px;">What the comparison does and does not establish</div>
<div style="color:#1a2e22;font-size:13.5px;line-height:1.75;">
Comparing the sequence models with the regression models is mostly uninformative, and the table says so
rather than inviting the reader to do it. The regressions carry 281 player intercepts and twenty-six
difficulty controls, so almost all of their likelihood advantage comes from knowing who took the shot and
from where. The sample bases also differ slightly. Neither fact has anything to do with streaks.
<br><br>
Two comparisons within the table do carry weight, and both are between properly nested models on the same
data. Between a constant probability and a converged two-state HMM there is no difference at all, so a
latent two-state process captures nothing here. Between an adjusted regression without a streak term and
the same regression with one, the gain is about three log-likelihood points for one parameter, which is
the same small and fragile effect Section 13 arrived at from the other direction and which BIC in Section
14 declined to keep.
</div>
</div>

<a id="sec17"></a>
<div style="background:linear-gradient(135deg,#0d3320 0%,#14532d 55%,#2f7a4d 100%);padding:26px 32px;border-radius:14px;margin:34px 0 20px 0;box-shadow:0 6px 24px rgba(13,51,32,0.25);">
<div style="color:rgba(255,255,255,0.55);font-size:11px;letter-spacing:4px;text-transform:uppercase;margin-bottom:8px;">Section 17</div>
<h2 style="color:#ffffff;font-family:Georgia,serif;font-size:24px;margin:0;font-weight:700;">Out-of-Sample Prediction</h2>
<p style="color:#d9cdb4;margin:6px 0 0 0;font-size:13.5px;">Whether knowing the streak helps predict the next shot, which is a different question</p>
</div>


Everything so far has asked whether a streak effect is detectable. Whether it is *useful* is separate,
and the two can come apart in both directions: a tiny effect can be detectable given enough data while
carrying no predictive value, and a model can predict well through variables that tell you nothing about
the mechanism. Section 13 found an effect of about -0.011 in log-odds, which is small enough that its
predictive contribution is worth measuring rather than assuming.

Shots are sequential, so a random train-test split would leak the future into the past: shots from the
same game, even the same possession, would land on both sides. The split here is chronological. Games
before 8 January 2015 train the models, games through 2 February form a validation period, and
everything after is held out and touched once.


```python
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import HistGradientBoostingClassifier
from sklearn.metrics import roc_auc_score, brier_score_loss, log_loss

pred_data = analysis.copy()
pred_data['game_date'] = pd.to_datetime(
    pred_data['MATCHUP'].str.split(' - ').str[0], format='%b %d, %Y')

cut_train, cut_valid = pred_data['game_date'].quantile([0.60, 0.80])
is_train = (pred_data['game_date'] < cut_train).values
is_valid = ((pred_data['game_date'] >= cut_train) &
            (pred_data['game_date'] < cut_valid)).values
is_test = (pred_data['game_date'] >= cut_valid).values

print(f"Season runs {pred_data['game_date'].min().date()} to {pred_data['game_date'].max().date()}")
print(f"  train      : shots before {cut_train.date()}       {is_train.sum():,}")
print(f"  validation : {cut_train.date()} to {cut_valid.date()}   {is_valid.sum():,}")
print(f"  test       : from {cut_valid.date()}               {is_test.sum():,}")
print(f"\nField-goal rate: train {y_obs[is_train].mean():.4f}, test {y_obs[is_test].mean():.4f}")
```

<div style="background:#fdf5ee;border-radius:14px;padding:22px 28px;border:1.5px solid #d9884f;font-family:Georgia,serif;line-height:1.8;color:#6b3410;margin:14px 0;">
<h4 style="color:#b4531f;margin:0 0 10px;font-size:1.0rem;">What is and is not held out</h4>

The player identities are shared across periods, which is deliberate: this is a question about predicting
future shots by known players, not about generalising to players never seen. The shot-difficulty
distribution is also broadly stable across a season, so the test period is not a different environment.
<br><br>
One real limitation remains. Player baseline rates are estimated from the training window and applied
forward, so a player who genuinely changed over the season is predicted with a stale rate. That
disadvantages every model in the comparison equally, and since the comparison is between feature sets
rather than against an absolute standard, it does not distort the quantity of interest.

</div>

```python
player_dummies = pd.get_dummies(pred_data['player_id'], dtype=float).values
K_feature = K_obs[:, None]
player_rate = pred_data.groupby('player_id')['FGM'].transform('mean').values[:, None]

feature_sets = {
    'player identity only': np.column_stack([player_dummies]),
    'player + shot difficulty': np.column_stack([player_dummies, X_difficulty]),
    'player + difficulty + streak': np.column_stack([player_dummies, X_difficulty, K_feature]),
}

results = []
for name, X in feature_sets.items():
    model = LogisticRegression(max_iter=300, solver='lbfgs')
    model.fit(X[is_train], y_obs[is_train])
    p_test = model.predict_proba(X[is_test])[:, 1]
    results.append({'Features': name, 'Model': 'logistic',
                    'ROC-AUC': roc_auc_score(y_obs[is_test], p_test),
                    'Log loss': log_loss(y_obs[is_test], p_test),
                    'Brier': brier_score_loss(y_obs[is_test], p_test)})

# A flexible learner, given the chance to find interactions a linear model would miss
for name, X in [('difficulty + player rate', np.column_stack([X_difficulty, player_rate])),
                ('difficulty + player rate + streak',
                 np.column_stack([X_difficulty, player_rate, K_feature]))]:
    gbm = HistGradientBoostingClassifier(max_iter=200, learning_rate=0.06,
                                          max_depth=4, random_state=RANDOM_SEED)
    gbm.fit(X[is_train], y_obs[is_train])
    p_test = gbm.predict_proba(X[is_test])[:, 1]
    results.append({'Features': name, 'Model': 'gradient boosting',
                    'ROC-AUC': roc_auc_score(y_obs[is_test], p_test),
                    'Log loss': log_loss(y_obs[is_test], p_test),
                    'Brier': brier_score_loss(y_obs[is_test], p_test)})

prediction_table = pd.DataFrame(results)
prediction_table.round(5)
```

```python
auc = prediction_table.set_index('Features')['ROC-AUC']
gain_difficulty = auc['player + shot difficulty'] - auc['player identity only']
gain_streak_lin = auc['player + difficulty + streak'] - auc['player + shot difficulty']
gain_streak_gbm = (auc['difficulty + player rate + streak']
                   - auc['difficulty + player rate'])

print("Incremental ROC-AUC on held-out games:")
print(f"  adding shot difficulty to player identity : {gain_difficulty:+.5f}")
print(f"  adding the streak term, logistic          : {gain_streak_lin:+.5f}")
print(f"  adding the streak term, gradient boosting : {gain_streak_gbm:+.5f}")
if abs(gain_streak_lin) > 1e-12:
    print(f"\nShot difficulty contributes roughly "
          f"{abs(gain_difficulty/gain_streak_lin):.0f} times as much as the streak term.")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14.5, 5.2))

ax = axes[0]
labels = ['player\nidentity', '+ shot\ndifficulty', '+ streak']
vals = [auc['player identity only'], auc['player + shot difficulty'],
        auc['player + difficulty + streak']]
bars = ax.bar(labels, vals, color=[SAND, GREEN_MID, GREEN_DEEP], edgecolor=INK, width=0.6)
for b, v in zip(bars, vals):
    ax.text(b.get_x() + b.get_width()/2, v + 0.004, f'{v:.4f}', ha='center',
            fontsize=11, fontweight='bold', color=GREEN_DEEP)
ax.axhline(0.5, color=GREY, ls=':', lw=1.5)
ax.set_ylim(0.48, 0.70); ax.set_ylabel('ROC-AUC on held-out games')
ax.set_title('(a) Where the predictive power comes from')

ax = axes[1]
best_X = np.column_stack([player_dummies, X_difficulty, K_feature])
best = LogisticRegression(max_iter=300, solver='lbfgs').fit(best_X[is_train], y_obs[is_train])
p_test = best.predict_proba(best_X[is_test])[:, 1]
bins = np.quantile(p_test, np.linspace(0, 1, 11))
idx = np.clip(np.digitize(p_test, bins[1:-1]), 0, 9)
obs_rate = np.array([y_obs[is_test][idx == b].mean() for b in range(10)])
pred_rate = np.array([p_test[idx == b].mean() for b in range(10)])
ax.plot([0.2, 0.75], [0.2, 0.75], color=GREY, ls='--', lw=1.5, label='perfect calibration')
ax.plot(pred_rate, obs_rate, marker='o', markersize=8, lw=2, color=GREEN_DEEP,
        label='observed')
ax.set_xlabel('Predicted probability (decile mean)')
ax.set_ylabel('Observed field-goal rate')
ax.set_title('(b) Calibration of the fullest model')
ax.legend(frameon=False, fontsize=9.5)

fig.suptitle('Figure 21. Predicting held-out shots under a chronological split',
             fontsize=14, color=GREEN_DEEP, fontweight='bold', y=1.03)
plt.tight_layout(); plt.show()
```

<div style="background:#f3f8f4;border-radius:12px;padding:20px 26px;border-left:5px solid #1c6b3f;font-family:Georgia,serif;line-height:1.85;color:#1a2e22;margin:14px 0;">
<h4 style="color:#14532d;margin:0 0 10px;">Result: streak information is worth essentially nothing out of sample</h4>

Knowing only who took the shot predicts the outcome barely better than a coin, at an ROC-AUC of about
0.54. Adding where the shot was taken from and how closely it was defended lifts that to about 0.64, an
increment of roughly 0.095. Adding the streak term on top moves it by about 0.0001, which is inside the
noise of a single held-out period and would not survive being measured again on a different cut.
<br><br>
Gradient boosting was given the same chance with a flexible learner that could pick up interactions
between streak and context that a linear model would miss. It finds a gain of the same negligible order,
about 0.0003. Two model families with very different inductive biases agree that there is nothing there
to find.
<br><br>
The comparison worth keeping is the ratio, which the cell above prints: shot difficulty contributes three
orders of magnitude more predictive information than the streak variable. The exact multiple is not
meaningful, since it is a ratio to a quantity indistinguishable from zero and it moves around with the
split, but its size is. Anyone building something that tries to anticipate the next shot should spend
their effort on shot quality and ignore the streak entirely, and that holds regardless of whether the
small inferential effect in Section 13 is real.
<br><br>
This also sharpens the reading of the earlier sections. A null in prediction and a null in inference are
different kinds of evidence, and here they point the same way, which is more informative than either
alone. The one thing this cannot rule out is an effect too small to matter predictively but real enough
to matter scientifically, which is precisely the regime the estimates have been circling.

</div>

The five data-generating processes and the competing detection methods are defined here, immediately before the study that uses them.

```python
"""
Operating characteristics of competing hot-hand methods.

Five data-generating processes, one of which has no hot hand at all, are each analysed by
every method the notebook uses. That turns the study from "what do these methods say about
the NBA" into "when do these methods get the right answer", which is the part that
generalises beyond one season of one league.
"""
import numpy as np
import scipy.sparse as sp


# --------------------------------------------------------------- data generation
def _draw_sequence(rng, n_shots, base_logit, effect_fn):
    """One player's season. effect_fn(k) returns the log-odds shift for streak length k."""
    out = np.empty(n_shots, dtype=np.int8)
    streak = np.empty(n_shots, dtype=np.int8)
    run = 0
    for t in range(n_shots):
        k = min(run, 3)
        streak[t] = k
        p = 1.0 / (1.0 + np.exp(-(base_logit + effect_fn(k))))
        made = rng.random() < p
        out[t] = made
        run = run + 1 if made else 0
    return out, streak


def generate(dgp, n_players, shots_each, rng):
    """Returns (outcomes, streaks, player_index) for one replication of the named DGP."""
    base = rng.normal(-0.2, 0.25, n_players)
    outcomes, streaks, idx = [], [], []

    if dgp == 'no effect':
        effects = [lambda k: 0.0 for _ in range(n_players)]
    elif dgp == 'constant effect':
        effects = [lambda k: 0.10 * k for _ in range(n_players)]
    elif dgp == 'heterogeneous effect':
        b = rng.normal(0.10, 0.20, n_players)
        effects = [(lambda k, bi=bi: bi * k) for bi in b]
    elif dgp == 'threshold at three':
        effects = [lambda k: 0.30 if k >= 3 else 0.0 for _ in range(n_players)]
    elif dgp == 'latent hot/cold state':
        effects = None
    else:
        raise ValueError(dgp)

    if dgp == 'latent hot/cold state':
        # shooting is driven by a persistent two-state process, not by observed history
        for i in range(n_players):
            p_low = 1.0 / (1.0 + np.exp(-(base[i] - 0.35)))
            p_high = 1.0 / (1.0 + np.exp(-(base[i] + 0.35)))
            z = 0
            o = np.empty(shots_each, dtype=np.int8)
            s = np.empty(shots_each, dtype=np.int8)
            run = 0
            for t in range(shots_each):
                s[t] = min(run, 3)
                made = rng.random() < (p_low if z == 0 else p_high)
                o[t] = made
                run = run + 1 if made else 0
                stay = 0.85 if z == 0 else 0.80
                z = z if rng.random() < stay else 1 - z
            outcomes.append(o); streaks.append(s); idx.append(np.full(shots_each, i))
    else:
        for i in range(n_players):
            o, s = _draw_sequence(rng, shots_each, base[i], effects[i])
            outcomes.append(o); streaks.append(s); idx.append(np.full(shots_each, i))

    return (np.concatenate(outcomes), np.concatenate(streaks).astype(float),
            np.concatenate(idx))


# --------------------------------------------------------------- methods
def ratio_statistic(y, idx, k=3):
    """P(make | previous k makes) minus P(make), pooled within player then averaged."""
    vals = []
    for i in np.unique(idx):
        yi = y[idx == i]
        num = den = 0
        run = 0
        for t in range(len(yi)):
            made = int(yi[t])          # cast: yi is int8 and the running sum would overflow
            if run >= k:
                den += 1
                num += made
            run = run + 1 if made == 1 else 0
        if den > 0:
            vals.append(num / den - yi.mean())
    return float(np.mean(vals)) if vals else np.nan


def ratio_permutation_null(y, idx, rng, n_perm, k=3):
    """Null distribution of the ratio statistic under reordering within each player."""
    out = np.empty(n_perm)
    for b in range(n_perm):
        yp = np.empty_like(y)
        for i in np.unique(idx):
            m = idx == i
            yp[m] = rng.permutation(y[m])
        out[b] = ratio_statistic(yp, idx, k)
    return out


def logistic_streak_coefficient(y, K, idx):
    """Streak coefficient from a logistic model with one intercept per player."""
    n_players = idx.max() + 1
    D = sp.csr_matrix((np.ones(len(y)), (np.arange(len(y)), idx)),
                      shape=(len(y), n_players))
    X = sp.hstack([sp.csr_matrix(K[:, None]), D], format='csr')
    beta = np.zeros(X.shape[1])
    for _ in range(30):
        eta = X @ beta
        mu = 1 / (1 + np.exp(-eta))
        W = np.clip(mu * (1 - mu), 1e-9, None)
        XtW = X.T.multiply(W)
        new = np.linalg.solve((XtW @ X).toarray(), XtW @ (eta + (y - mu) / W))
        if np.max(np.abs(new - beta)) < 1e-9:
            beta = new
            break
        beta = new
    eta = X @ beta
    mu = 1 / (1 + np.exp(-eta))
    W = np.clip(mu * (1 - mu), 1e-9, None)
    se = np.sqrt(np.linalg.inv((X.T.multiply(W) @ X).toarray())[0, 0])
    return float(beta[0]), float(se)


def logistic_null_draws(y, K, idx, rng, n_null):
    """Refit the logistic model on independent shooting at each player's own rate."""
    player_rate = np.empty(len(y))
    for i in np.unique(idx):
        m = idx == i
        player_rate[m] = y[m].mean()
    draws = np.empty(n_null)
    for b in range(n_null):
        ys = (rng.random(len(y)) < player_rate).astype(np.int8)
        Ks = streaks_from(ys, idx)
        draws[b] = logistic_streak_coefficient(ys, Ks, idx)[0]
    return draws


def streaks_from(y, idx):
    """Rebuild the capped streak variable after resampling outcomes."""
    K = np.empty(len(y))
    run = 0
    prev = -1
    for t in range(len(y)):
        if idx[t] != prev:
            run = 0
            prev = idx[t]
        K[t] = min(run, 3)
        run = run + 1 if y[t] == 1 else 0
    return K
```

<a id="sec18"></a>
<div style="background:linear-gradient(135deg,#0d3320 0%,#14532d 55%,#2f7a4d 100%);padding:26px 32px;border-radius:14px;margin:34px 0 20px 0;box-shadow:0 6px 24px rgba(13,51,32,0.25);">
<div style="color:rgba(255,255,255,0.55);font-size:11px;letter-spacing:4px;text-transform:uppercase;margin-bottom:8px;">Section 18</div>
<h2 style="color:#ffffff;font-family:Georgia,serif;font-size:24px;margin:0;font-weight:700;">Model Recovery Study</h2>
<p style="color:#d9cdb4;margin:6px 0 0 0;font-size:13.5px;">Turning the methods themselves into the object of study</p>
</div>


Everything so far has applied methods to one season of NBA shooting and reported what they said. That
leaves an obvious gap: when these methods disagree, or agree, there is no way to know which of them was
capable of being right. The way to close it is to generate data where the truth is known by construction
and see which methods recover it.

Five processes are simulated, each with 120 players taking 400 shots, sized to resemble a real season.
One contains no hot hand at all, which makes it the false-positive test. Three contain a hot hand of
different shapes. The fifth contains no streak mechanism but a persistent latent state, which is the
competing theory from Section 16 in generative form.


```python
dgp_descriptions = pd.DataFrame([
    ('no effect', 'shooting is independent; streak length is irrelevant', 'beta = 0'),
    ('constant effect', 'each extra preceding make shifts the log-odds equally',
     'beta = 0.10 for every player'),
    ('heterogeneous effect', 'players differ in how much streaks matter to them',
     'beta_i ~ N(0.10, 0.20^2)'),
    ('threshold at three', 'nothing happens until three makes, then a jump',
     '+0.30 once K >= 3, else 0'),
    ('latent hot/cold state', 'a persistent unobserved state drives shooting, not the streak',
     'two states +/- 0.35 in log-odds, persistence 0.85 / 0.80'),
], columns=['Data-generating process', 'What it represents', 'Parameters'])
dgp_descriptions
```


Four decision rules are applied to every replication. Two use the ratio statistic from Section 8, once
compared against zero as the classical literature did and once against its own permutation null. Two use
the logistic streak coefficient from Section 12, again compared against zero and against a simulated null
built by resampling independent shooting at each player's own rate. The point of running all four is that
the pairs differ only in their benchmark, so any difference in their behaviour is attributable to the
benchmark alone.


```python
DGPS = ['no effect', 'constant effect', 'heterogeneous effect',
        'threshold at three', 'latent hot/cold state']
N_REP, N_PERM, N_NULL = 40, 60, 30
N_PLAYERS_SIM, SHOTS_SIM = 120, 400

recovery_rows = []
for dgp in DGPS:
    for rep in range(N_REP):
        rng_r = np.random.default_rng(10_000 + 97 * rep)
        y_s, K_s, idx_s = generate(dgp, N_PLAYERS_SIM, SHOTS_SIM, rng_r)

        ratio_obs = ratio_statistic(y_s, idx_s)
        ratio_null = ratio_permutation_null(y_s, idx_s, rng_r, N_PERM)

        beta_obs, beta_se = logistic_streak_coefficient(y_s, K_s, idx_s)
        beta_null = logistic_null_draws(y_s, K_s, idx_s, rng_r, N_NULL)

        recovery_rows.append({
            'dgp': dgp, 'rep': rep,
            'ratio_obs': ratio_obs, 'ratio_null_mean': ratio_null.mean(),
            'ratio_corrected': ratio_obs - ratio_null.mean(),
            'ratio_z': (ratio_obs - ratio_null.mean()) / ratio_null.std(),
            'beta_obs': beta_obs, 'beta_null_mean': beta_null.mean(),
            'beta_corrected': beta_obs - beta_null.mean(),
            'beta_z': (beta_obs - beta_null.mean()) / beta_null.std(),
            'beta_z_vs_zero': beta_obs / beta_se,
        })

recovery_raw = pd.DataFrame(recovery_rows)
print(f"{len(recovery_raw)} replications completed "
      f"({N_REP} per process, {len(DGPS)} processes)")
```

```python
summary_rows = []
for dgp in DGPS:
    g = recovery_raw[recovery_raw['dgp'] == dgp]
    summary_rows.append({
        'Data-generating process': dgp,
        'ratio, naive': g['ratio_obs'].mean(),
        'ratio, corrected': g['ratio_corrected'].mean(),
        'beta, naive': g['beta_obs'].mean(),
        'beta, corrected': g['beta_corrected'].mean(),
        'detect: ratio vs zero': (g['ratio_obs'] > 0).mean(),
        'detect: ratio vs null': (g['ratio_z'] > 1.645).mean(),
        'detect: beta vs zero': (g['beta_z_vs_zero'].abs() > 1.96).mean(),
        'detect: beta vs null': (g['beta_z'].abs() > 1.96).mean(),
    })
recovery_summary = pd.DataFrame(summary_rows)
print("Detection rates. On the 'no effect' row these are false-positive rates and should")
print("sit at the nominal level (0.05 for the two-sided tests); elsewhere they are power.")
recovery_summary.round(4)
```

```python
fig, axes = plt.subplots(1, 3, figsize=(17.5, 5.2))
short = ['no\neffect', 'constant', 'heterog.', 'threshold', 'latent\nstate']
xs = np.arange(len(DGPS))

ax = axes[0]
w = 0.38
ax.bar(xs - w/2, recovery_summary['beta, naive'], w, color=SAND,
       edgecolor=COPPER, label='naive')
ax.bar(xs + w/2, recovery_summary['beta, corrected'], w, color=GREEN_SOFT,
       edgecolor=GREEN_DEEP, label='bias corrected')
ax.axhline(0, color=INK, lw=1.2)
ax.set_xticks(xs); ax.set_xticklabels(short, fontsize=9)
ax.set_ylabel(r'Mean estimated $\beta$')
ax.set_title('(a) Estimates, with and without correction')
ax.legend(frameon=False, fontsize=9)

ax = axes[1]
fp = recovery_summary.set_index('Data-generating process').loc['no effect']
methods = ['ratio vs null', 'beta vs zero', 'beta vs null']
vals = [fp['detect: ratio vs null'], fp['detect: beta vs zero'], fp['detect: beta vs null']]
nominal = [0.05, 0.05, 0.05]
cols = [COPPER if v > 0.08 else GREEN_DEEP for v in vals]
bars = ax.bar(range(3), vals, color=cols, edgecolor=INK, width=0.55)
ax.axhline(0.05, color=GREY, ls='--', lw=1.8, label='nominal level')
for b, v in zip(bars, vals):
    ax.text(b.get_x()+b.get_width()/2, v+0.004, f'{v:.3f}', ha='center',
            fontsize=11, fontweight='bold', color=GREEN_DEEP)
ax.set_xticks(range(3))
ax.set_xticklabels(['ratio\nvs null', 'beta\nvs zero', 'beta\nvs null'], fontsize=9.5)
ax.set_ylabel('False-positive rate')
ax.set_title('(b) Type I error when there is no effect')
ax.legend(frameon=False, fontsize=9)

ax = axes[2]
ax.bar(xs - w/2, recovery_summary['detect: beta vs zero'], w, color=SAND,
       edgecolor=COPPER, label='beta vs zero')
ax.bar(xs + w/2, recovery_summary['detect: beta vs null'], w, color=GREEN_SOFT,
       edgecolor=GREEN_DEEP, label='beta vs simulated null')
ax.axhline(0.05, color=GREY, ls='--', lw=1.5)
ax.set_xticks(xs); ax.set_xticklabels(short, fontsize=9)
ax.set_ylabel('Detection rate'); ax.set_ylim(0, 1.08)
ax.set_title('(c) Power, and the cost of the wrong benchmark')
ax.legend(frameon=False, fontsize=9)

fig.suptitle('Figure 22. How the competing methods behave when the truth is known',
             fontsize=14, color=GREEN_DEEP, fontweight='bold', y=1.03)
plt.tight_layout(); plt.show()
```

<div style="background:#f3f8f4;border-radius:12px;padding:20px 26px;border-left:5px solid #1c6b3f;font-family:Georgia,serif;line-height:1.85;color:#1a2e22;margin:14px 0;">
<h4 style="color:#14532d;margin:0 0 10px;">Result one: the wrong benchmark doubles the false-positive rate</h4>

On data generated with no hot hand whatsoever, the logistic coefficient compared against zero rejects in
10% of replications at a nominal 5% level. The identical coefficient compared against its own simulated
null rejects in exactly 5%. Nothing changes between those two rows except what the estimate is measured
against.
<br><br>
This is the practical cost of Section 12.1's finding stated as an error rate. A researcher using standard
software, reading the p-value it prints, and applying a 5% threshold is running a test with twice the
advertised false-positive rate whenever the regressor is built from the outcome's own history. The
permutation-calibrated ratio test comes in at 2.5%, slightly conservative, which is the safer direction.
<br><br>
Both corrections also recover the truth on the estimate itself. Under the null the naive ratio statistic
averages -0.011 and the naive coefficient -0.007, while their corrected versions average +0.001 and
-0.000 respectively. The corrections are doing exactly what they were designed to do.

</div>

<div style="background:#f3f8f4;border-radius:12px;padding:20px 26px;border-left:5px solid #1c6b3f;font-family:Georgia,serif;line-height:1.85;color:#1a2e22;margin:14px 0;">
<h4 style="color:#14532d;margin:0 0 10px;">Result two: every method has ample power, so the real-data null is not a power failure</h4>

Against a constant effect of 0.10 in log-odds, every method detects it in 100% of replications. The same
holds for heterogeneous effects and for a threshold that only switches on at three makes. At the sample
size used here, 120 players taking 400 shots, an effect of that size is not something these methods can
miss.
<br><br>
That matters for reading Sections 13 through 17. The real data are larger than these simulations, 281
players and 113,646 usable shots, and the methods found nothing. This study establishes that the absence
is not an artefact of underpowered tools: had an effect of the size the literature discusses been
present, these same methods would have found it essentially every time. The one qualification is that
power was measured at effect sizes around 0.10, and the fully adjusted real-data estimate is about
-0.011, an order of magnitude smaller. Nothing here claims the methods could detect an effect that small.

</div>

<div style="background:#fdf5ee;border-radius:14px;padding:22px 28px;border:1.5px solid #d9884f;font-family:Georgia,serif;line-height:1.8;color:#6b3410;margin:14px 0;">
<h4 style="color:#b4531f;margin:0 0 10px;font-size:1.0rem;">Result three: no method can tell a streak from a latent state</h4>

The fifth process contains no streak mechanism at all. Shooting is governed by a persistent hidden state,
and the streak variable is a downstream symptom rather than a cause. Yet the streak coefficient detects
"an effect" in 97.5% of replications, and averages +0.044 after correction.
<br><br>
So a significant streak coefficient does not license the interpretation that streaks matter. It is equally
consistent with a latent process that streaks merely reflect, and the methods used throughout this
notebook cannot separate those two stories. The reverse holds as well: Section 16's HMM collapsing into a
single state on real data says a two-state process does not fit, but a genuine streak mechanism would not
necessarily show up as one.
<br><br>
This is a limit on interpretation rather than on the estimates, and it is worth carrying into the
conclusion. Had this notebook found a robust positive streak coefficient, the honest reading would still
have been that something produces short-term dependence, not that the dependence works through the
streak. As it happens the coefficient is null, so the question does not arise here, but it would arise
immediately for anyone who found one.

</div>

<a id="sec19"></a>
<div style="background:linear-gradient(135deg,#0d3320 0%,#14532d 55%,#2f7a4d 100%);padding:26px 32px;border-radius:14px;margin:34px 0 20px 0;box-shadow:0 6px 24px rgba(13,51,32,0.25);">
<div style="color:rgba(255,255,255,0.55);font-size:11px;letter-spacing:4px;text-transform:uppercase;margin-bottom:8px;">Section 19</div>
<h2 style="color:#ffffff;font-family:Georgia,serif;font-size:24px;margin:0;font-weight:700;">Robustness</h2>
<p style="color:#d9cdb4;margin:6px 0 0 0;font-size:13.5px;">Whether any of the analytical choices along the way were load-bearing</p>
</div>


A dozen decisions have been made in getting here: which rows to exclude, where to cap the streak
variable, how flexibly to control for difficulty, which players to include. Each was justified when it
was made, but justification is cheap. The test is whether the answer moves when the decision is reversed.


```python
def difficulty_design_for(subset, flexible=True):
    """Rebuild the control block for an arbitrary subset, dropping any column that
    a subset has made constant (which would otherwise make the design singular)."""
    if flexible:
        blocks = [
            pd.get_dummies(pd.cut(subset['SHOT_DIST'], [0, 4, 8, 12, 16, 20, 22, 24, 27, 60],
                                   right=False), drop_first=True, dtype=float).values,
            pd.get_dummies(pd.cut(subset['CLOSE_DEF_DIST'], [0, 2, 3, 4, 5, 6, 8, 60],
                                   right=False), drop_first=True, dtype=float).values]
    else:
        blocks = [subset['SHOT_DIST'].values[:, None],
                  subset['CLOSE_DEF_DIST'].values[:, None]]
    blocks += [
        pd.get_dummies(subset['PERIOD'].clip(upper=5), drop_first=True, dtype=float).values,
        subset['TOUCH_TIME'].fillna(subset['TOUCH_TIME'].median()).values[:, None],
        subset['DRIBBLES'].values[:, None],
        (subset['PTS_TYPE'] == 3).astype(float).values[:, None],
        (subset['LOCATION'] == 'H').astype(float).values[:, None],
        subset['SHOT_CLOCK'].fillna(subset['SHOT_CLOCK'].median()).values[:, None],
        subset['shot_clock_missing'].values[:, None],
        subset['game_clock_sec'].values[:, None] / 60.0,
        subset['touch_time_invalid'].values[:, None]]
    X = np.column_stack(blocks)
    return X[:, X.std(0) > 1e-12]


def fit_subset(subset, K_values, flexible=True):
    codes_sub = pd.factorize(subset['player_id'])[0]
    Xd = np.column_stack([K_values, difficulty_design_for(subset, flexible)])
    b, se = sparse_logit(subset['FGM'].values, Xd, codes_sub, codes_sub.max() + 1)
    return b[0], se[0]


K_base = np.minimum(analysis['prev_streak'], 3).astype(float).values
inconsistent = (((analysis['PTS_TYPE'] == 3) & (analysis['SHOT_DIST'] < 22)) |
                ((analysis['PTS_TYPE'] == 2) & (analysis['SHOT_DIST'] > 23.75)))
high_volume = analysis['player_id'].map(analysis['player_id'].value_counts()) >= 200

specifications = [
    ('primary specification', analysis, K_base, True),
    ('drop shot-clock-missing rows', analysis[analysis['shot_clock_missing'] == 0],
     K_base[(analysis['shot_clock_missing'] == 0).values], True),
    ('drop shot-value inconsistencies', analysis[~inconsistent],
     K_base[(~inconsistent).values], True),
    ('drop invalid touch-time rows', analysis[analysis['touch_time_invalid'] == 0],
     K_base[(analysis['touch_time_invalid'] == 0).values], True),
    ('streak capped at 2', analysis,
     np.minimum(analysis['prev_streak'], 2).astype(float).values, True),
    ('streak capped at 4', analysis,
     np.minimum(analysis['prev_streak'], 4).astype(float).values, True),
    ('linear rather than binned difficulty', analysis, K_base, False),
    ('players with 200+ shots only', analysis[high_volume],
     K_base[high_volume.values], True),
]

rob_rows = []
for name, subset, K_values, flexible in specifications:
    b, se = fit_subset(subset, K_values, flexible)
    rob_rows.append({'Specification': name, 'N': len(subset),
                     'beta (vs zero)': b, 'se': se, 'z vs zero': b / se,
                     'bias-corrected': b - null_beta_m2.mean()})
robustness_table = pd.DataFrame(rob_rows)
robustness_table['z vs null'] = ((robustness_table['beta (vs zero)'] - null_beta_m2.mean())
                                 / null_beta_m2.std())
robustness_table.round(5)
```

```python
fig, ax = plt.subplots(figsize=(11.5, 6))
order = np.arange(len(robustness_table))[::-1]
vals = robustness_table['beta (vs zero)'].values
errs = 1.96 * robustness_table['se'].values
corrected = robustness_table['bias-corrected'].values

ax.axvspan(null_beta_m2.mean() - 1.96 * null_beta_m2.std(),
           null_beta_m2.mean() + 1.96 * null_beta_m2.std(),
           color=SAND, alpha=0.55, label='95% band of the simulated null')
ax.axvline(0, color=GREY, ls=':', lw=1.6, label='zero')
ax.axvline(null_beta_m2.mean(), color=COPPER, ls='--', lw=1.8,
           label=f'null mean ({null_beta_m2.mean():+.4f})')
ax.errorbar(vals, order, xerr=errs, fmt='o', markersize=8, color=GREEN_DEEP,
            capsize=4, lw=1.8, label='estimate vs zero, 95% CI')
ax.scatter(corrected, order, marker='D', s=60, color=COPPER_LT,
           edgecolor=INK, zorder=4, label='bias corrected')
ax.set_yticks(order)
ax.set_yticklabels(robustness_table['Specification'], fontsize=9.5)
ax.set_xlabel(r'Streak coefficient $\beta$')
ax.set_title('Figure 23. Every specification lands in the same place, and none of them clears the null',
             fontsize=13, color=GREEN_DEEP, fontweight='bold')
ax.legend(frameon=False, fontsize=9, loc='lower right')
plt.tight_layout(); plt.show()
```

<div style="background:#f3f8f4;border-radius:12px;padding:20px 26px;border-left:5px solid #1c6b3f;font-family:Georgia,serif;line-height:1.85;color:#1a2e22;margin:14px 0;">
<h4 style="color:#14532d;margin:0 0 10px;">Result: highly stable, and stably non-significant</h4>

Across eight specifications the coefficient sits between -0.014 and -0.019, with z-statistics against
zero between -2.0 and -2.9. Dropping the shot-clock-missing rows, the shot-value inconsistencies or the
invalid touch-time rows barely moves it. Capping the streak at two or at four barely moves it. Restricting
to high-volume players barely moves it. The one choice that matters at all is using linear rather than
binned difficulty controls, which shrinks the estimate to -0.014, and in the direction that fits Section
7.1: a linear control cannot absorb a non-monotone relationship, so some of the difficulty effect leaks
into the streak coefficient.
<br><br>
This stability is worth separating from significance, because the two get conflated. The estimate is
robust. Its interpretation is not, and that is the whole point of the figure. Every specification sits
comfortably inside the 95% band of the simulated null, which is centred near -0.006 rather than at zero.
Judged against zero the finding looks consistent and significant across eight robustness checks. Judged
against the benchmark that is actually correct, it is consistently and robustly nothing.
<br><br>
A robustness table is usually read as an argument that a finding is real. Here it is doing the opposite
work: it shows that the null result is not an artefact of any particular analytical choice either.

</div>

The loader for the multi-season archive is defined here so the notebook stays self-contained. It harmonises a schema that is not perfectly stable across the 22 files and reconstructs shot order from the game clock.

```python
"""
Multi-season NBA shot data: loading, harmonisation, and streak construction.

Two datasets are used in this project and they are complementary rather than
interchangeable:

  2014-15 shot logs    defender distance, touch time, dribbles, shot clock.
                       The only source here that measures how contested a shot was,
                       which is what the shot-difficulty adjustment requires.

  2004-2025 shot chart 22 seasons, true court coordinates, NBA zone labels and
                       action types, but no defender information.

The first supports the replication and the difficulty adjustment; the second supports
season heterogeneity, temporal validation, and any analysis that needs court geometry.
"""
import zipfile
import numpy as np
import pandas as pd

USECOLS = ['SEASON_2', 'PLAYER_ID', 'PLAYER_NAME', 'POSITION_GROUP', 'GAME_ID',
           'GAME_DATE', 'SHOT_MADE', 'SHOT_TYPE', 'ACTION_TYPE', 'BASIC_ZONE',
           'ZONE_RANGE', 'LOC_X', 'LOC_Y', 'SHOT_DISTANCE', 'QUARTER',
           'MINS_LEFT', 'SECS_LEFT']

COMPACT = {'SHOT_MADE': 'bool', 'LOC_X': 'float32', 'LOC_Y': 'float32',
           'SHOT_DISTANCE': 'float32', 'QUARTER': 'int8',
           'MINS_LEFT': 'int8', 'SECS_LEFT': 'int8'}


def load_season(year, data_dir):
    """One season, with compact dtypes so all of them fit in memory.

    The schema is not perfectly stable across the archive: the 2024-25 file omits
    POSITION_GROUP and POSITION, which every earlier season carries.  Rather than
    assume a fixed column list, the loader intersects the requested columns with
    what the file actually has and records the difference, so a silently missing
    variable cannot turn into a silently dropped control.
    """
    path = f'{data_dir}/NBA_{year}_Shots.csv.zip'
    with zipfile.ZipFile(path) as z:
        name = [n for n in z.namelist() if n.endswith('.csv') and not n.startswith('__')][0]
        available = pd.read_csv(z.open(name), nrows=0).columns.tolist()
        wanted = [c for c in USECOLS if c in available]
        missing = [c for c in USECOLS if c not in available]
        df = pd.read_csv(z.open(name), usecols=wanted,
                         dtype={k: v for k, v in COMPACT.items() if k in wanted})
    for c in ('SEASON_2', 'POSITION_GROUP', 'SHOT_TYPE', 'BASIC_ZONE',
              'ZONE_RANGE', 'ACTION_TYPE'):
        if c in df.columns:
            df[c] = df[c].astype('category')
    df.attrs['missing_columns'] = missing
    return df


def add_shot_order(df):
    """Chronological order of shots within each player-game.

    The file has no shot counter, so order is reconstructed from the game clock:
    seconds elapsed since tip-off, treating every period as 12 minutes and any
    period beyond the fourth as a 5-minute overtime.
    """
    period = df['QUARTER'].astype('int16')
    regulation = np.minimum(period, 4)
    overtime = np.maximum(period - 4, 0)
    period_start = (regulation - 1) * 720 + overtime * 300
    period_length = np.where(period <= 4, 720, 300)
    remaining = df['MINS_LEFT'].astype('int16') * 60 + df['SECS_LEFT'].astype('int16')
    df['elapsed'] = (period_start + (period_length - remaining)).astype('int32')
    return df


def build_streaks(df):
    """Sort within player-game and attach the preceding-make count.

    Shots sharing an identical timestamp cannot be ordered, so the first of any
    tied group keeps its place and the rest are flagged: their preceding-make
    count is undefined and they are excluded from streak-conditional analysis,
    exactly as the gap rule in the single-season data does.
    """
    df = df.sort_values(['PLAYER_ID', 'GAME_ID', 'elapsed'],
                        kind='mergesort').reset_index(drop=True)

    unit = (df['PLAYER_ID'].astype(str) + '_' + df['GAME_ID'].astype(str)).values
    new_unit = np.empty(len(df), dtype=bool)
    new_unit[0] = True
    new_unit[1:] = unit[1:] != unit[:-1]

    tied = np.zeros(len(df), dtype=bool)
    same_unit = ~new_unit
    tied[1:] = same_unit[1:] & (df['elapsed'].values[1:] == df['elapsed'].values[:-1])

    df['seq_break'] = new_unit | tied

    made = df['SHOT_MADE'].values
    brk = df['seq_break'].values
    streak = np.zeros(len(df), dtype=np.int8)
    run = 0
    for i in range(len(df)):
        if brk[i]:
            run = 0
        streak[i] = min(run, 127)
        run = run + 1 if made[i] else 0
    df['prev_streak'] = streak
    return df


def load_multi_season(years, data_dir, verbose=True):
    """Load, harmonise and stack several seasons."""
    frames = []
    for y in years:
        d = add_shot_order(load_season(y, data_dir))
        frames.append(d)
        if verbose:
            print(f"  {d['SEASON_2'].iloc[0]}: {len(d):,} shots", flush=True)
    out = pd.concat(frames, ignore_index=True)
    del frames
    return build_streaks(out)
```

<a id="sec19b"></a>
<div style="background:linear-gradient(135deg,#0d3320 0%,#14532d 55%,#2f7a4d 100%);padding:26px 32px;border-radius:14px;margin:34px 0 20px 0;box-shadow:0 6px 24px rgba(13,51,32,0.25);">
<div style="color:rgba(255,255,255,0.55);font-size:11px;letter-spacing:4px;text-transform:uppercase;margin-bottom:8px;">Section 19b</div>
<h2 style="color:#ffffff;font-family:Georgia,serif;font-size:24px;margin:0;font-weight:700;">Extension to Twenty-Two Seasons</h2>
<p style="color:#d9cdb4;margin:6px 0 0 0;font-size:13.5px;">An independent data source, real court geometry, and whether the answer holds across eras</p>
</div>


Everything to this point rests on one season of tracking data, which raises two questions a single
source cannot answer. Would an independent measurement of the same season agree? And is whatever the
season shows a property of basketball or a property of 2014-15?

A second archive answers both. It carries every regular-season field-goal attempt from 2003-04 through
2024-25, drawn from NBA.com and distributed as season files, and its variables overlap only partly with
the tracking log. The two are complementary rather than interchangeable, and the difference decides which
questions each can address.


```python
source_comparison = pd.DataFrame([
    ('Seasons covered', '2014-15 only', '2003-04 to 2024-25'),
    ('Field-goal attempts', '128,069', '4,450,789'),
    ('Shot outcome', 'yes', 'yes'),
    ('Distance from basket', 'yes', 'yes'),
    ('Court coordinates (x, y)', 'no', 'yes'),
    ('NBA zone labels', 'no', 'yes'),
    ('Action type (dunk, floater, ...)', 'no', 'yes'),
    ('Distance to closest defender', 'yes', 'no'),
    ('Touch time before the shot', 'yes', 'no'),
    ('Dribbles before the shot', 'yes', 'no'),
    ('Shot clock', 'yes', 'no'),
    ('Explicit within-game shot order', 'yes', 'reconstructed from the game clock'),
], columns=['Variable', 'Tracking log (Sections 5-18)', 'Shot archive (this section)'])
source_comparison
```

<div style="background:#f4efe4;border-radius:12px;padding:20px 26px;border-left:5px solid #b4531f;font-family:Georgia,serif;line-height:1.85;color:#1a2e22;margin:14px 0;">
<h4 style="color:#b4531f;margin:0 0 10px;">Why both are kept rather than one replacing the other</h4>

The shot archive is thirty-five times larger and carries geometry the tracking log lacks, which makes it
tempting to treat as an upgrade. It is not, for one decisive reason: it has no defender distance, no
touch time and no dribbles. Those three variables are what Section 13's shot-difficulty adjustment is
built from, and that adjustment is where two thirds of the apparent streak effect went. Without them the
central result of this notebook could not have been obtained at all.
<br><br>
So the division of labour follows the data. The tracking log keeps the replication, the bias work and
the difficulty adjustment, because only it can measure how contested a shot was. The archive takes on
what a single season cannot support: an independent check on that season, court geometry, and whether
the answer is stable across twenty-two years of a changing game.

</div>

<div style="background:linear-gradient(135deg,#2f7a4d 0%,#14532d 100%);border-radius:8px;padding:11px 20px;margin:24px 0 14px 0;">
<span style="color:#d9cdb4;font-family:Georgia,serif;font-size:12px;font-weight:700;letter-spacing:1px;">19b.1</span>
<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">Reconstructing shot order without a shot counter</span>
</div>


The archive records the period and the game clock but no shot number, so the sequence the whole analysis
depends on has to be rebuilt. Seconds elapsed since tip-off gives the ordering, treating each regulation
period as twelve minutes and each overtime as five. Where two shots by the same player in the same game
carry an identical timestamp their order is genuinely unknown, and those shots are flagged and excluded
from streak-conditional analysis rather than ordered arbitrarily, exactly as the gap rule of Section 6.4
handles missing shots in the tracking log.


```python
from multi_season import load_season, add_shot_order, build_streaks

SEASON_DIR = '/kaggle/input/nba-shots-2004-2025'

archive_2015 = build_streaks(add_shot_order(load_season(2015, SEASON_DIR)))
archive_analysis = archive_2015[~archive_2015['seq_break']]

tied = int(archive_2015['seq_break'].sum() -
           archive_2015.groupby(['PLAYER_ID', 'GAME_ID']).ngroups)
print(f"2014-15 in the archive: {len(archive_2015):,} shots, "
      f"{archive_2015['PLAYER_ID'].nunique()} players, {archive_2015['GAME_ID'].nunique()} games")
print(f"  shots with no observed predecessor (first of a game, or a timestamp tie): "
      f"{archive_2015['seq_break'].sum():,}")
print(f"  of which unorderable ties: {tied:,} ({100*tied/len(archive_2015):.2f}% of shots)")
print(f"  analysable: {len(archive_analysis):,}")
```

<div style="background:linear-gradient(135deg,#2f7a4d 0%,#14532d 100%);border-radius:8px;padding:11px 20px;margin:24px 0 14px 0;">
<span style="color:#d9cdb4;font-family:Georgia,serif;font-size:12px;font-weight:700;letter-spacing:1px;">19b.2</span>
<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">Do two independent sources agree about the same season?</span>
</div>

```python
rows = []
for k in range(5):
    arch_mask = (archive_analysis['prev_streak'] == k if k < 4
                 else archive_analysis['prev_streak'] >= 4)
    trk_mask = (analysis['prev_streak'] == k if k < 4 else analysis['prev_streak'] >= 4)
    rows.append({
        'K': f'{k}+' if k == 4 else str(k),
        'archive FG%': archive_analysis.loc[arch_mask, 'SHOT_MADE'].mean(),
        'archive n': int(arch_mask.sum()),
        'tracking FG%': analysis.loc[trk_mask, 'FGM'].mean(),
        'tracking n': int(trk_mask.sum()),
    })
cross_source = pd.DataFrame(rows)
cross_source['difference'] = cross_source['tracking FG%'] - cross_source['archive FG%']

coverage = len(analysis) / len(archive_analysis)
print(f"The tracking log covers {100*coverage:.1f}% of the season's analysable shots")
print(f"  players: {analysis['player_id'].nunique()} of {archive_2015['PLAYER_ID'].nunique()}")
print(f"  games  : {analysis['GAME_ID'].nunique()} of {archive_2015['GAME_ID'].nunique()}")
cross_source.round(4)
```

<div style="background:#f3f8f4;border-radius:12px;padding:20px 26px;border-left:5px solid #1c6b3f;font-family:Georgia,serif;line-height:1.85;color:#1a2e22;margin:14px 0;">
<h4 style="color:#14532d;margin:0 0 10px;">The sources agree on the pattern, and the coverage gap is worth knowing about</h4>

Both datasets show field-goal percentage falling as the preceding run of makes lengthens, and both put
the drop between the no-streak and four-make categories at one to three percentage points. The tracking
log's decline is the steeper of the two. That an entirely separate measurement of the same games
reproduces the qualitative pattern is real external validation, and it was not guaranteed.
<br><br>
The coverage comparison is the more useful discovery. The tracking log holds about 62% of the season's
attempts, 281 of 490 players and 904 of 1,230 games, because optical tracking was not capturing every
venue and every player in 2014-15. Every result in Sections 5 through 18 therefore rests on a substantial
and possibly non-random subset rather than the full season, which was not visible from inside that
dataset. The archive shows the subset behaves like the whole on this dimension, which is reassuring, but
the limitation belongs in Section 21 regardless.

</div>

<div style="background:linear-gradient(135deg,#2f7a4d 0%,#14532d 100%);border-radius:8px;padding:11px 20px;margin:24px 0 14px 0;">
<span style="color:#d9cdb4;font-family:Georgia,serif;font-size:12px;font-weight:700;letter-spacing:1px;">19b.3</span>
<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">Where shots are actually taken from</span>
</div>


The archive records the x and y position of every attempt, so Section 7.4's distance-band approximation
can be replaced with the real thing. The two panels below map shot density by streak state on a common
grid, and the third shows the difference between them: where a player who has just made three shots is
more likely to be shooting from, and where they are less.


```python
archive_recent = build_streaks(add_shot_order(load_season(2024, SEASON_DIR)))
recent = archive_recent[~archive_recent['seq_break']]
streak_recent = np.minimum(recent['prev_streak'].values, 3)

X_EDGES = np.linspace(-25, 25, 41)
Y_EDGES = np.linspace(0, 40, 33)

def density_grid(sub):
    """Share of that state's attempts falling in each court cell."""
    H, _, _ = np.histogram2d(sub['LOC_X'].values, sub['LOC_Y'].values,
                             bins=[X_EDGES, Y_EDGES])
    return H / H.sum()

def draw_court(ax):
    from matplotlib.patches import Circle, Rectangle, Arc
    ax.add_patch(Circle((0, 0), 0.75, fill=False, color=INK, lw=1.4))
    ax.add_patch(Rectangle((-3, -1.1), 6, 0.2, color=INK))
    ax.add_patch(Rectangle((-8, 0), 16, 19, fill=False, color=INK, lw=1.2))
    ax.add_patch(Arc((0, 19), 12, 12, theta1=0, theta2=180, color=INK, lw=1.2))
    ax.add_patch(Rectangle((-22, 0), 0.12, 14, color=INK))
    ax.add_patch(Rectangle((22, 0), 0.12, 14, color=INK))
    ax.add_patch(Arc((0, 0), 47.5, 47.5, theta1=22, theta2=158, color=INK, lw=1.3))
    ax.set_xlim(-25, 25); ax.set_ylim(-2, 40); ax.set_aspect('equal'); ax.axis('off')

cold = density_grid(recent[streak_recent == 0])
hot = density_grid(recent[streak_recent == 3])

fig, axes = plt.subplots(1, 3, figsize=(17.5, 6.6))
for ax, grid, title in [(axes[0], cold, 'No preceding make (K = 0)'),
                        (axes[1], hot, 'After three makes (K = 3)')]:
    ax.pcolormesh(X_EDGES, Y_EDGES, grid.T, cmap=CMAP_GREEN,
                  vmin=0, vmax=max(cold.max(), hot.max()), shading='flat', zorder=0)
    draw_court(ax)
    ax.set_title(title, fontsize=12, color=GREEN_DEEP, fontweight='bold')

ax = axes[2]
diff = (hot - cold) * 1000
lim = np.abs(diff).max()
mesh = ax.pcolormesh(X_EDGES, Y_EDGES, diff.T, cmap=CMAP_DIV,
                     vmin=-lim, vmax=lim, shading='flat', zorder=0)
draw_court(ax)
ax.set_title('Difference: where a hot shooter shoots more\n(green) and less (copper)',
             fontsize=12, color=GREEN_DEEP, fontweight='bold')
plt.colorbar(mesh, ax=ax, shrink=0.55, label='change in share per 1,000 shots')

fig.suptitle('Figure 24. True shot charts by streak state, 2023-24 season',
             fontsize=14, color=GREEN_DEEP, fontweight='bold', y=0.99)
plt.show()

for state in (0, 3):
    s = recent[streak_recent == state]
    beyond = (s['SHOT_TYPE'].astype(str).str.contains('3PT')).mean()
    print(f"K = {state}: {len(s):,} shots | mean distance {s['SHOT_DISTANCE'].mean():.2f} ft "
          f"| {100*beyond:.1f}% threes | FG% {s['SHOT_MADE'].mean():.4f}")
```

<div style="background:linear-gradient(135deg,#f3f8f4,#e6f0e9);border:1px solid #7aa98c55;border-left:5px solid #1c6b3f;border-radius:0 12px 12px 0;padding:18px 24px;margin:14px 0;">
<div style="font-weight:700;color:#14532d;margin-bottom:10px;font-size:14px;">What the difference map shows, and why it is read cautiously</div>
<div style="color:#1a2e22;font-size:13.5px;line-height:1.75;">
The difference panel is the first genuinely spatial view of shot selection this notebook has had. It is
also the noisiest figure here: differencing two density estimates amplifies sampling variation in every
cell, so no significance attaches to any individual patch of court, and the printed summary underneath is
the part to rely on.
<br><br>
That summary is worth reading carefully, because it does not simply repeat the 2014-15 result. In
2023-24 the average shot after three makes is about a quarter of a foot longer, a smaller gap than the
2014-15 tracking data showed, and the share of attempts from beyond the arc is essentially unchanged at
38%. Field-goal percentage after three makes is actually slightly <em>higher</em> than after none, 48.3%
against 47.8%, which is the opposite of the 2014-15 direction.
<br><br>
None of that overturns anything, because this is a single season compared without adjustment, and the
pooled twenty-two-season estimate below is what settles the question. But it is a useful corrective to
reading the 2014-15 shot-selection result as a fixed law of basketball. In a league where nearly forty
percent of shots already come from beyond the arc, there is less room to reach further out, and the
behavioural shift that was clear in 2014-15 is muted here.
</div>
</div>

<div style="background:linear-gradient(135deg,#2f7a4d 0%,#14532d 100%);border-radius:8px;padding:11px 20px;margin:24px 0 14px 0;">
<span style="color:#d9cdb4;font-family:Georgia,serif;font-size:12px;font-weight:700;letter-spacing:1px;">19b.4</span>
<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">Is the effect stable across seasons?</span>
</div>


The game changed enormously over the archive's span. The share of attempts from beyond the arc rose from
18.8% in 2003-04 to 42.1% in 2024-25 and average shot distance moved from 11.6 to 14.0 feet. If the
streak coefficient is a stable property of shooting it should survive that; if it is a feature of a
particular tactical era it should not.

Each season is fitted separately with the same specification, a logistic model with player-specific
intercepts, NBA zone controls, and the preceding-make count, restricted to players with at least 100
attempts that season. Each also gets its own simulated null, since Section 12.1 established that the
coefficient is biased under independence and the size of that bias depends on the data.


```python
from scipy.linalg import qr as _qr

def independent_columns(X, tol=1e-9):
    """Keep a maximal linearly independent subset of columns.

    Zone labels and zone ranges overlap: "Restricted Area" already implies the
    shortest range band, so a variance filter leaves exact collinearity behind and
    the normal equations become singular.  A pivoted QR identifies which columns
    genuinely add rank.
    """
    X = X[:, X.std(0) > 1e-12]
    if X.shape[1] == 0:
        return X
    _, r, piv = _qr(X, mode='economic', pivoting=True)
    rank = int(np.sum(np.abs(np.diag(r)) > tol * max(X.shape) * abs(r[0, 0])))
    return X[:, np.sort(piv[:rank])]


def sparse_logit_stable(y, X_dense, codes, n_groups, max_iter=40, tol=1e-9):
    """Sparse IRLS with a small ridge on the diagonal.

    Across 22 seasons some players are perfectly separated by the zone controls,
    which sends the IRLS weights to zero and makes the normal equations singular.
    A 1e-6 ridge removes that failure mode without materially moving any estimate.
    """
    D = sp.csr_matrix((np.ones(len(y)), (np.arange(len(y)), codes)), shape=(len(y), n_groups))
    X = sp.hstack([sp.csr_matrix(X_dense), D], format='csr')
    beta = np.zeros(X.shape[1])
    for _ in range(max_iter):
        eta = X @ beta
        mu = 1 / (1 + np.exp(-eta))
        W = np.clip(mu * (1 - mu), 1e-9, None)
        XtW = X.T.multiply(W)
        A = (XtW @ X).toarray()
        A[np.diag_indices_from(A)] += 1e-6
        new = np.linalg.solve(A, XtW @ (eta + (y - mu) / W))
        if np.max(np.abs(new - beta)) < tol:
            beta = new
            break
        beta = new
    eta = X @ beta
    mu = 1 / (1 + np.exp(-eta))
    W = np.clip(mu * (1 - mu), 1e-9, None)
    A = (X.T.multiply(W) @ X).toarray()
    A[np.diag_indices_from(A)] += 1e-6
    se = float(np.sqrt(np.linalg.inv(A)[0, 0]))
    return float(beta[0]), se


def rebuild_streaks(outcomes, breaks):
    """Preceding-make count, capped at three, after resampling outcomes."""
    st = np.zeros(len(outcomes), dtype=np.int8)
    run = 0
    for i in range(len(outcomes)):
        if breaks[i]:
            run = 0
        st[i] = min(run, 3)
        run = run + 1 if outcomes[i] else 0
    return st.astype(float)
```

```python
from scipy.stats import chi2

def season_streak_estimate(year, n_null=25, min_shots=100, seed_base=1000):
    """One season: the streak coefficient, and the value it takes under independence."""
    d = build_streaks(add_shot_order(load_season(year, SEASON_DIR)))
    counts = d['PLAYER_ID'].value_counts()
    d = d[d['PLAYER_ID'].isin(counts[counts >= min_shots].index)].reset_index(drop=True)
    brk = d['seq_break'].values
    keep = ~brk
    codes = pd.factorize(d['PLAYER_ID'])[0]
    Z = np.column_stack([
        pd.get_dummies(d['BASIC_ZONE'], drop_first=True, dtype=float).values,
        pd.get_dummies(d['ZONE_RANGE'], drop_first=True, dtype=float).values])
    Z = independent_columns(Z)

    y = d['SHOT_MADE'].values.astype(np.int8)
    K = np.minimum(d['prev_streak'].values, 3).astype(float)
    beta, se = sparse_logit_stable(y[keep], np.column_stack([K[keep], Z[keep]]),
                                   codes[keep], codes.max() + 1)

    rate = d.groupby('PLAYER_ID')['SHOT_MADE'].transform('mean').values.astype(float)
    rng = np.random.default_rng(seed_base + year)
    nulls = []
    for _ in range(n_null):
        ys = rng.random(len(d)) < rate
        Ks = rebuild_streaks(ys, brk)
        b, _ = sparse_logit_stable(ys[keep].astype(np.int8),
                                   np.column_stack([Ks[keep], Z[keep]]),
                                   codes[keep], codes.max() + 1)
        nulls.append(b)
    nulls = np.array(nulls)
    return {'season': str(d['SEASON_2'].iloc[0]), 'shots': len(d),
            'players': int(d['PLAYER_ID'].nunique()),
            'fg': float(d['SHOT_MADE'].mean()),
            'three_share': float(d['SHOT_TYPE'].astype(str).str.contains('3PT').mean()),
            'beta': beta, 'se': se, 'null_mean': float(nulls.mean()),
            'corrected': beta - float(nulls.mean())}


season_rows = [season_streak_estimate(y) for y in range(2004, 2026)]
season_table = pd.DataFrame(season_rows)
season_table.round(5)
```

```python
c = season_table['corrected'].values
w = 1.0 / season_table['se'].values ** 2
pooled = float((w * c).sum() / w.sum())
pooled_se = float(np.sqrt(1.0 / w.sum()))
Q = float((w * (c - pooled) ** 2).sum())
df_q = len(c) - 1
p_Q = float(1 - chi2.cdf(Q, df_q))
I2 = max(0.0, (Q - df_q) / Q) * 100

print(f"Seasons analysed                         : {len(c)}")
print(f"Seasons with a negative corrected effect : {(c < 0).sum()} of {len(c)}")
print(f"Fixed-effect pooled estimate             : {pooled:+.5f}  (se {pooled_se:.5f})")
print(f"  95% CI                                 : [{pooled-1.96*pooled_se:+.5f}, "
      f"{pooled+1.96*pooled_se:+.5f}]")
print(f"\nCochran's Q                              : {Q:.2f} on {df_q} df, p = {p_Q:.4f}")
print(f"I-squared (variation beyond sampling noise): {I2:.1f}%")
print(f"\nCorrelation, three-point share vs effect  : "
      f"{np.corrcoef(season_table['three_share'], c)[0,1]:+.3f}")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 5.6),
                          gridspec_kw={'width_ratios': [1.55, 1]})

ax = axes[0]
yrs = np.arange(len(season_table))
ax.axhspan(pooled - 1.96 * pooled_se, pooled + 1.96 * pooled_se,
           color=SAND, alpha=0.6, label='95% CI of the pooled estimate')
ax.axhline(pooled, color=COPPER, lw=2, label=f'pooled {pooled:+.4f}')
ax.axhline(0, color=INK, lw=1.3, ls=':')
ax.errorbar(yrs, c, yerr=1.96 * season_table['se'], fmt='o', markersize=6,
            color=GREEN_DEEP, capsize=3, lw=1.5, label='season estimate')
ax.set_xticks(yrs)
ax.set_xticklabels(season_table['season'], rotation=70, fontsize=7.5)
ax.set_ylabel(r'Bias-corrected streak coefficient $\beta$')
ax.set_title(f"(a) Twenty-two seasons  (Q = {Q:.1f}, p = {p_Q:.2f}, "
             f"I\u00b2 = {I2:.0f}%)")
ax.legend(frameon=False, fontsize=9)

ax = axes[1]
sc = ax.scatter(season_table['three_share'], c, s=70,
                c=np.arange(len(c)), cmap=CMAP_GREEN, edgecolor=INK, linewidth=0.7)
ax.axhline(pooled, color=COPPER, ls='--', lw=1.6)
ax.set_xlabel('Share of attempts from beyond the arc')
ax.set_ylabel(r'Bias-corrected $\beta$')
ax.set_title('(b) The game changed; the effect did not')
cb = plt.colorbar(sc, ax=ax, shrink=0.8)
cb.set_label('season (earliest to latest)')

fig.suptitle('Figure 25. The streak effect across twenty-two NBA seasons',
             fontsize=14, color=GREEN_DEEP, fontweight='bold', y=1.02)
plt.tight_layout(); plt.show()
```

<div style="background:#f3f8f4;border-radius:12px;padding:20px 26px;border-left:5px solid #1c6b3f;font-family:Georgia,serif;line-height:1.85;color:#1a2e22;margin:14px 0;">
<h4 style="color:#14532d;margin:0 0 10px;">Result: small, negative, and the same in every era</h4>

Nineteen of twenty-two seasons return a negative corrected coefficient, and the fixed-effect pooled
estimate across 4.45 million shots is about -0.005 with a tight confidence interval. That is smaller than
the tracking-data estimate of -0.011 to -0.014, which is expected: the archive has no defender distance,
so its zone controls absorb less of the shot-selection effect than Section 13's controls do, and part of
the coefficient is being held by the controls rather than left in the streak term.
<br><br>
The heterogeneity test is the interesting part. Cochran's Q comes in at 22.1 on 21 degrees of freedom
with p = 0.39, and I-squared is 5%. Essentially all of the season-to-season spread in the estimates is
sampling noise, with no detectable real variation between seasons. Panel (b) makes the same point
differently: the share of attempts from beyond the arc more than doubled over this period, and its
correlation with the estimated effect is +0.02, which is nothing.
<br><br>
This answers a question the single-season analysis could not. Whatever produces the small negative
association is not a feature of 2014-15, of the pre-three-point-revolution game, or of any particular
tactical era. It has been the same size for twenty-two years, through a period in which almost everything
else about NBA shooting changed.

</div>

<div style="background:#fdf5ee;border-radius:14px;padding:22px 28px;border:1.5px solid #d9884f;font-family:Georgia,serif;line-height:1.8;color:#6b3410;margin:14px 0;">
<h4 style="color:#b4531f;margin:0 0 10px;font-size:1.0rem;">What the multi-season result does not settle</h4>

Stability across seasons is not evidence that the effect is causal, and the same confounding argument
from Section 20 applies to all twenty-two years equally. A mechanism that operates consistently, such as
players reaching for slightly harder shots after makes, would produce exactly this pattern: a small
negative coefficient, stable across eras, shrinking when better difficulty controls are available. That
reading is more consistent with the evidence than a stable psychological cold hand would be.
<br><br>
The archive's controls are also weaker than the tracking log's by construction. Zone labels capture where
a shot came from but not how contested it was, so the season-by-season estimates should be read as an
upper bound on the magnitude rather than a refinement of Section 13's number. The value here is in the
stability, not in the point estimate.

</div>

<div style="background:linear-gradient(135deg,#2f7a4d 0%,#14532d 100%);border-radius:8px;padding:11px 20px;margin:24px 0 14px 0;">
<span style="color:#d9cdb4;font-family:Georgia,serif;font-size:12px;font-weight:700;letter-spacing:1px;">19b.5</span>
<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">Where on the court, and what kind of shot</span>
</div>


The archive carries two variables the tracking log does not: the NBA's own court-zone labels, and an
action type naming what kind of attempt each shot was, from a cutting layup to a step-back three. Neither
is available for the 2014-15 analysis, so this subsection asks a question the earlier sections could not.

Everything so far has treated the streak effect as one number. If shooting after a run of makes really
behaves differently depending on where the shot comes from, a single pooled coefficient would be
averaging over something structured. Four recent seasons are pooled here for power.


```python
ZONE_SEASONS = [2022, 2023, 2024, 2025]
zone_frames = [build_streaks(add_shot_order(load_season(y, SEASON_DIR))) for y in ZONE_SEASONS]
zone_df = pd.concat(zone_frames, ignore_index=True)
del zone_frames
zone_a = zone_df[~zone_df['seq_break']].copy()
zone_a['K'] = np.minimum(zone_a['prev_streak'], 3).astype(float)

print(f"Seasons pooled: {', '.join(sorted(zone_df['SEASON_2'].astype(str).unique()))}")
print(f"Shots: {len(zone_df):,}  |  analysable: {len(zone_a):,}  "
      f"|  players: {zone_df['PLAYER_ID'].nunique()}")

zone_rows = []
for zone, g in zone_a.groupby('BASIC_ZONE', observed=True):
    if len(g) < 20_000:
        continue
    at_zero = g.loc[g['K'] == 0, 'SHOT_MADE'].mean()
    at_three = g.loc[g['K'] == 3, 'SHOT_MADE'].mean()
    share0 = (zone_a.loc[zone_a['K'] == 0, 'BASIC_ZONE'] == zone).mean()
    share3 = (zone_a.loc[zone_a['K'] == 3, 'BASIC_ZONE'] == zone).mean()
    zone_rows.append({'Zone': zone, 'shots': len(g),
                      'FG% at K=0': at_zero, 'FG% at K=3': at_three,
                      'difference': at_three - at_zero,
                      'share at K=0': share0, 'share at K=3': share3,
                      'mix shift (pp)': 100 * (share3 - share0)})
zone_table = pd.DataFrame(zone_rows).sort_values('difference')
zone_table.round(4)
```

<div style="background:#f3f8f4;border-radius:12px;padding:20px 26px;border-left:5px solid #1c6b3f;font-family:Georgia,serif;line-height:1.85;color:#1a2e22;margin:14px 0;">
<h4 style="color:#14532d;margin:0 0 10px;">The effect changes sign depending on where the shot comes from</h4>

The three zones behind the arc all move the same way and all move down: above the break -0.013, left
corner -0.016, right corner -0.015. The three zones inside it all move the same way and all move up:
restricted area +0.031, in the paint +0.014, mid-range +0.006. Six zones, two clean groups, and the split
falls exactly on the three-point line.
<br><br>
A single pooled coefficient averages these into one small negative number, which is what Sections 13 and
19b report. That average is not wrong, but it conceals a structured pattern underneath, and the structure
is the more interesting object. Whatever is happening after a run of makes is not one effect applied
uniformly to every shot.
<br><br>
The shot mix moves too, and not in the direction a simple story would predict. Attempts from above the
break rise by two percentage points and mid-range by one, while both corner threes fall by more than a
point each and the restricted area falls by almost a point. Players are not simply shooting from further
out; they are shifting from corner threes toward above-the-break threes, which are different kinds of
shot despite similar distances.

</div>

<div style="background:linear-gradient(135deg,#2f7a4d 0%,#14532d 100%);border-radius:8px;padding:11px 20px;margin:24px 0 14px 0;">
<span style="color:#d9cdb4;font-family:Georgia,serif;font-size:12px;font-weight:700;letter-spacing:1px;">19b.6</span>
<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">A hypothesis about shot creation, and what happened to it</span>
</div>


The corner-to-above-the-break shift suggests a mechanism. Corner threes are overwhelmingly catch-and-shoot
attempts created by someone else's pass, because the geometry makes them hard to generate off the dribble.
Above-the-break threes are far more often self-created pull-ups and step-backs. If a player who has just
made several shots starts hunting their own shot rather than taking what the offence gives them, the mix
would shift exactly this way, and self-created shots are harder, which would explain the negative
three-point coefficient without any appeal to shooting touch.

The action-type labels let this be tested rather than asserted.


```python
action_label = zone_a['ACTION_TYPE'].astype(str)
zone_a['self_created'] = action_label.str.contains(
    'Pullup|Step Back|Driving|Turnaround|Fadeaway|Running', case=False, regex=True)

# within-player deviations, so this cannot be a composition effect
zone_a['self_created_dev'] = (zone_a['self_created'].astype(float)
                              - zone_a.groupby('PLAYER_ID')['self_created']
                                      .transform('mean').astype(float))

creation = (zone_a.groupby('K')
            .agg(self_created_share=('self_created', 'mean'),
                 within_player_deviation=('self_created_dev', 'mean'),
                 n=('SHOT_MADE', 'size'))
            .reset_index())
print("Share of attempts that are self-created, by preceding-make count:")
print(creation.round(5).to_string(index=False))

# does the three-point decline live in the self-created threes, as the hypothesis predicts?
threes = zone_a[zone_a['BASIC_ZONE'].astype(str).str.contains('3')]
split_rows = []
for label, mask in [('self-created threes', threes['self_created']),
                    ('other threes', ~threes['self_created'])]:
    at0 = threes.loc[mask & (threes['K'] == 0), 'SHOT_MADE'].mean()
    at3 = threes.loc[mask & (threes['K'] == 3), 'SHOT_MADE'].mean()
    split_rows.append({'Three-point attempts': label, 'FG% at K=0': at0,
                       'FG% at K=3': at3, 'difference': at3 - at0,
                       'n at K=3': int((mask & (threes['K'] == 3)).sum())})
three_split = pd.DataFrame(split_rows)
three_split.round(4)
```

<div style="background:#fdf5ee;border-radius:14px;padding:22px 28px;border:1.5px solid #d9884f;font-family:Georgia,serif;line-height:1.8;color:#6b3410;margin:14px 0;">
<h4 style="color:#b4531f;margin:0 0 10px;font-size:1.0rem;">The behavioural half of the hypothesis holds; the explanatory half does not</h4>

Players really do create more of their own shots after making some. Measured as within-player deviations,
so that differences between players cannot produce it, the self-created share runs -0.012 with no
preceding make and +0.018 after three, a swing of about three percentage points that increases
monotonically with the run. That is a genuine behavioural finding and it is new here, since the tracking
log carries no action labels.
<br><br>
It does not explain the three-point result. If self-creation were the mechanism, the decline should
concentrate in self-created threes and largely vanish in the rest. It does not: self-created threes fall
by 0.011 and other threes fall by 0.017, so if anything the decline is slightly larger among the threes
the hypothesis predicted would be unaffected.
<br><br>
The hypothesis is therefore reported as tested and rejected rather than quietly dropped. Something makes
three-point shooting slightly worse and inside shooting slightly better after a run of makes, and
self-creation is not it. Defensive closeout intensity, which no dataset used here measures, is the
obvious next candidate and the obvious thing this analysis cannot reach.

</div>

```python
fig, axes = plt.subplots(1, 2, figsize=(15.5, 5.6))

ax = axes[0]
zt = zone_table.copy()
order = np.arange(len(zt))
colours = [COPPER if v < 0 else GREEN_DEEP for v in zt['difference']]
ax.barh(order, zt['difference'], color=colours, edgecolor=INK)
ax.axvline(0, color=INK, lw=1.3)
ax.set_yticks(order); ax.set_yticklabels(zt['Zone'], fontsize=9.5)
for i, (v, n) in enumerate(zip(zt['difference'], zt['shots'])):
    ax.text(v + (0.0012 if v > 0 else -0.0012), i, f'{v:+.3f}',
            va='center', ha='left' if v > 0 else 'right', fontsize=9.5, color=INK)
ax.set_xlabel('Field-goal rate after three makes, minus after none')
ax.set_title('(a) Inside the arc it rises, outside it falls')

ax = axes[1]
ax.barh(order, zt['mix shift (pp)'],
        color=[GREEN_DEEP if v > 0 else COPPER for v in zt['mix shift (pp)']],
        edgecolor=INK)
ax.axvline(0, color=INK, lw=1.3)
ax.set_yticks(order); ax.set_yticklabels(zt['Zone'], fontsize=9.5)
ax.set_xlabel('Change in share of attempts (percentage points)')
ax.set_title('(b) Corner threes give way to above-the-break threes')

fig.suptitle('Figure 26. Zone-specific shooting and shot mix after a run of makes, 2021-22 to 2024-25',
             fontsize=13.5, color=GREEN_DEEP, fontweight='bold', y=1.02)
plt.tight_layout(); plt.show()
```

<h2 style="color:#ffffff;font-family:Georgia,serif;font-size:24px;margin:0;font-weight:700;">Causal Reasoning</h2>
<p style="color:#d9cdb4;margin:6px 0 0 0;font-size:13.5px;">What the design could have identified, and what it could not</p>
</div>


No causal claim has been made anywhere above, and this section explains why that restraint is forced
rather than cautious. The useful device is a diagram of what generates what, because it makes visible
which arrows the data can speak to.


```python
fig, ax = plt.subplots(figsize=(11, 6.5))
ax.set_xlim(0, 10); ax.set_ylim(0, 7); ax.axis('off')

nodes = {
    'Player\nability': (1.3, 5.6), 'Previous\nmakes': (4.3, 5.6),
    'Shot\nselection': (4.3, 3.0), 'This shot\ngoes in': (8.2, 4.3),
    'Defensive\nresponse': (1.3, 1.2), 'Unobserved\nform / fatigue': (7.0, 1.0),
}
observed = {'Player\nability', 'Previous\nmakes', 'Shot\nselection', 'This shot\ngoes in'}
for name, (x, y) in nodes.items():
    seen = name in observed
    box = plt.matplotlib.patches.FancyBboxPatch(
        (x - 0.85, y - 0.48), 1.7, 0.96, boxstyle='round,pad=0.45',
        facecolor='#e6f0e9' if seen else '#fdf5ee',
        edgecolor=GREEN_DEEP if seen else COPPER, linewidth=1.8)
    ax.add_patch(box)
    ax.text(x, y, name, ha='center', va='center', fontsize=10,
            color=INK, fontweight='bold')

def arrow(a, b, dashed=False, colour=None):
    ax.annotate('', xy=nodes[b], xytext=nodes[a],
                arrowprops=dict(arrowstyle='-|>', lw=1.8,
                                color=colour or (COPPER if dashed else GREEN_DEEP),
                                linestyle='dashed' if dashed else 'solid',
                                shrinkA=38, shrinkB=38))

arrow('Player\nability', 'Previous\nmakes')
arrow('Player\nability', 'This shot\ngoes in')
arrow('Previous\nmakes', 'Shot\nselection')
arrow('Shot\nselection', 'This shot\ngoes in')
arrow('Previous\nmakes', 'This shot\ngoes in')
arrow('Defensive\nresponse', 'Shot\nselection', dashed=True)
arrow('Unobserved\nform / fatigue', 'Previous\nmakes', dashed=True)
arrow('Unobserved\nform / fatigue', 'This shot\ngoes in', dashed=True)

ax.text(5, 6.6, 'Figure 27. What generates a made shot',
        ha='center', fontsize=13.5, color=GREEN_DEEP, fontweight='bold')
ax.text(5, 0.05, 'Green: measured in this dataset.   Copper dashed: not measured.',
        ha='center', fontsize=9.5, color=GREY, style='italic')
plt.tight_layout(); plt.show()
```


Three features of the diagram determine what this analysis could establish.

**Player ability is a confounder, and it is handled.** Better shooters make more shots and therefore
accumulate more streaks, so without adjustment the streak variable would partly be measuring who is
shooting. Player-specific intercepts block that path, which is why every model from Section 12 onward
carries them.

**Shot selection sits on the causal path, which makes adjusting for it a choice with consequences.**
Previous makes change what shot gets taken (Section 13 measured this directly), and the shot taken
changes whether it goes in. Controlling for difficulty therefore estimates the effect of a streak
*holding the shot fixed*, which answers "does a player who has made three shoot better on an otherwise
identical attempt" and not "does a player who has made three score more often". Those are different
questions with different answers, and the notebook reports the first because the second is dominated by a
behavioural response that has nothing to do with shooting skill.

**The unmeasured arrows are the ones that rule out a causal reading.** Form, fatigue, and whatever else
varies within a game are unobserved here, and they plausibly affect both whether the previous shots went
in and whether this one does. That is a classic confounding structure with no instrument, no
discontinuity, and no assignment mechanism available. Nothing in a single season of observational shot
logs identifies it.


<div style="background:#fdf5ee;border-radius:14px;padding:22px 28px;border:1.5px solid #d9884f;font-family:Georgia,serif;line-height:1.8;color:#6b3410;margin:14px 0;">
<h4 style="color:#b4531f;margin:0 0 10px;font-size:1.0rem;">What the null result does and does not mean causally</h4>

Because the confounding cannot be ruled out, the adjusted estimate near zero should be read as an
association, not as evidence that streaks have no causal effect on shooting. Two distinct situations
would produce the same number: there is genuinely nothing there, or there is a real positive effect
offset by an unmeasured negative one, such as fatigue accumulating over the same possessions in which
makes accumulate.
<br><br>
Section 18 adds a second interpretive limit from the other direction. Even a clearly positive coefficient
would not have identified the mechanism, because a latent-state process with no streak mechanism at all
produced a significant streak coefficient in 97.5% of simulated replications. Between the unmeasured
confounders and the mechanism ambiguity, the honest scope of this notebook is descriptive: it establishes
what the conditional associations are, how much of them survive correction, and how much predictive value
they carry. It does not establish why.

</div>

<h2 style="color:#ffffff;font-family:Georgia,serif;font-size:24px;margin:0;font-weight:700;">Limitations and Ethics</h2>
<p style="color:#d9cdb4;margin:6px 0 0 0;font-size:13.5px;">Collected in one place rather than scattered as hedges</p>
</div>

<div style="background:#f7f6f2;border-radius:12px;padding:20px 26px;border-left:5px solid #8a8f8a;font-family:Georgia,serif;line-height:1.85;color:#1a2e22;margin:14px 0;">
<h4 style="color:#14532d;margin:0 0 10px;">Design and data</h4>

One season of one league, field goals only. Free throws are excluded by the dataset, and they are the
setting where several studies have found their clearest signals, so this is a harder test than some of
the literature ran and the results are not directly comparable to free-throw work. The shot logs come
from optical tracking and carry the measurement problems Section 6 documented: 312 impossible touch
times, 1,411 shot-value inconsistencies, and missingness concentrated at period ends. Defensive scheme,
fatigue, lineup, and score pressure beyond the raw clock are all unmeasured.

</div>

<div style="background:#f7f6f2;border-radius:12px;padding:20px 26px;border-left:5px solid #8a8f8a;font-family:Georgia,serif;line-height:1.85;color:#1a2e22;margin:14px 0;">
<h4 style="color:#14532d;margin:0 0 10px;">Statistical scope</h4>

The primary analysis was specified before modelling, but the notebook as a whole was built iteratively on
the same data it analyses, so the secondary results in Sections 14 to 16 carry more researcher degrees of
freedom than the primary one. Section 14 is explicit about the case where that mattered. Power was
established in Section 18 at effect sizes around 0.10 in log-odds; the adjusted estimate is roughly
-0.011, an order of magnitude smaller, and no claim is made that these methods could detect an effect
that small. The simulated nulls use 150 to 200 draws, enough for a stable normal approximation but not
for precise tail probabilities, which is why Section 13 reports both an empirical and an approximated
p-value.

</div>


**Ethics and data provenance.** The data are publicly released professional-sport statistics describing
the on-court actions of athletes performing in public. No private information is involved and no consent
question arises in the way it would for human-subjects data. Two restraints still apply. Players are
named in the dataset, and the notebook does not rank them by apparent streakiness or single anyone out as
a streak shooter, because Section 11 showed those individual estimates are dominated by noise and naming
people on the strength of them would be misleading about real identifiable individuals. And no
psychological interpretation is placed on any shooting sequence: a latent state in Section 16 is a
statistical construct, and describing a player as confident or rattled on the basis of four made shots
would be an inference the data cannot support.


<a id="sec22"></a>
<div style="background:linear-gradient(135deg,#0d3320 0%,#14532d 55%,#2f7a4d 100%);padding:26px 32px;border-radius:14px;margin:34px 0 20px 0;box-shadow:0 6px 24px rgba(13,51,32,0.25);">
<div style="color:rgba(255,255,255,0.55);font-size:11px;letter-spacing:4px;text-transform:uppercase;margin-bottom:8px;">Section 22</div>
<h2 style="color:#ffffff;font-family:Georgia,serif;font-size:24px;margin:0;font-weight:700;">Conclusion</h2>
<p style="color:#d9cdb4;margin:6px 0 0 0;font-size:13.5px;">What the analysis supports, suggests, leaves open, and cannot address</p>
</div>

<div style="background:linear-gradient(135deg,#f4efe4 0%,#ece5d6 50%,#f4efe4 100%);border-radius:16px;padding:34px 42px;font-family:Georgia,serif;line-height:1.85;color:#1a2e22;border:1.5px solid #d9cdb4;margin:20px 0;">
<h2 style="color:#14532d;font-size:1.4rem;margin:0 0 18px;">Conclusion</h2>

<p style="margin:0 0 16px 0;"><strong>What the data support.</strong> Read without correction, 2014-15 NBA
shooting looks like a cold hand: the conditional rate after three consecutive makes is 2.4 percentage
points below the unconditional rate, and a logistic model with player effects gives a streak coefficient
of -0.051 with a p-value near 1e-14. That apparent finding decomposes almost entirely into two parts.
Roughly a fifth is the streak-selection bias Miller and Sanjurjo identified, reproduced here from exact
enumeration, measured in the permutation nulls of real players at about -0.021, and shown in Section 12.1
to persist inside generalised linear models at about -0.006. Two thirds of the remainder is shot
selection: players attempt shots about 1.7 feet farther out after three makes than after a miss. What is
left after both corrections is about -0.011, with a two-sided p-value near 0.10, and it is stable at that
value across eight alternative specifications.</p>

<p style="margin:0 0 16px 0;"><strong>Where that conclusion is fragile.</strong> Section 13b found the adjusted
model does not actually fit: its binned residuals miss in ten of twenty bins, in a smile-shaped pattern
indicating a missing interaction between shot distance and defender distance. Adding that interaction
restores calibration to three bins of twenty, and moves the corrected coefficient to about -0.014 with a
two-sided p-value near 0.039. The two estimates agree on sign and are close in size, so this is not a
reversal, but significance does depend on how the difficulty controls are built, and the better-fitting
build is the one that clears the threshold. The fully adjusted effect is best described as small,
negative, and of borderline significance, tipping toward significance as the controls improve. A residual
negative of that size is also what incomplete difficulty measurement would produce, given Section 13
established that shot selection genuinely changes after makes, so better tracking data would be expected
to shrink it further rather than confirm it.</p>

<p style="margin:0 0 16px 0;"><strong>What the data suggest.</strong> Three independent lines point the same
way. A hierarchical Bayesian model puts between-player variation in the streak effect at 0.023 and finds
no player among 281 with a 95% posterior probability of a positive effect, so the scattered individual
estimates that separate tests produce are noise rather than real differences. A two-state hidden Markov
model, run far enough, collapses exactly into a single state, so a latent-state description buys nothing
either. And under chronological validation, adding streak information changes held-out ROC-AUC by less
than 0.001 against 0.095 for shot difficulty, so whatever effect might exist carries no predictive value.
Agreement across an inferential null, a latent-variable null, and a predictive null is more informative
than any of them alone.</p>

<p style="margin:0 0 16px 0;"><strong>What twenty-two seasons add.</strong> Section 19b repeats the core
estimate on an independent archive of 4.45 million attempts spanning 2003-04 to 2024-25. Nineteen of the
twenty-two seasons return a negative corrected coefficient, the pooled estimate is about -0.005, and
Cochran's Q gives no evidence of between-season heterogeneity (p = 0.39, I-squared 5%) even though the
share of attempts from beyond the arc rose from 19% to 42% across the period. The archive's estimate is
smaller than the tracking data's, as expected, since zone labels absorb less shot selection than defender
distance does. That same section also found the tracking log covers only 62% of 2014-15's attempts, which
is a coverage limitation invisible from inside that dataset.</p>

<p style="margin:0 0 16px 0;"><strong>What remains uncertain.</strong> The methods were shown in Section 18 to
have full power against effects around 0.10 in log-odds, which is the scale the literature discusses, but
the adjusted estimate is an order of magnitude smaller and nothing establishes that an effect that small
would be detectable. The shot-difficulty controls capture only observable difficulty, so a real effect
masked by unmeasured defensive adjustment cannot be excluded. And the analysis covers field goals in one
season; free throws, where the strongest published effects sit, are absent from the data entirely.</p>

<p style="margin:0 0 16px 0;"><strong>What the data cannot establish.</strong> Nothing here identifies a
causal effect. Form and fatigue plausibly influence both whether previous shots went in and whether this
one does, there is no instrument or discontinuity to exploit, and a single cross-section of observational
shot logs does not support causal identification. Section 18 adds the sharper limit: a simulated process
containing no streak mechanism at all, only a persistent hidden state, produced a significant streak
coefficient in 97.5% of replications. Even a clear positive finding would not have identified the
mechanism behind it.</p>

<p style="margin:0;"><strong>The methodological findings may outlast the empirical one.</strong> Two emerged
from checking rather than from analysis, and both run in the same direction. The conventional
inverse-gamma prior on a slope variance returns more between-player heterogeneity from data simulated
with none (0.173) than from the real data (0.136), so any hierarchical hot-hand analysis using it is
liable to report player differences that do not exist. The hidden Markov model's state separation shrinks
slowly enough that a default stopping rule halts while the two states still differ by 0.13, manufacturing
a latent-state finding that longer optimisation dissolves. Section 18 quantifies the cost of a third such
default: testing a history-derived regressor against zero rather than against its own simulated null
doubles the false-positive rate, from the nominal 5% to 10%. In every case the routine choice errs toward
finding structure that is not there, which is the more dangerous direction for a literature built on
detecting subtle effects.</p>
